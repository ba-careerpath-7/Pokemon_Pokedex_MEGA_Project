## NA and NaN values Case Study, do we delete them?

* Usually, we do NOT want to delete rows that have NA (Not Available) or NaN (Not a Number) values!!!

* Sometimes there may even be a systematic reason for NA values!

* Even if the NA values have no pattern, it is still not a good idea to simply drop them. It is not ethical to even drop one observation that contains a NA value. We should at least impute NA values with some kind of value. Furthermore, we want to use all the observations because it can be useful information!


---
### Examples

* Example 1, F1 Racing: in F1 racing there are 3 qualifying rounds. "In the first qualifying round, the slowest 5 drivers will not advance to the next round" (F1Racing). These 5 drivers will have NA values for the second qualfying round data. Are we just going to delete those 5 drivers from the data set?

NO! Those NA values are actually information! It shows those drivers did not make it to the second round!!! We could change the NA value to string value such as "did not make it to second round".

* Example 2, Airport flights: Perhaps we have NA values in a airport data set. We may observe that that the departure time are usually NA when the temperature is really cold, like -30 celcius! This means the weather caused the flight to be canceled and no departure time is recorded.  

These NA values are SYSTEMATIC! Cold weather causes flights to be canceled! And therefore, we can NOT just remove these rows. These NA values are actually informative!!!

---

### What should we impute?

1. Missing continous values:

When the NA value is a continous number, what should we impute? Perhaps we can impute the mean value of the column, although that is not always reliable since we don't know if the missing value is close to the mean.

A more reliable option would be to use a K Nearest Neighbors Imputer, where this imputer fills missing values by taking the average of the K neighbor observations that do have values!

2. Missing categorical values:

When the NA value is categorical, we can simply write "None" or "Missing" to denote that we do not know the category or if there is no reasonable category to write.

Or if there is reason too, we can use the K Nearest Neighbors classification model to fill the missing value by looking at the K number neighbor observations and use majority vote to select the most likely category.

---

## Outlier Case Study, do we delete them?

* Outlier defintion: A data point that is drastically different from majority of the data.

* Outlier example: Lets pretend that the NBA league's average point per game stat is 10 points per game. We observe a player who consistently scores 27 points per game. Are we going to simply remove data regarding that player just because he is a outlier?

NO! It is unethical to remove outliers, that player is still part of the data. There are bound to be outliers in life and it provides a interesting perspective to the data. Removing outliers will distort reality!

It would also make the player who averages 27 points per game a sad player as well!

* One more thing, we actually want outliers!
We have a machine learning model called Isolation Forests that tries to identify outliers in a data set! (For more information of Isolation Forests, check Section F, Un-Supervised Model 9)

But why would people want to find outliers?
Because interesting and abnormal observations are usually the outliers!

In finance, the GeeksForGeeks Isolation Forest article states that "Isolation Forests are used to identify fraudulent transactions by analyzing credit card transactions for anomalies such as large, unexpected purchases or rapid consecutive transactions. It helps to prevent financial fraud and protects customers from potential losses" ("What is a Isolation Forest?").

So outliers can even be a dangerous observation, and we want to detect dangerous operations so that we can deal with whatever is happening.


---

### Is there any situation where we would delete observations that have NA or are outliers?

There is NO reason to delete NA or outliers if the data set was recorded well.

The only possible exception is if there is a data entry mistake, such as a wrong recording of a observation's NA value or a outlier.

For typos or incorrect measurements, that is up to discretion. We could always ask the person who made the typo or incorrect measurement and change those values instead of deleting the observation.

Now lets try to replace our NA values, and lets use all Pokemon observations, even if they are extreme outliers such as Magikarp or Arceus.

```python
# Lets check if we have NA values:

pokemon_df.isna().sum() # this takes the sum of all NA values for each column!

# steps of the code above:
# 1: .isna() checks for NA's in each column with True and False
# 2: .sum() adds up all True's (Trues equal 1)
# and the numbers of Trues are the number of NA!
```

We have NA/NaN values, so lets replace them!

```python
#pokemon_df["gender_male_ratio"].fillna("Not Available")
#pokemon_df["Type 2"].fillna("None")
# pokemon_df["Forms"].fillna("None")
# NA ALERT: Doing this only prints the series and fills the NA values OF this CERTAIN series!
# writing this ALONE will not fix the series!!!

# CLEANING DATA BY FILLING NA's:
# To actually fix NA values in a dataframe, we either assign the fillna series with the original series:
# pokemon_df["gender_male_ratio"] = pokemon_df["gender_male_ratio"].fillna("Not Available")

# or we assign the whole data frame:
# pokemon_df = pokemon_df.fillna{"gender_male_ratio":"Not Available"}

pokemon_df["gender_male_ratio"] = pokemon_df["gender_male_ratio"].fillna("Not Available")

pokemon_df["gender_male_ratio"]
```


### Feature Engineering:

A feature is another word for a independent variable.

Feature engineering is when we need to make new feature that are inspired by features in the data set. Here is one example of feature engineering below, I wanted to make a feature that shows what region a Pokemon comes from, not just what Generation they are from.

Feature engineering will happen throughout the project. For instance, for Un-supervised learning, labeling each point with the 18 Pokemon types can be too confusing.

Therefore, making 5 groups and that each contain several Pokemon Types can be more wise. In Section F, Un-Supervised Model 1 PCA, there was a instance where I put rock,steel, and ground Pokemon in a group called Mineral Pokemon.

## So should I consider Mythical Pokemon as Legendary in this project?

There is the potential problem of predicting ,Mythical Pokemon as Legendaries in the machine learning part of this project. Many have high total base stats and some even have distinctly different abilities.  Additionally, the small number of Legendary Pokémon (only 64) could make it tempting to combine Legendary and Mythical Pokemon into a single category.


On the other hand, there is also a "Is_Ultra_Beast" variable. So I could combine
the ultra beasts, the mythicals, and legendaries as a rare pokemon group. However the term "rare" can be subjective. Many Pokemon that are rare in the wild, such as Gabite or Dratini, are NOT Mythical or Legendary.

Additionally, not all Mythical Pokemon are strong, Phione has total base stats of 480 and Meltan has total base stats of 300.

Furthermore, the Bulbapedia article states that "Mythical Pokemon are a group of Pokemon obtained by special means or events. Mythical Pokemon are related to Legendary Pokemon, but are a distinct group" (Bulbapedia).


Overall, it seems better to stick with the dataset labels and classify only the 64 Legendary Pokémon as Legendary. This approach preserves the original distinctions and avoids subjective groupings.

# 🔎 SECTION C: Exploratory Data Analysis (EDA) for Predictors!

Exploratory Data Analysis (EDA) is something any statistician, data analyst, quant analyst, or data scientist MUST do before using machine learning models!

If possible, we should look at the data, find any interesting aggregates and filters, and do basic data visualization because this counts as EDA as well. We want to get a feel for the data and know what we are getting into!

But most importantly, we can not go blindly making machine learning models right at the start! We have to use good predictor variables for our models! If we end up using garbage predictors, we will end up with garbage results. This is known as "garbage in, garbage out"!

* For supervised models, one way to do this is to make separate plots of our predictor variables against the response variables.

* For un-supervised models, we can make plots of predictor variables against other predictor variables. A fast way to plot this is by a pair plot. Another informative plot could be a correlation plot, to see how correlated the predictors are too each other.




### ALERT: We are doing EDA ON 3 sets of variables!
1. Classification
2. Regression
3. Patterns



---

### What do the words "predictors" and "response" mean?

* Predictor variables- The $x$ variables, also known as independent variables or features. The model uses this to make predictions. Predictors are usually stored in a matrix, commonly known as a feature matrix.  The feature matrix is made up of multiple columns from a data frame.

* Response variable- The $y$ variable, also known as dependent variable or target. The model tries to predict these values. The response is usually stored in a vector (1D array). The response vector is made up of a single column from a data frame. Or, if we want estimated values, then the estimated response vector is made up values predicted by the model!

* Throughout this project, I will mostly be using the words "predictor" and "response".  



---

* This is the predictor/feature matrix:


$$X_{[n \times p]} =
\begin{bmatrix}
x_{1,1} & x_{1,2} & \cdots & x_{1,p} \\
x_{2,1} & x_{2,2} & \cdots & x_{2,p} \\
\vdots  & \vdots  & \ddots & \vdots \\
x_{n,1} & x_{n,2} & \cdots & x_{n,p}
\end{bmatrix}$$

Both Supervised Learning and Un-Supervised Learning models use the predictor matrix.


* This is the response vector:

$$\vec y_{[n \times 1]} =
\begin{bmatrix}
y_1 \\
y_2 \\
\vdots \\
y_n
\end{bmatrix}$$

Only the Supervised Learning models use a response vector.

Where:

$n$ is the number of observations.
$p$ is the number of predictors.

$x_{3,4}$ is the 3rd observation's 4th predictor value.

$y_3$ is the 3rd observation's response value.

## One problem with Type 1 and Type 2 predictors!

Type 1 and Type 2 has string values like "fire" and "water". This can be a problem for classification models such as KNN or logistic regression. They do NOT accept string values!!!

We may think that one hot coding is a option. One hot coding is when for a fire type pokemon, we output 1 if fire, 0 otherwise. The problem with this approach is that there are 18 types in pokemon. So we would have 17 columns of 0s, and 1 column of 1 for a single pokemon's type. This is over kill for the predictor space!


## Target Encoding:

One good option is to assign a pokemon type a value with the probablity of being legendary.

The Geeks for Geeks website states: "In target encoding, we replace each categorical value with the mean of the response variable, and the mean is based on that category. Target encoding can be  useful for categorical variables with a big amount of unique values" (Geek 1).


For instance, lets say 20% of all legendaries are dragon. Then we would assign ALL dragon pokemon with a 0.2 probablity of being dragon.

This way, we would only have 1 column for Type 1 target encoded, and 1 column Type 2 target encoded. With one hot encoding, we will have of 18 columns of Type 1 one hot encoded + 19 columns of Type 2 one hot encoded (Type 2 also has a none value).

Lets do some feature engineering!

```python
# training_set.groupby("Type 1").mean(numeric_only= True)
# gives mean of each numeric variable, but we are only interested in Is_Legendary

# training_set.groupby("Type 1")["Is_Legendary"].mean()
# Index the "Is_Legendary" variable to get the mean of JUST Is_Legendary!

# training set: type 1 and 2
type1_means_training = training_set.groupby("Type 1")["Is_Legendary"].mean()
print(type1_means_training) # we can NOT just put this in the data frame right away!
# the training set df has 900 rows, but we only got 18 rows here!
# we need to use map()!

training_set["type1_encoded"] = training_set["Type 1"].map(type1_means_training)


type2_means_training = training_set.groupby("Type 2")["Is_Legendary"].mean()
print(type2_means_training) # we can NOT just put this in the data frame right away!
# the training set df has 900 rows, but we only got 18 rows here!
# we need to use map()!

training_set["type2_encoded"] = training_set["Type 2"].map(type2_means_training)

training_set


```

### Potential problem with target encoding

You may have noticed that we have one potential problem with target encoding. It essentially gives "hints" on if a pokemon is legendary or not based on their types. This is data leakage, when a model knows information it is not supposed to know. To put it another way, data leakage is when a model or data gives information that gives away if a pokemon is legendary or not.

Therefore we need to be CAREFUL! How can we "mitigate" during predictions?

Well we can mitigate the possible data leakage problem by giving the testing set the SAME encoded values as the training set!

Therefore, the model will NOT have hints of the testing set pokemon. It will ONLY have informtation of the training set pokemon!


```python
# testing set: type 1 and 2

testing_set["type1_encoded"] = testing_set["Type 1"].map(type1_means_training)

testing_set["type2_encoded"] = testing_set["Type 2"].map(type2_means_training)

testing_set


```

## 🔴 Supervised Classification Model 1: K Nearest Neighbors (KNN)



* This model is simple. We select a testing set point to classify. Then look at the K nearest training set points and take the majority vote of the classes.

* We usually select the nearest points by the shortest euclidian distance possible. Think of the distance formula. Here is a example with two variables: $d_E = \sqrt{(x_0 - x_1)^2 + (y_0 -y_1)^2}$


* If we have a tie of classes, R would "flip a coin" but Python's scikit-learn would default on picking the lowest class value. EX: "If we have a binary class of 0,1 then KNN would pick 0 in a tie" (Possibly Biased KNeighborsClassifier for Ties). To avoid tie breakers, it is more wise to set K to be a odd  number, but with multiple classes, we can still get potential voting ties anyways.

* For example, say we select K = 5. In this point, what the class labels of its 5 nearest points? Lets say the 4 other points are not legendary, but 5th point is a legendary. Then by majority vote of 4 to 1, we classify the selected point as a non-legendary pokemon.


* BE CAREFUL! KNN has weaknesses!!! It can't really use categorical predictors since categorical predictors dont have numbers! How do we plug a categorical predictor value into the distance formula. Furthermore, we need to standardize the predictor numbers so that not one predictor dominates the distance formula.

### What about target encoded predictors of Type 1 and Type 2?

I will NOT use the the target encoded predictors of Type 1 and Type 2 for KNN. Yes, they have numerical values, but those values mainly came from strings. These values also do not represent a distance either.

### Standardization formula:

$x_{i,standardized} = \frac{x_i - \bar x}{\sigma}$

The StandardScalar function from sklearn.preprocessing can automatically calculate the above formula for each observation.

We have two categorical variables of Type 1 and Type 2, we may have to take those off for KNN to make KNN work.



```python
X_train_knn = training_set[["Base_Stats", "capturing_rate"]]
print(X_train_knn)


X_test_knn = testing_set[["Base_Stats", "capturing_rate"]]
print(X_test_knn)

```

## Accuracy Score vs F1 Score, what score do we use? Also, can we visualize our errors?


### Accuracy Score:

Accuracy score is a good starting point to see how well we classified, BUT it does NOT tell the whole story! If for example 99% of observations are class A, then a model can just guess all observations as part A and have a accuracy score of 99%! Clearly we need another metric to use!


### F1 Score:

Since there are over 90% of non-legendaries, a model can simply predict all pokemon as a non legendary and do well. This is called a class imbalance, when one class dominates. Using a F1 score can help us examine how well a model predicts a smaller class.


Let's use the F1 score and that will be our metric on deciding the best binary classification model. If two models have a tie by having the same F1 score, perhaps we can use the accuracy score as a tie breaker.





* Confusion Matrix:

If we want to visually see what types of misclassifications there are, we use a binary confusion matrix!



## F1 Score Case Study

The GeeksForGeeks website states that "The F1 Score is a metric in machine learning that evaluates how well a classification model performs when there is a class imbalance"(GeeksForGeeks).

Here are the notations and formulas:

* TN = True Negative
* FN = False Negative
* FP = False Positive
* TP = True Positive

1. Measuring Precision

* Precision is the portion of total positive predictions that are correct.

* Another definition: Measures how accurate the positive predictions are.

* In other words, how often do we classify legendary Pokemon correctly? A high precision means that model does not make many false legendary Pokemon mistakes.

$$Precision = \frac{TP}{TP + FP}$$


2. Measuring Recall.
$$Recall = \frac{TP}{TP + FN}$$

* Recall is the portion of true positives divided by the total amount of predicted positives.

* Another definition: Measures how many of the actual positives the model correctly identified.

* In other words, does this model find most legendary Pokemon? A high recall means the model catches the most legendary Pokemon available, even if it means misclassifying some Pokemon.  

3. Measuring F1 Score

* Both precision and recall have to do well to get a high F1 score. If either precision and recall are low, this will result in a low F1 score.

* A high F1 score shows well our model is both confident and correct in classifying Legendary Pokemon.

$$F1~Score = 2* \frac{Precision * Recall}{Precision + Recall}$$




### What is considered a good F1 score?

Well a good F1 score is subjective and depends on our data set and predictors. Here are my subjective cut offs:

If $0 \% \le F1 ~ score < 30 \%$, then model is predicting terribly. Model  absoultely struggles to predict legendary Pokemon.

If $30 \% \le F1 ~ score < 40 \%$ then model is predicting weakly. Model predicts some legendary Pokemon but still may misclassify a lot of legendary Pokemon.

If $40 \% \le F1 ~ score < 60 \%$ then model is predicting okay. Model seems to capture some patterns in legendary Pokemon.

If $60 \% \le F1 ~ score < 80 \%$ then model is predicting good. Model has a good balance of precision and recall.

If $80 \% \le F1 ~ score \le 100 \%$ then model is predicting amazingly. Most legendary Pokemon are predicted correctly and there are only a few misclassifications.


```python
# function that checks the f1 score in our subjective cutoffs

def f1_check(f1_score):
  if 0 <= f1_score < 0.3:
    print("Model's F1 score is terrible...")
  elif 0.3 <= f1_score < 0.4:
    print("Model's F1 score is weak.")
  elif 0.4 <= f1_score < 0.6:
    print("Model's F1 score is okay.")
  elif 0.6 <= f1_score < 0.8:
    print("Model's F1 score is good.")
  elif 0.8 <= f1_score <= 1.0:
    print("Model's F1 score is amazing!")
  else:
    print("This may not be a F1 score, F1 score is a decimal between 0 to 1")
```

```python
from sklearn.metrics import f1_score

knn_model1_f1 = round(f1_score(y_test, y_preds),3)
print("KNN model 1's F1 score:", knn_model1_f1)

f1_check(knn_model1_f1)
```

### How to read binary confusion matrix in Scikit-Learn:

Reading a binary confusion matrix in Scikit-Learn is a bit different than  binary regular confusion matrices. A person who never used Scikit-Learn would think: "top left: true positive" and "bottom left: true negative", but it is the opposite for Scikit-Learn!

Scikit-Learn's confusion matrix website states that: "$C_{0,0}$ is true negative, $C_{0,1}$ is false positive,  $C_{1,0}$ is false negative, $C_{1,1}$ is true positive" (Pedregoso).

Here are the confusion matrix layouts:

$$
\begin{bmatrix}
C_{0,0} & C_{0,1}  \\
C_{1,0}  & C_{1,1}
\end{bmatrix} = \begin{bmatrix}
TN & FP \\
FN & TP
\end{bmatrix}
$$


1. Top Left: True Negative, TN  -> true non legendary (predicted pokemon as a non legendary and is correct)

2. Top Right: False Positive, FP -> false legendary (predicted pokemon as legendary, but it was actually a non legendary!)

3. Bottom Left: False Negative, FN -> false non legendary (predicted pokemon as non legendary, BUT it was actually a legendary!)

4. Bottom Right: True Positive, TP -> true legendary (predicted pokemon as legendary and is correct)

We want to see a lot of numbers in the left diagonal of a confusion matrix! This indicates that we are classifying many numbers correctly!



```python
from sklearn.metrics import confusion_matrix

fig = plt.figure(figsize = (12,7))

mat = confusion_matrix(y_test, y_preds)

sns.heatmap(mat, square=True, annot=True, fmt="d", cbar=False,
            xticklabels = ["not legendary", "legendary"],
            yticklabels = ["not legendary", "legendary"])

plt.title("Confusion Matrix")
plt.xlabel("true label")
plt.ylabel("predicted label");
```

### Can we find the best K value?

Using our scaled predictors, can we find the best k value? What k value will give us the best accuracy score?

```python
k_range = range(1,21,1) # start at 1, stop at 20 (since end is EXCLUSIVE), step by 1

k_scores = [] # empty list that stores our accuracy scores

f_scores = [] # empty list that stores our f1 scores

# we want to use the scaled data!!!

# recall in Python and R, the x in the FOR loop is a temp variable
for x in k_range:
  knn_model = KNeighborsClassifier(n_neighbors = x)
  knn_model.fit(X_train_knn_scaled, y_train)
  y_preds = knn_model.predict(X_test_knn_scaled)
  temp1_score = accuracy_score(y_test,y_preds)
  k_scores.append(temp1_score)

for y in k_range:
  knn_model = KNeighborsClassifier(n_neighbors = y)
  knn_model.fit(X_train_knn_scaled, y_train)
  y_preds = knn_model.predict(X_test_knn_scaled)
  temp2_score = f1_score(y_test,y_preds)
  f_scores.append(temp2_score)



# print(k_scores)

# use the numpy aggregate function np.max() to get the max value of the array
# use vectorized scalar addition of + 1,
# this goes from index 0 to 19, we want 1 to 20 though

all_k_max = np.where(k_scores == np.max(k_scores))[0] + 1
# [0] returns a 1d array

all_f_max = np.where(f_scores == np.max(f_scores))[0] + 1


print("Best possible k values for accuracy score:", all_k_max)
print("Best accuracy score from these k values: ", np.max(k_scores) )

print()

print("Best possible k values for f1 score: ", all_f_max)
print("Best f1 score from these k values: ", round(np.max(f_scores)) )




```

```python
fig = plt.figure(figsize = (12,7))


ax = plt.axes()

ax.plot(k_range, k_scores, label = "Accuracy Score")
ax.scatter(k_range, k_scores) # lets plot each point to make visualization easier

ax.plot(k_range, f_scores, label = "F1 Score")
ax.scatter(k_range,f_scores)

ax.set_xlim(left = 1, right = 20)
ax.set_ylim(bottom = -0.05, top = 1.0)

ax.set_xlabel("The K neighbors number")
ax.set_ylabel("Percentage from 0 to 1")

ax.set_title("What amount of K neighbors gives us the best scores?");

ax.legend();
```

## 🔴 Supervised Classification Model 2: Naive Bayes

Naive Bayes is a model that is inspired by the Bayes Theorem!

Bayes Theorem:

$$P(C|X) = \frac{P(X|C)*P(C)}{P(X)}$$

Let C denote the class of a pokemon. Is the pokemon in class of non legendary or being a legendary?

Let X denote the predictors of Base_Stats, capturing_rate, Type 1, Type 2!


*   P(X|C) is the likelihood. What is the probability that event X happens given event C?
*   P(C) is the prior, this is the probability that we assume before hand. Usually the probability of the occurances in our data. AKA how common is the class in our data?
*   P(X) is the normalizing constant. This probability is the least important part of the formula. In many cases, it can be thrown away since all calculations have the same value of P(X), so all P(X) values can get canceled out!
*   P(C|X) is the posterior probability, this is the final probability that resulted from the information from the likelihood and prior probablities.

### Naive Bayes Process:

All Naive Bayes does is multiply probabilities in Bayes theorem to get the a probability of class 0 - not legendary, and probability of class 1- legendary.

In this model:

* C contains the outcome class, let k be the specific class, k = [0- not legendary, 1- legendary].

* X are the independent variables, X = [Base_Stats, capturing_rate, Type 1, Type 2!].

$$P(C = k|x_1,x_2,x_3,x_4) =  \frac{P(x_1,x_2,x_3,x_4| C = k)*P(C=k)}{P(x_1,x_2,x_3,x_4)} = \frac{P(x_1| C = k) * P(x_2| C= k) * P(x_3| C = k) * P(x_4| C = k) *P(C=k)}{P(x_1,x_2,x_3,x_4)}$$

We can ignore the denominator: $P(x_1,x_2,x_3,x_4)$

---

Why can we ignore the denominator?

Because this value is the SAME for each calculation. Recall that this is the normalizing constant and we get the same proportions regardless if we use the normalizing constant or not!

Lets pretend that

$$\frac{P(X|C = 0)*P(C=0)}{P(X)} > \frac{P(X|C = 1)*P(C=1)}{P(X)}$$

We can take off $P(X)$ and the outcome will be the same, the proportions are the same even if the numbers of left hand side and right hand side do not have $P(X)$.

---


Lastly, calculate class 0 the probability of pokemon not being legendary:

$$P(y = 0|x_1,x_2,x_3,x_4)$$

Also calculate the class 1 probability of pokemon being legendary:

$$P(y = 1|x_1,x_2,x_3,x_4)$$

Do this for each observation.

* If $P(y=0|X) > P(y = 1|X)$, predict class 0: classify pokemon as non-legendary.

* If $P(y=1|X) > P(y = 0|X)$, predict class 1: classify pokemon as legendary.




One down side of the gaussian naive bayes model in Python is that it can NOT take both continous and categorical predictors. Recall that the gaussian distribution is a fancy name for the normal distribution. And also recall that the normal distribution uses continous numbers like 1.249, not categorical values like "water pokemon".

Thankfully we can avoid this issue since we did target encoding, we gave each pokemon type a numerical value. And unlike KNN, naive bayes is not about distance!









```python
from sklearn.naive_bayes import GaussianNB # 1. from module, import class or function

gnb_model = GaussianNB() # 2. intiate model

gnb_model.fit(X_train,y_train) # 3. fit the model

# priors:
print(gnb_model.class_prior_) # P(C=0) = 0.9414, P(C=1) = 0.0585

# number of observations in class 0 and class 1
print(gnb_model.class_count_)

# the means of each predictor.
print(gnb_model.theta_)
# the first row is class 0 means. the second row is class 1 means
# the first element is mean of base_stats
# second element is mean of capturing rate,
# third and fourth elements are means of type 1 encoded and type 2 encoded respectively

# the variances of each predictor
print(gnb_model.var_)
# the variance array structure follows the same array stucture as the means!

# together the means and variances define the normal distribution for each predictor!
```

The prior probablities are:

* Class of non legendary: $P(C=0) = 0.9414$

* Class of legendary: $P(C=1) = 0.0585$

The means and variances of each predictor and each class are listed above.

Our next model will be logistic regression, but before diving into logistic regression, lets learn about the bias and variance trade off theory for F1 Score.

## Bias and Variance Trade off for F1 Score:

Before we head towards more regression models, I want to mention a thoery that is important for getting the the best possible predictons.

If the model is too biased it means the model is too simple. Simple could mean too few predictors or the constraints are too heavy. (A example of heavy constraints are big lambda values for LASSO and Ridge, we will talk about that soon.) A simple model is underfitted and will get a lot of error when using a testing set.  The underfitted model's F1 score drops in both the training and testing set.

If the model has too much variance, it is a complex model. It has too much predictors or there are minimal constraints. (A example of heavy constraints are tiny lambda values for LASSO and Ridge, we will talk about that soon.)  Now, for the training set, the overfitted model F1 score increases, but this is not necesarily good.

This means the model is only accustomed to predicting the training set well, but not unknown or future data.  The overfitted model will have it's F1 score drop when using a testing set.

We are fixed to having 4 predictors, so we are not changing the number of predictors.

But the upcoming models will have parameters that can be tuned, such as the number of tree branches, the lambda values, the learning rate, and more. The less complex we make the tuning parameters, the more bias it will have. The more complex they are, the more variance it will have.

Ultimately we want the bias and variance combination that gives the  highest amount of test set F1 Score. Think of this shape $\cap$, and we want to reach the absolute top of this shape $\cap$.

## 🔴 Supervised Classification Model 3: Logistic Regression

Logistic Regression is a binary classification model that predicts two classes.

The main idea of logistic regression is this: What is the probability that class B happens?
If $P(B) > 0.5$ predict class B. If $P(B) \le 0.5$, predict class A.

### Theory of Logistic Regression:

---
**Part A: Odds**

The odds is another way to talk about probability, and this is a important concept on how logistic regression works.

The odds formula is:

$$(\frac{p}{1-p}) = e^{\beta_0 + \beta_1 x_{1i} + ... + \beta_k x_{ki}}$$

This has a exponential shape, due to the exponent.

Here is a example of how to interpret the odds.  Let p denote the probablity. If $p = 0.8$, then the odds are: $\frac{0.8}{1-0.8} = \frac{0.8}{0.2}= 4$. We have a 4 to 1 chance of predicting a legendary pokemon.

But as the probability increases towards 1, the odds become expontentially higher. It makes the intrepretation extreme, so what can we do?

---
**Part B: Log Odds (Logit)**

We can take the log of the odds to make it a linear shape, which makes the interpretation more managable. Log of odds is also known as a logit.

The log odds (logit) formula is:

$$\eta_i = logit(p_i) =ln(\frac{p}{1-p}) = \beta_0 + \beta_1 x_{1i} + ... + \beta_k x_{ki}$$

This has a linear shape. (similar to $y= mx+b$ with one predictor.)

Simply by taking the log or ln of the odds, we get a additive effective instead of a exponetial effect. As the slope increases, the more likely the pokemon is legendary. But probability and classes are not associated with linear lines. So what can we do next?

---
**Part C: Logistic Regression from Inverse Logit (Sigmoid Function)**

If we take the inverse of the logit, we end up with probability values instead of linear values.

The logistic regression formula is:

$$p_i = logit^{-1}(\eta_i) = \frac{e^{\eta_i}}{1 + e^{\eta_i}} = \frac{e^{\beta_0 + \beta_1 x_{1i} + ... + \beta_k x_{ki}}}{1 + e^{\beta_0 + \beta_1 x_{1i} + ... + \beta_k x_{ki}}}$$

Logistic regression with 1 predictor is famously known for  it's S-shape curve. Formally this is called a sigmoid. With multiple predictors we have a sigmoid shape in multiple dimensions.



Now by doing a inverse logit on $logit(p_i)$, we isolate $p_i$ and get our probability graph. As some predictors make the the curve go up, the probability of being a legendary class goes up as well.


Where:

* $i$ is i-th observation

* $p_i$ is the probability of class B happening for the i-th observation

* $k$ represents k amount of predictors.

* $x_1,...,x_k$ are the predictors like Base_Stats and capture_rate.

* $\beta_0,..., \beta_k$ are known as the beta coeffcieints, they are just numbers created by the training set.

Enough theory talk, lets go make our logistic regression model!

```python
from sklearn.linear_model import LogisticRegression

logistic_model = LogisticRegression()

logistic_model.fit(X_train, y_train)

y_preds = logistic_model.predict(X_test)

#print(y_preds)

results_df = pd.DataFrame({"Actual values": y_test, "Predicted values": y_preds})
results_df
```

```python
logistic_score = round(accuracy_score(y_test, y_preds),3)
print("Logistic Regression accuracy score:",logistic_score)

logistic_f1 = round(f1_score(y_test, y_preds),3)
print("Logistic Regression F1 score:", logistic_f1)

f1_check(logistic_f1)

```

```python
print("beta coefficients for logistic regression: ", logistic_model.coef_)
print("beta_0 coefficient for logistic regression: ", logistic_model.intercept_ )
```

```python
mat = confusion_matrix(y_test, y_preds)

fig = plt.figure(figsize = (12,7))

sns.heatmap(mat, square=True, annot=True, fmt="d", cbar=False,
            xticklabels = ["not legendary", "legendary"],
            yticklabels = ["not legendary", "legendary"])

plt.title("Confusion Matrix")
plt.xlabel("true label")
plt.ylabel("predicted label");
```

### Conclusion for Logistic Regression:


It is interesting to see that the logistic regression model predicted 4 legendaries but only 2 were correct. It had 9 false non legendaries as well.

KNN is still our best model with 30.8% F1 score. Logistic regression had a F1 score of about 26.7%.

## 🔴 Supervised Classification Model 4: Logistic Ridge


Logistic Ridge is a special type of logistic regression where the beta coeffcients shrink towards zero, but never become zero.

The reason we use Logistic Ridge is to increase the accuracy rate! Lower beta coefficients can reduce overfitting which can increase accuracy when using testing data!


We normally use Logistic Ridge if they believe the predictors are correlated. Total base stats and capture rates seem correlated, since powerful pokemon have high base stats, which usually means they have low capture rates. But I do not think type 1 and type 2 are strongly correlated.

### Theory of Logistic Ridge:



* In logistic LASSO, a lambda($\lambda$) penalty term is used.  The bigger the lambda penalty, the more the beta coefficients shrink towards 0. BUT for Logistic Ridge, the beta coefficients will only be close to 0, not exactly 0!  




* The L2 norm: $||\beta||_2 = \sum^p_{j =1}  \widehat \beta_j^2$ is the reason why our beta coeffcients can become close to zero, but not exactly zero! Think of a 2D plot where the the range of values for x and y are [-1,1]. With L2 norm, we can only make circular lines, and due to the [-1,1] range, we end up making a circle shape. The circle has 4 reference points: (1,0),(0,1), (-1,0), (0,-1).

It is true that a x value ($\widehat \beta_1$) or a y value ($\widehat \beta_2$) can be 0, but the RSS countours usually touch a area near the reference points, not on it. That is why Ridge can only give values close to 0.


* Be careful! We can not make the lambda penalty big and call it a day! Making the lambda penalty too big and making the beta coefficients being zero is not helpful ($\widehat \beta_1^R = 0,...,\widehat \beta_p^R = 0$),  we would just end up with a $\widehat \beta_0^R$ value. This means we predict all observations as class A or class B depending on the value of $\widehat \beta_0^R$.  We want to find the lambda value that gives the highest accuracy rate.



* We want to put a shrink the beta coeffcients so that the model has less variance, but this causes the model to be more biased. What lambda value can give the best compromise value for the most amount of accuracy rate?



* Cross validation (CV) is one way to find the lambda value that gives highest accuracy score. So our models will try to implement CV.


### ALERT:
Python's scikit-learn uses the parameter of $\alpha$ instead of $\lambda$. So please keep this in mind!





You will notice in the code below, we dont have a ($\alpha$) value. Instead we have a C value, which is just the inverse alpha.

$$C = \frac{1}{\alpha}$$  



```python
from sklearn.linear_model import LogisticRegressionCV

# ALERT: We wan to use cross validation to find the best C value!
# Once we find the C value, we can easily get alpha by doing: alpha =  1/C!

# Ridge uses a l2 penalty


logistic_ridge_model = LogisticRegressionCV(
    Cs=10,           # list of log spaced C's to try EX: C's = [0.001,0.01,0.1,1,10,100,]
    cv=5,            # 5-fold cross-validation
    penalty="l2",    # L2 = Ridge
    scoring="accuracy", # accuracy score
    max_iter=1000   # number of steps needed for convergence to find a good C
)

logistic_ridge_model.fit(X_train,y_train)

y_preds = logistic_ridge_model.predict(X_test)

#print(y_preds)

results_df = pd.DataFrame({"Actual values": y_test, "Predicted values": y_preds})
results_df
```

```python
logistic_ridge_score = round(accuracy_score(y_test,y_preds),3)

print("Logistic Ridge accuracy score:",logistic_ridge_score)

logistic_ridge_f1 = round(f1_score(y_test, y_preds),3)
print("Logistic Ridge F1 score:", logistic_ridge_f1)

f1_check(logistic_ridge_f1)

```

```python
print("best C value for logistic ridge: ", logistic_ridge_model.C_)
print("best alpha value for logistic ridge: ", (1/logistic_ridge_model.C_))
print()
print("beta coefficients for logistic ridge: ", logistic_ridge_model.coef_)
print("beta_0 coefficient for logistic ridge: ", logistic_ridge_model.intercept_ )
```

```python
mat = confusion_matrix(y_test, y_preds)


fig = plt.figure(figsize = (12,7))


sns.heatmap(mat, square=True, annot=True, fmt="d", cbar=False,
            xticklabels = ["not legendary", "legendary"],
            yticklabels = ["not legendary", "legendary"])


plt.title("Confusion Matrix")
plt.xlabel("true label")
plt.ylabel("predicted label");
```

Seems like logistic ridge did not predict legendary pokemon well.

### Conclusion for Logistic Ridge:

It is a surprise that regular logistic regression has a F1 score of about 26.7%, but logistic ridge has a F1 score of 20.0%.

Perhaps our predictors were not strongly collinear.

But we can try one more method to modify logistic regression and hopefully get a better accuract score. Let us try Logistic LASSO!

## 🔴 Supervised Classification Model 5: Logistic LASSO (Least Absolute Shrinkage and Selection Operator)


Logistic LASSO is a special type of logistic regression where the beta coeffcients shrink towards zero, and CAN become zero. This means Logisstic  LASSO acts as a variable selector. Lets say $\hat \beta_1 = 0$, then the Base_stats predictor of $x_1$ does nothing, since $0 * any~number = 0$.

So:

 $$p_i = logit^{-1}(\eta_i) = \frac{e^{\eta_i}}{1 + e^{\eta_i}} = \frac{e^{\beta_0 + 0* x_{1i} + ... + \beta_4 x_{4i}}}{1 + e^{\beta_0 + 0 * x_{1i} + ... + \beta_4 x_{4i}}} =  \frac{e^{\beta_0  + ... + \beta_4 x_{4i}}}{1 + e^{\beta_0 + + ... + \beta_4 x_{4i}}}$$


The Base_stats predictor is basically selected out of the model!

Just like Logistic Ridge, the reason we use Logistic LASSO is to improve the accuracy score!


We normally use  Logistic LASSO if it is possible some of the predictors are not important. But this feels sketchy since part C of the classification EDA pretty much showed that the predictors of:  base_stats, capture_rate, type1_encoded, and type2_encoded were all important. Perhaps the type1_encoded and type2_encoded are the predictors that are least important.

### Theory of Logistic LASSO:

* In logistic LASSO, a lambda($\lambda$) penalty term is used.  The bigger the lambda penalty, the more the beta coefficients shrink towards 0. If the lambda penalty becomes big enough, some beta coefficients will be effectively 0.




* The L1 norm: $||\beta||_1 = \sum^p_{j =1}|  \widehat \beta_j|$ is the reason why our beta coeffcients can become zero! Think of a 2D plot where the the range of values for x and y are [-1,1]. With L1 norm, we can only make straight lines, and due to the [-1,1] range, we end up making a diamond shape. There are 4 sharp axis points of the diamond: (1,0),(0,1), (-1,0), (0,-1). If the RSS countour touches one of those axis points is bound to be a x value ($\widehat \beta_1$) or a y value ($\widehat \beta_2$) that has a value equal to zero.


* Careful, we can not make the lambda penalty big and call it a day! Making the lambda penalty too big and making the beta coefficients being zero is not helpful ($\widehat \beta_1^L = 0,...,\widehat \beta_p^L = 0$),  we would just end up with a $\widehat \beta_0^L$ value. This means we predict all observations as class A or class B depending on the value of $\widehat \beta_0^L$.  We want to find the lambda value that gives the highest accuracy rate.



* We want to put a shrink the beta coeffcients so that the model has less variance, but this causes the model to be more biased. What lambda value can give the best compromise value for the most amount of accuracy rate?



* Cross validation (CV) is one way to find the lambda value that gives highest accuracy score. So our models will try to implement CV.


### ALERT:
Python's scikit-learn uses the parameter of $\alpha$ instead of $\lambda$. So please keep this in mind!


You will notice in the code below, we dont have a ($\alpha$) value. Instead we have a C value, which is just the inverse alpha.

$$C = \frac{1}{\alpha}$$


```python

logistic_lasso_model = LogisticRegressionCV(
                        Cs = 10,
                        cv = 5,
                        penalty = "l1", # l1 = lasso
                        solver = "liblinear", # lasso requires liblinear solver
                        scoring = "accuracy",
                        max_iter = 1000
                        )

logistic_lasso_model.fit(X_train,y_train)

y_preds = logistic_lasso_model.predict(X_test)

#print(y_preds)

results_df = pd.DataFrame({"Actual values": y_test, "Predicted values": y_preds})
results_df
```

```python
logistic_lasso_score = round(accuracy_score(y_test,y_preds),3)
print("Logistic LASSO accuracy score:", logistic_lasso_score)

logistic_lasso_f1 = round(f1_score(y_test, y_preds),3)
print("Logistic LASSO F1 score:", logistic_lasso_f1)

f1_check(logistic_lasso_f1)

```

```python
print("best C value for logistic LASSO: ", logistic_lasso_model.C_)
print("best alpha value for logistic LASSO: ", (1/logistic_lasso_model.C_))
print()
print("beta coefficients for logistic LASSO: ", logistic_lasso_model.coef_)
print("beta_0 coefficient for logistic LASSO: ", logistic_lasso_model.intercept_ )
```

```python
mat = confusion_matrix(y_test, y_preds)

fig = plt.figure(figsize = (12,7))


sns.heatmap(mat, square = True, annot = True, fmt = "d", cbar = False,
            xticklabels = ["not legendary", "legendary"],
            yticklabels = ["not legendary", "legendary"])

plt.title("Confusion Matrix")
plt.xlabel("true label")
plt.ylabel("predicted label");
```

It is shocking, but apparently Logistic LASSO and Logistic Ridge has the same F1 score of 20.0% and did worse than regular logistic regression's F1 score of 26.7%.

## 🔴 Supervised Classification Model 6: Support Vector Machine, SVM

Support Vector Machines (SVM) are fascinating machine learning models. The main idea of SVM is that we create a decision boundary, and points called support vectors are involved in making that decision boundary. The decision boundary separates class A and class B points.

There are 4 commonly used kernels for SVM:

* Linear: $K(\vec x_i,\vec x_j) = \vec x_i^T * \vec x_j$

Where $\vec x_i$ and $\vec x_j$ are vectors that contain predictors values of a observation. (I will showcase a vector example in the XGBoost models.)

For data that can be linearly separated, we use a linear kernel. The decision boundary usually a line for 2D points, a plane for 3D points, and a hyperplane for 4D and higher points.

According to the geeks for geeks website, "are commonly used in credit scoring and fraud detection models because they are fast, easy to implement and produce interpretable results" (night_fury1).

* Polynomial: $K(\vec x_i,\vec x_j) = (\vec x_i^T * \vec x_j + c)^d$

Where $d$ is the polyonmial degree and $c$ is a constant.

If data is not linearly separable but still follows a pattern, use the polynomial kernel (night_fury1).

* Radial (formally known as RBF, radial basis function): $K(\vec x_i,\vec x_j) = e^{-\gamma * (\vec x_i - \vec x_j)^2}$

For non-linear data, we usually a radial kernel, although  kernel options such as sigmoid exist.

The radial kernel is most versatile because it can even be used for linear data. So if we do not know what type of data we have, the radial kernel is usually a solid choice.


* Sigmoid: $K(\vec x_i,\vec x_j) = tanh(\alpha \vec x_i^T * \vec x_j + \beta)$

Where $\alpha$ and $\beta$ are constants.

And $tanh(x) = \frac{e^x - e^{-x}}{e^x + e^{-x}}$ is the hyperbolic
tangent where any real numbers of $x$ will be between -1 to 1. So values near -1 are assigned to class A, and values near 1 are assigned to class B.

We can use the sigmoid kernel if the data can not be linearly separated. This is similar to the sigmoid function from logistic regression, and usually best for binary classification (night_fury1).



---

### The radial kernel:


I will be using the radial kernel for SVM. So what does it do?


The radial kernel projects data into a higher dimensional space in order to classify the observations as class A or class B. A hyperplane is created so that the points of A and B are separated. But why?

Suppose a classroom has boys and girls. Usually, boys and girls are sitting in random seats, so a regular 3D plane will not divide boys and girls cleanly. But if we project this classroom to a higher dimension, a hyperplane of maybe 5D or higher can divide the boys and girls cleanly. This can be hard to visualize since we live in 3D (or 4D if you treat time as a dimension).

But perhaps you can imagine two groups of 2D points. Lets say the points of group A is centered at the origin of the plot, and the points of group B surrounds group A.


Next, add a Z axis to the 2D plot. This is how the 2D points get projected to 3D space.

Lets say that the the points of group A have a average of the Z = 10. Meanwhile the points of group B have a average of Z = 2.

Try to picture a 3D plane separating the two groups. Thats the 3D decision boundary.

If we project the 3D space back to 2D space, we get a 2D decision boundary that may look curvy. Inside the boundary are the group A points.

This is what the radial kernel of SVM does.

---

### SVM Process:

1. Before we make a SVM model, we need to standardize the data. SVM uses distance so we do not want one predictor variable that dominates other predictor variables.

2. We fit the SVM model using the standardized data. We then use tuning parameters:

* $C$ is the cost. A high $C$ means that the cost of error is high, so SVM would want to make strict decision boundary. The two margins outside the decision boundary can be narrower or more constricted.   

* $\gamma$ tunes the complexity of the hyperplane. A higher $\gamma$ makes a complex fit, and a lower $\gamma$ makes a simple fit.

A higher $\gamma$ is not always good. A higher $\gamma$ can do well in classifying the training set, but this could overfit and then do worse in classifying the testing set.


3. SVM projects the training data into a higher dimension, and creates the decision boundary.

4. Next, we project back to the original data. the decision boundary can look curvy or non-linear.

5. To be clear, SVM makes a training set decision boundary first. The training set decision boundary is STILL used when using the testing set points.


Enough talk, now lets make a SVM model. We first need to standardize the data.

```python
scaler_function = StandardScaler()

scaler_function.fit(X_train) # let the data only be trained on training data
# do NOT use any testing set info!

X_train_scaled = scaler_function.transform(X_train)

# the testing set is standardized by training set's sample mean and sample std dev
X_test_scaled = scaler_function.transform(X_test)
```

```python
from sklearn.svm import SVC

# rbf is the radial kernel, known as radial basis function

svm_model = SVC(kernel = "rbf", gamma = 1/4, C = 1 )

svm_model.fit(X_train_scaled, y_train)
y_preds = svm_model.predict(X_test_scaled)

# print(y_preds)

results_df = pd.DataFrame({"Actual values": y_test, "Predicted values": y_preds})
results_df
```

```python
mat = confusion_matrix(y_test, y_preds)


fig = plt.figure(figsize = (12,7))


sns.heatmap(mat, square = True, annot = True, fmt = "d", cbar = False,
            xticklabels = ["not legendary", "legendary"],
            yticklabels = ["not legendary", "legendary"])

plt.title("Confusion Matrix")
plt.xlabel("true label")
plt.ylabel("predicted label");
```

So the SVM model for $C=1$ and $\gamma = \frac{1}{4}$ did not classify legendaries well.

Lets try to find the best $C$ and $\gamma$ values!

### Using a FOR loop to find the best C and gamma for SVM

It seems a bit weird for me to use a FOR loop instead of cross validation, but recall that cross validation would use different sets of the WHOLE pokemon data set. This project's main purpose is to just make Gen 9 the testing set, so cross validation can ruin that main purpose.

Lets test numbers in a log scale. (Example of some log scale numbers are: 0.01,0.1,1,10,...)



Many models are affected with parameter changes in log scale, including SVM.

```python
C_list = [0.001,0.01,0.1,1,10,100,1000]
gamma_list = [0.001,0.01,0.1,1,10,100,1000]


accuracy_score_list = []
f1_score_list = []
y_preds_list = []
models_list = []

for C_value in C_list:
  for gamma_value in gamma_list:
    svm_model = SVC(kernel = "rbf", C = C_value, gamma = gamma_value)
    current_model = svm_model.fit(X_train_scaled, y_train)
    models_list.append(current_model)

    y_preds = svm_model.predict(X_test_scaled)
    y_preds_list.append(y_preds)

    svm_accuracy_score = accuracy_score(y_test,y_preds)
    accuracy_score_list.append(svm_accuracy_score)

    svm_f1_score = f1_score(y_test,y_preds)
    f1_score_list.append(svm_f1_score)



#best_f1_score = np.max(f1_score_list) # we want highest score so use max
#print(best_f1_score)

highest_index = np.argmax(f1_score_list)
# we want location of highest score, use f1_score_list!


svm_best_f1_score = round(f1_score_list[highest_index],3) # we want highest score so use max

print("Optimal SVM F1 score:",svm_best_f1_score)


optimal_svm_model = models_list[highest_index]
print(optimal_svm_model)

print()

print("The C value of this SVR model is:", optimal_svm_model.C )
print("The gamma value of this SVR model is:", optimal_svm_model.gamma)
```

```python
y_preds = y_preds_list[highest_index]

results_df = pd.DataFrame({"Actual values": y_test, "Predicted values": y_preds})
results_df
```

### 3D F1 Score Visualization


```python
# first we need to extract all model's C values and gamma values

C_coords = []
gamma_coords = []

for i in range(0,len(models_list)):
  current_svm_model = models_list[i]
  current_C = current_svm_model.C
  C_coords.append(current_C)

  current_svm_model = models_list[i]
  current_gamma = current_svm_model.gamma
  gamma_coords.append(current_gamma)


#print(C_coords)
#print(gamma_coords)
#print(f1_score_list)

print()
print(C_coords[highest_index])
print(gamma_coords[highest_index])
print(f1_score_list[highest_index])

```

```python
# now we can plot!

fig = plt.figure(figsize = (20,12)) # width = 20, height = 12


ax = plt.axes(projection = "3d")

ax.scatter(C_coords, gamma_coords, f1_score_list,
           color = "red", label = "Model's F1 score", s = 100)


# Make one circle in point with highest F1 score!
ax.scatter(optimal_svm_model.C, optimal_svm_model.gamma,
            svm_best_f1_score, facecolors= "none", s = 300,
            edgecolors= "blue", linewidths= 2,
            label= "This model gives highest F1 score, the highest point model")

ax.set_title("3D Plot of F1 Scores")
ax.set_xlabel("C values")
ax.set_ylabel("gamma values")
ax.set_zlabel("F1 score")

ax.view_init(10,75)

ax.legend();
```

Based on this plot, it appears to get high F1 scores, we need a combination of:

(high $C$  of about 1000, and a extremely low $\gamma$ value about 1) this combination gave the highest F1 score in our case.

(low $C$ value of about 100, and extremely low $\gamma$ value of about 1), this combination can still output a high F1 score.

Keep in mind that we can have similar combinations and still have a low F1 score.

### Support Vector analysis:

Some people may think word "vector" is a line, but this is not the case for SVM or even in Linear Algebra.

* The word "vector" for SVM are just the coordinates of the actual points.

* In Linear Algebra, we can draw a vector as a arrow from the origin point, but the vector is still just the coordinates of a point. More specifically, a vector just represents magnitude and direction.

Support vectors are simply the actual data points that "support" the training set decision boundary.

* The support vectors are usually the points that are closest to the training set decision boundary. That's it.

* If we remove these support vectors, the decision boundary will change.

* But removing the non-support vectors will not affect the decision boundary.


```python
### Checking out the support vectors:

optimal_svm_model =  models_list[highest_index]

number_sv = len(optimal_svm_model.support_vectors_)

print("Number of support vectors:", number_sv)
print("First support vector coordinates:", optimal_svm_model.support_vectors_[0])
print("Last support vector coordinates:", optimal_svm_model.support_vectors_[number_sv-1])
print()
print("Number of support vectors for class 0, class of non-legendary Pokemon:",
      optimal_svm_model.n_support_[0])
print("Number of support vectors for class 1, class of legendary Pokemon:",
      optimal_svm_model.n_support_[1])

# It makes sense that we have less support vectors for legendary Pokemon.
# We have significantly less legendary Pokemon then regular Pokemon.
```

```python
svm_score = round(accuracy_score_list[highest_index],3)
print("SVM accuracy score:", svm_score)

svm_f1 = round(svm_best_f1_score,3)
print("SVM F1 score:", svm_f1)

f1_check(svm_f1)

```

```python
mat = confusion_matrix(y_test, y_preds)


fig = plt.figure(figsize = (12,7))


sns.heatmap(mat, square = True, annot = True, fmt = "d", cbar = False,
            xticklabels = ["not legendary", "legendary"],
            yticklabels = ["not legendary", "legendary"])

plt.title("Confusion Matrix")
plt.xlabel("true label")
plt.ylabel("predicted label");
```

### Conclusion for SVM:

## 🔴 Supervised Classification Model 7: Decision Tree (classification version)

Decision trees are a machine learning model that gives easy interpretation, but the prediction could be really off target or extremely biased.

Anyone who does not know machine learning model can grasp the intrepretation of a  decision tree!

For instance, lets say we want to classify the position of a NBA player. We make a simple decision tree with one predictor, age.


*   This tree has one root node (where the tree splits).

*   It has two branches (the pathways of the tree).

*   And has two leaf nodes (the predicted results of a tree).  

If the NBA player is less than 26 years old, go left and classify him as point guard. If the NBA player is 26 years old or more, go right and classify him as a center.


Decision Trees have great power in classification!

* Logistic Regression can only predict binary outcomes of 0,1  or False/True.

* Meanwhile, decision trees can classify MULTIPLE classses!

Logistic Regression would not be appropiate in the NBA example of classifying positions. But a decision tree can do this!

---

### Partitioning:

You can think of paritions as regions or territories.

Classification decision trees are made by partitioning. The partitions depend on the training data.


* Parition 1, a area with a mostly point guards would be classified as the point guard area.

* Partion 2, another area with mostly centers would be classified as the center area.

* Repeat until for all partitions, up to partition K.

---

### Classification Tree goals:

* Partition goal: Create partitions (territories) that mostly contain one class.

* Split goal: A split is chosen by maximizing purity. A partition is pure if most or all points belong to a certain class.

---
### Using a Test set:

When validating with a testing set, the partitions do not change, and we will see if the parititions do well in classifying testing set observations.

If we wanted, perhaps we could have classified which type a pokemon is.

* For instance, maybe a tree would classify a pokemon with high defense as a steel pokemon!

* Maybe it would classify a pokemon with high special attack as a psychic pokemon.

* Or perhaps a pokemon with low capture rate would be classified as a dragon pokemon, since dragon pokemon tend to be hard to capture!

But our focus is on classifying legendary pokemon, lets do that!


```python
from sklearn.tree import DecisionTreeClassifier

# tree 1:

tree_model = DecisionTreeClassifier(random_state = 1)

tree_model.fit(X_train, y_train)

y_preds = tree_model.predict(X_test) # no y_test! we use y_test to compare with y_preds!

#print(y_preds)

results_df = pd.DataFrame({"Actual values": y_test, "Predicted values": y_preds})
results_df
```

```python
tree1_accuracy_score = round(accuracy_score(y_test,y_preds),3)
print("Tree 1's accuracy score",tree1_accuracy_score)

tree1_f1 = round(f1_score(y_test, y_preds),3)
print("Tree 1 F1 score:", tree1_f1)

f1_check(tree1_f1)

```

```python
mat = confusion_matrix(y_test, y_preds)


fig = plt.figure(figsize = (12,7))


sns.heatmap(mat, square=True, annot=True, fmt= "d", cbar=False,
            xticklabels = ["not legendary", "legendary"],
            yticklabels = ["not legendary", "legendary"])

plt.title("Confusion Matrix")
plt.xlabel("true label")
plt.ylabel("predicted label");
```

It honestly seems like the decision tree classified a lot of legendaries right, classifying around 8 legendaries correctly in a pokemon region is a impressive feat.

Tree 1 even had a F1 score of about 57.1%, so Tree 1 did a decent job of predicting legendaries.

But can we make the classification tree predict even better?

### Now lets make the tree simpler!

Tree 1 had no constraints and was able to made so that it could have many branches, layers, and leafs as it wants to adjust for the training set. This is called a un-pruned tree.

HOWEVER, this is NOT ideal when testing against the testing set. The testing set have totally different data compared to the training set. Therefore, a non-pruned tree may not predict well against a testing set! This situation is called overfitting! Tree 1 is a complex tree based on the training set, but this complex tree only predicts the training set well, it will usually predict the testing set horribly!

If we can make the tree 2 more simpler, we could put resrtcitions on how much branches it makes, how much layers it makes, how much observations is required for a leaf, then we force the tree to be more simpler. This is called a pruned tree.

This way, tree 2 can generalize way better and hopefully get a lower error when we predict against the testing set!


```python
# random_state = 1 -> set.seed(1) in R, always make the same tree

# max_depth = 5 -> 5 is the max layers of a tree

# min_samples_split = 10 -> needs at least 10 observations in a node for a split
# a split makes a branch, no split means no branch is made

# min_samples_leaf = 5 -> a leaf needs at least 5 pokemon.
# The leaf is the end part of a tree, where we get our predictions

# all these parameters above can reduce overfitting and hopefully gives less error
# random_state does NOT contribute to reducing overfitting


tree_model = DecisionTreeClassifier(random_state= 1, max_depth = 5,
                                    min_samples_split= 10, min_samples_leaf = 5)



tree_model.fit(X_train, y_train)

y_preds = tree_model.predict(X_test)

#print(y_preds)

results_df = pd.DataFrame({"Actual values": y_test, "Predicted values": y_preds})
results_df
```

```python
tree2_accuracy_score = round(accuracy_score(y_test,y_preds),3)
print("Tree 2's accuracy score:", tree2_accuracy_score)


tree2_f1 = round(f1_score(y_test, y_preds),3)
print("Tree 2's F1 score:", tree2_f1)


f1_check(tree2_f1)

```

```python
mat = confusion_matrix(y_test, y_preds)

fig = plt.figure(figsize = (12,7))


sns.heatmap(mat, square=True, annot=True, fmt= "d", cbar=False,
            xticklabels = ["not legendary", "legendary"],
            yticklabels = ["not legendary", "legendary"])

plt.title("Confusion Matrix")
plt.xlabel("true label")
plt.ylabel("predicted label");

```

### Conclusion for Classification Tree:


Interestingly, tree 1 had a better F1 Score score of 57.1% compared to tree 2's accuracy score of 18.2%. In fact, tree 2 did not do a good job at prediciting legendaries, it only predicted two legendaries! Perhaps we do need a complex tree for prediciting legendaries!

This is possible since a simple tree may have not captured complex patterns. In terms of the bias variance trade off,  this is also a sign that the bias is higher since the tree is not capturing as much information, but the variance is lower since there is less overfitting.

In the section of "Supervised Regression Model 7: Descision Tree (regression version)" we will see that that the root Mean Squared Error for the pruned true is lower than the un-pruned tree.

## 🔴 Supervised Classification Model 8: Random Forest (classification version)

* In many situations, psychology tells us that a group decision is better than a single person's decision. The group is the Random Forest, and the single person is a Decision Tree. This is the intuition of Random Forest, multiple trees tend to give better results than single trees.

* The classification Random Forest model's prediction is actually simple. We use many decision trees, and each tree outputs a class. We take all the classes and do a majority vote. For instance, if class A has the most votes, then we classify that observation as class A.

* For example, let's say we want to classify an NBA player's position. The NBA positions are: Point Guard (PG), Shooting Guard (SG), Small Forward (SF), Power Forward (PF), and Center (C). Let us also say this is the first observation of the testing set, and that this Random Forest has 100 trees. Let's say that  the 100 trees produced the following results: 40 PG, 10 SG, 20 SF, 10 PF, and 20 C.

Since the PG position wins by majority vote, classify observation 1 as PG.

Repeat this process for observation/player 2,...,n!

* Before we talk about how Random Forests are made, we need to talk about bootstrapping.

---

### Bootstrapping:

* Bootstrapping: This is when we have an original sample of data, then randomly pick observations (re-sample) from that original sample with replacement. The new sample is called the bootstrap sample.

We do bootstrapping to mimic real-world data.

* Now this might seem weird, why would I want to re-pick the same data, shouldn't I just get more data? Doesn't bootstrapping feel "weird"? Why am I allowed to use the same observation multiple times as real observations?

* Well sometimes its expensive or just not possible to get more actual data, so this is why bootstrapping was made.

* So what about the "weird" part? Well, MANY objects, animals, or people in real life contain the same elements. For example, let's say we have a sample of 50 students.

Student 3 has blue eyes, is 6 feet tall, has yellow hair, and is American. Does that mean that student 3 is realistically the only person in the world who has blue eyes, is 6 feet, has yellow hair, and is American?

NO! Why? Because in the real world,  there are other people in the world who have blue eyes, are 6 feet, have yellow hair, and are American! The fact that things, animals, or people have common traits is why bootstrapping is so powerful! Bootstrapping may select student 3 multiple times, and this is okay to do!


* An example of bootstrapping: Say we have container 1 that contains 4 colored balls: red, green, blue, yellow. A machine will replace any ball that is taken out of the container, and the ball's respective color will be put in.

The person closes his eyes and picks the balls at random.  The goal is to put 10 balls in container 2.

Perhaps the person picks a red ball, he puts it at container 2, and the machine replaces the red ball with another red ball. He does the process 9 more times. The process soon goes on and container 2 has 5 reds, 2 greens, 3 blues, and 0 yellow.

Random re-sampling happened when the person closed his eyes and picked a ball from container 1 and moved it to container 2.  Replacement happened when the machine replaced the color ball. The bootstrap sample is container 2.

Notice that some colors are picked multiple times due to replacement, and yellow was not picked at all since there is a rare chance any color ball may not even be picked.

---

### The process of Random Forests:

1. A Random Forest model uses the training set, and uses bootstrapping on the training set for each tree. Now each tree will be different since bootstrapping is a random process. Each tree re-samples the same number of observations from the training set. Due to the random process of bootstrapping, roughly 60% of unique training set observations are used, some of the tree's bootstrap observations are bound to be repeats of some training set observations.

Let's say we have 50 observations in the training set. Tree 1 is trained on $obs_1$, $obs_1$, $obs_{37}$,..., $obs_{21}$. But observations: $obs_{14}$, $obs_{18}$, ..., $obs_{47}$ are NEVER selected to train tree 1.

Next, Tree 2 is trained on $obs_7$, $obs_7$, $obs_7$,... $obs_{40}$. But observations: $obs_{33}$, $obs_{35}$, ..., $obs_{50}$ are NEVER selected to train tree 2.

2. The observations that are not selected for training a tree are called Out Of Bag (OOB). These OOB observations are basically the internal testing set. Notice that $obs_1$  appeared multiple times for tree 1's training data, this is due to the randomness of bootstrapping.

For classification random forests, we use OOB observations to evaluate OOB accuracy.

A low OOB accuracy signals that the Random Forest will probably predict testing set observations poorly.

A high OOB accuracy signals that the Random Forest will predict observations in the actual testing set well. (We should note that there are a few cases where a Random Forest with high OOB accuracy predicts terribly.)

As a side note, some people use OOB Error. It is simply: $OOB~Error = 1 - OOB ~ Accuracy$

3. After the random forest is trained and the OOB accuracy is calculated, we test the random forest with an ACTUAL testing set. We get our testing set accuracy score here.

Note that the trained bootstrapped trees remain the same when making predictions on the testing set. Even in an extreme case where a tree is trained on only 10 bootstrap observations while the testing set contains 1,000 observations, the same trained trees from steps 1 and 2 are used for predicitions.

I am aware that we can tune and search for the best possible parameters using GridSearchCV from sklearn.model_selection.
```python
# What each parameter does:

# n_estimators = 500 -> number of trees, 500 trees in this case

# oob_score=True -> gives out-of-bag error


# max_depth = 3 -> 3 is the max layers of a tree

# min_samples_split = 10 -> needs at least 10 observations in a node for a split
# a split makes a branch, no split means no branch is made

# min_samples_leaf = 5 -> a leaf needs at least 5 pokemon.
# The leaf is the end part of a tree, where we get our predictions

# max_features = "sqrt" -> each tree only uses the square root amount of the
# total predictors. Here, we got 4 predictors, so sqrt(4) = 2,
# each tree uses 2 predictors, and these 2 predictors are selected at random
# ALERT: we usually USE this since we want to make each tree totally different!

# class_weight = "balanced" -> handles class imbalance, since we have
# less legendaries random forest tries to give a higher weight to legendaries.
# higher weights means the models tries pay more attention to it

# bootstrap = True by default, don't need to specify this!

# random_state = 1 -> set.seed(1) in R, always make the same tree

# all these parameters above can reduce overfitting and hopefully gives less error
# random_state does NOT contribute to reducing overfitting

# rf model 2:

rf2_model = RandomForestClassifier(n_estimators = 500,
                                  oob_score = True,
                                  max_depth = 3,
                                  min_samples_split = 10,
                                  min_samples_leaf = 5,
                                  max_features = "sqrt",
                                  class_weight = "balanced",
                                  random_state = 1)

rf2_model.fit(X_train,y_train)

y_preds = rf2_model.predict(X_test)

print(y_preds)

results_df = pd.DataFrame({"Actual values": y_test, "Predicted values": y_preds})
results_df
```

## 🔴 Supervised Classification Model 9: XGBoost (classification version)

* XGBoost stands for Extreme Gradient Boosting. For classification, it sums the  baseline number and predictions from multiple trees to  produce one number. Next, it converts that number to a probability.

We will be re-using the log odds and sigmoid function concepts from the Logistic Regression Model.

* Here is an intuition of XGBoost: Let us imagine an archer who is trying to shoot the bullseye. The archer already has some experience with shooting arrows, but he is not an expert. He shoots his first arrow, and it goes way left of the bullseye. He then tries to adjust his second arrow so that hits more to the right, but it still does not hit the bullseye.

He keeps shooting arrows, keeps learning from his mistakes, and keeps noting the distance from his arrow mark to the bullseye.

At his 500th arrow, he shoots the arrow and it finally hits the bullseye! Now, the arrow is not at the middle of the bullseye, but hitting any part of the bullseye is a big feat!

This is XGBoost. This model tries to give "bullseye" predictions.

Here each "arrow" is a tree. Each tree tries to learn the errors of the previous tree. The goal is to minimize the error distance between the prediction and the true value. Usually, the prediction won't exactly be the true value, but it can be really close.


* Random Forest is a parallel approach since it uses independent trees. When training  the trees, they are made randomly due to the bootstrap process.

* XGBoost is a sequential approach. The trees are trained by each tree predicting then learning:

1. We start with a baseline value.

2. Next, tree 1 learns from the baseline value and predicts a residual.

3. then tree 2 learns from the ensemble of tree 1 and the baseline value, and tree 2 gives its own residual. (Ensemble means collection of trees.)

4. Next, tree 3 learns  the ensemble of tree 2, tree 1, and the residual, and tree 3 gives its own predicition.

5. Repeat this for M trees. Combine all tree values plus the baseline value to get our final prediction for the observation.

6. Repeat these steps 1 to 5 for observations 2 to n.   




### Binary classification XGBoost Notation and Formula:

To understand the process of XGBoost, here are the mathematical formulas of XGBoost:

* 1. This is our log odds equation:

$$\widehat y_M(\vec x) = \widehat f_0 + \eta \widehat f_1(\vec x) + \eta \widehat f_2(\vec x) + ... \eta \widehat f_M(\vec x)$$

Where:

* $\vec x$ is the vector of predictors for a certain observation. We plug in those predictors for each tree.

For instance, for observation 1, the first Pokemon is Bulbasaur. We would put Bulbasaur's predictor values of Base_Stats = 318,  capturing_rate = 45, type1_encoded = 0.011628 and type2_encoded = 0,

The predictor vector for Bulbasaur is:
$\vec x_1 =
\begin{bmatrix}
 Base~Stats \\
capturing~rate \\
type1~encoded \\
type2~encoded
\end{bmatrix}  = \begin{bmatrix}
 318 \\
45\\
0.011628 \\
0
\end{bmatrix}$  

We would plug each these values for Bulbasaur in XGBoost and get our predictions. Do the same for the rest of the Pokemon.


* $\widehat f_0$ is the baseline prediction. For classification, we first find the proportion of class B in our training set. we convert that probablity into log odds.


EX: If the proportion of getting a legendary pokemon in a training set is: $\frac{45}{900} = 0.05$, Then lets plug that value into the log odds formula!

$$\widehat f_0 =  ln(\frac{p}{1-p}) = ln(\frac{0.05}{1-0.05}) = ln(\frac{0.05}{0.95}) = -2.944$$

Our baseline value would be $\widehat f_0 = -2.944$ in this scenario.

* $\widehat f_m(\vec x)$ is the prediction from the m-th tree. The m-th tree has learned information of its previous trees.

EX: $\widehat f_2(\vec x)$ is the prediction of tree 2.

Be careful, $\widehat y_2$ is the prediction of the baseline plus tree 1 with learning rate, and tree 2 with learning rate.

* $\eta$ is the learning rate. It is basically the multiplier of each tree's contribution.

We usually use small learning rates  such as 0.01 to 0.30. Each tree makes a small correction. This is a more stable approach but requires more trees such as 300 to 1000.

Big learning rates  such as 0.30 or higher carry some risk. Each tree makes a bigger corrections, but overfitting can happen and the predictions may over estimate.



* $M$ is the total number of trees.

* $\widehat y_M(x)$ is the final prediction after all trees.


2. This is our sigmoid function:

$$\widehat p_i = \frac{e^{\widehat y_M}}{1 + e^{\widehat y_M}}$$

We can input $\widehat y_M$ into the sigmoid function, and get a probablity. If the probablity ends of being higher than 0.5, then XGBoost will classify the observation as class B. If equal to or lower than 0.5, XGBoost will classify the observation as class A.





### Process of XGBoost for classification:

* Training Set process:

Let's say we are using XGBoost to predict if the pet is a dog or a cat. The training set involves families in the state of Ohio who have a cat or dog.

Assume that the proportion of a families having a dog is 0.7.

We want to predict the probability of a family owning a dog. Perhaps the predictors are about
how much time the pet spends outside, how loud the pet is, how often it sleeps, etc.

1. XGBoost makes a baseline prediction which is just calculating the proportion and plugging it in the log odds. Recall that the proportion of dogs is about 0.7. Then:

$$\widehat f_0 = ln(\frac{p}{1-p}) = ln(\frac{0.70}{1-0.70}) = ln(\frac{0.70}{0.30}) = 0.847$$

$\widehat f_0 = 0.847$ is the baseline value for the log odds. (NOT the probability value, the proportion was 0.7!)

Let's say in the training set, observation 1 is a dog, so $y_1 = 1$.

Let $\eta = 0.1$

2. Next, XGBoost makes tree 1. Lets say tree 1 contains leaf 1 of ($hours ~ outside \le 2$), this predicts a value of -0.1. Meanwhile leaf 2 of ($hours ~ outside > 2$) predicts a value of 0.2.

Let's say observation goes to leaf 2.


Then the log odds value is:

$$\widehat y_M = \widehat f_0 + \eta \widehat f_1(\vec x) = 0.847 + (0.1)(0.2) = 0.867$$

And plugging the log odds value into the sigmoid function gives our probablity:

$$\widehat p_1 = \frac{e^{0.867}}{1 + e^{0.867}} = 0.704$$

$\widehat p_1 = 0.704$ is our estimated probablity after 1 tree. Now the true class is 1, so the error is: $1 - 0.704 = 0.296$.

Our prediction of 0.704 is still low and we want to increase that prediction.


3. XGBoost creates tree 2. Lets say tree 2 contains leaf 1 of ($ hours ~ sleep \le 12$), this predicts a value of 0.3. Meanwhile leaf 2 of ($hours ~ sleep > 12$) predicts a value of -0.3.

Let's say observation goes to leaf 2.

Lets say this is a unique dog who sleeps about 14 hours a day. A Thrive Health Pet Care article states "Normal adult dogs sleep about 12 hours a day, but normal adult cats sleep about 15 hours a day" (Thrive Pet Healthcare). So XGBoost probably thinks observation 1 is a cat just because it does not have the same sleeping patterns as most dogs.


Then the log odds value is:

$$\widehat y_M = \widehat f_0 + \eta \widehat f_1(\vec x) + \eta \widehat f_2(\vec x)  = 0.847 + (0.1)(0.2) + (0.1)(-0.3) = 0.837$$

And plugging the log odds value into the sigmoid function gives our probability:

$$\widehat p_1 = \frac{e^{0.837}}{1 + e^{0.837}} = 0.698$$

$\widehat p_1 = 0.698$ is our estimated probablity after tree 1 and tree 2. Now the true class is 1, so the error is: $1 - 0.698 = 0.302$.

Our prediction of 0.698 is still low and we want to increase that prediction. One thing to note is that our adjustments might even lower the probablity.

The next trees will keep adding small adjustments and hopefully this final prediction becomes close to 1.







4. Repeat the process of step 3 for Tree 3 to Tree M, this will get our final log odds prediction ($\widehat y_m$) for observation 1.

Repeat steps 1 to 4 again for observations 2 to n.



---

Let's pretend that the final log odds predictions after 500 trees is:

$$\widehat y_{500} = 3.120$$.

Next the probablity is:

$\widehat p_i = \frac{e^{\widehat y_M}}{1 + e^{\widehat y_M}} = \frac{e^{3.12}}{1 + e^{3.12}} = 0.958$

This is a solid prediction! XGBoost would classify this pet as a dog with greater confidence!

---


* Testing Set Process:

5. When using a testing set, we still use the same log odds formula, we just plug in the predictor values of the testing set:

$$\widehat y_M(\vec x_{test}) = \widehat f_0 + \eta \widehat f_1(\vec x_{test}) + \eta \widehat f_2(\vec x_{test}) + ... +  \eta \widehat f_M(\vec x_{test})$$

Be careful. We do NOT find the log odds of the testing set response value! $\widehat f_0 = ln(\frac{p_{training}}{1-p_{training}}) = ln(\frac{0.70}{1-0.70}) = ln(\frac{0.70}{0.30}) = 0.847$ is always baseline for this dog and cat testing set!

We do not do ANY residual calculations when plugging in testing set values! We still use the same training set trees as well!

6. After we get the testing set predictions, compare them with the actual testing set response values. Calculate the F1 score.

### CV ALERT:
I am aware that we can tune and search for the best possible parameters using GridSearchCV from sklearn.model_selection. However, this is computationally expensive, especially given that  there are many machine learning models in this project. Furthermore, basic tuned parameters can still give great F1 scores.

Our first XGBoost model did not do a good job at classifying legendaries. What if we tuned the xgboost model a lot more?

### Lets make a tuned XGBoost model!

```python

# XGBoost parameters:

# 1. objective = "reg:squarederror" -> use MSE
# this is the default, so no need to mention this!

# 2. n_estimators -> amount of trees to use, called boosting rounds for XGBoost
# n_estimators = 500 trees

# 3. learning_rate (eta) -> how much each tree learns from previous tree
# learning_rate = 0.05 -> eta = 0.05

# 4. max_depth -> how complex each tree is. depth = 1 -> 2 branches.
# depth = 2 -> at most 4 branches, depth = 3 -> at most 8 branches

# 5. subsample -> fraction of training rows used for each tree, prevents overfitting
# subsample = 0.8 -> only use 80% of training data for each tree

# 6. colsample_bytree -> fraction of total predictors to use for each tree
# similar to max_features in random forest
# colsample_bytree = 0.8 -> use 80% of predictors, in our case we have 4 so use 3.

# 7. gamma -> gamma = 0 makes big complex trees, gamma = 5 makes small simple trees

# 8. min_child_weight -> number of samples or weights needed to make a split.
# similar to min_sample_split in random forest, BUT if we use weights, can be different
# min_child_weight = 5 -> need at least 5 pokemon to make a split
# for instance, if we give a legendary a weight of 5, and regular pokemon a weight of 1,
# 2 legendary pokemon and 5 regular pokemon -> (2*5) + 5 = 15 as weight.

# 9. scale_pos_weight -> do number_majority_class / number_rare_class
# this accounts for class imbalance! We have a lot more non legendaries than legendaries
# NOTE: This only works for binary classification!

# 10. random_state = 1 -> set the seed as 1, this makes results reproducible.
# think of this as a minecraft seed, the same minecraft seed always gives a same world!



# xgboost model 2:

# lets get the number of  non legendaries and legendaries in the training test:

number_non_legendary = sum(y_train == 0)
print("Training set's number of non legendaries:", number_non_legendary)

number_legendary = sum(y_train == 1)
print("Training set's number of legendaries:", number_legendary)


xgb2_model = XGBClassifier(n_estimators = 500,
                          learning_rate = 0.05, # eta
                          max_depth = 3,
                          subsample = 0.8,
                          colsample_bytree = 0.8,
                          gamma = 3,
                          min_child_weight = 10,
                          scale_pos_weight = (number_non_legendary/number_legendary),
                          random_state = 1
                          )

xgb2_model.fit(X_train, y_train)

y_preds = xgb2_model.predict(X_test)

#print(y_preds)

results_df = pd.DataFrame({"Actual values": y_test, "Predicted values": y_preds})
results_df
```

```python
xgb2_model_score =  round(accuracy_score(y_test, y_preds),3)
print("XGBoost model 2 accuracy score:", xgb2_model_score)

xgb2_model_f1 = round(f1_score(y_test, y_preds),3)
print("XGBoost model 2 F1 score:", xgb2_model_f1)

f1_check(xgb2_model_f1)
```

```python
mat = confusion_matrix(y_test, y_preds)

fig = plt.figure(figsize = (12,7))


sns.heatmap(mat, square = True, annot = True, fmt = "d", cbar = False,
            xticklabels = ["not legendary", "legendary"],
            yticklabels = ["not legendary", "legendary"])

plt.title("Confusion Matrix")
plt.xlabel("true label")
plt.ylabel("predicted label");
```

## 🔴 Supervised Classification Model 10: Neural Networks (classification version)

* Neural networks are a machine learning model  inspired by our brain neurons. Brain neurons are connected with other neurons. Similarly, Neural Networks use artificial neurons (often symbolized as circles) that are connected with other neurons using lines.

* GeeksForGeeks states that "Neural networks serve as the backbone of AI. They are the core driver of artificial intelligence progress; continue shaping the future of technology and innovation" (Ghorakavi).

* Additionally, GeeksForGeeks states "Neural networks drive key technologies such as powering natural language processing (NLP); enabling self-driving vehicles; supporting automated decision-making systems" (Ghorakavi).

* Let's jump into how Neural Networks work!

---
### Process for Neural Networks (Training Set)

**NOTE:** Before we even begin to make a neural network model, we need to standardize the data. We do this so the the one weight will not dominate another weight. We will talk about weights soon.      

Training a Neural Network consists of 3 parts:

**Part 1- The Input Layer:**

This is the first layer. There will be 4 neurons  in my neural network since I have 4  predictors variables of Base_Stats, capturing_rate, Type 1 encoded, and Type 2 encoded. (Some people call neurons as nodes.)

Also, the training set has around 900 observations, so I feed the neural network about 900 times.

Lets say for example my first Pokemon observation is Bulbasaur. We would put Bulbasaur's predictor values of Base_Stats = 318,  capturing_rate = 45, type1_encoded = 0.011628 and type2_encoded = 0.

The model will input those values in the first layer. Repeat for observations 2 to n.

**Part 2- The Hidden Layer:**

These are the middle layers of the neural network. We can have 1 hidden layer or multiple hidden layers.

The hidden layer involves 4 steps:

1. The hidden layers have neurons that are a linear combination of weights and the variables in the input layer PLUS a bias number (b).

So each hidden layer node has the linear combination: $z = w_1 x_1 + w_2 x_2 + ... w_p x_p + b$.

Where:

$b$ is the bias number.

$x_k$ is a predictor variable.

$w_k$ is the weight number.

$z$ is the linear combination of $w_k$, $x_k$, and $b$.     


2. We usually make the neural network use an activation function. A activation function simply converts our linear combination number of $z$ into a different number.

For the middle layer, we usually use the ReLU activation:

$$f(z) = ReLU(z) = max(0,z)$$

If $z$ is negative, it becomes a value 0. If $z$ is positive, the value remains as $z$.

But why would we want to use a ReLU function in the first place? It is because Neural Networks want to capture non-linear information in predictors. ReLU is a piecewise function:

$$ReLU(z) =
 \begin{cases}
 z, & \text{if } z > 0 \\
 0, & \text{if } z \le 0 \\
 \end{cases}$$

As a refresher, a linear function is any straight line with a constant slope, $y = mx + b$.

It is true that each linear part separately is a linear function. The first part is a straight line that increases. The second part is a horizontal line with a slope of 0.

HOWEVER, a piecewise function in its TOTALITY is NOT linear! A line that has zero slope but suddenly has a slope of $z$ is NOT a linear function! This does not follow the linear function concept $y = mx + b$ for all $x$!    

According to GeeksForGeeks, "ReLU has a Non-Linearity property. Although it seems like a piecewise linear function, ReLU is still a non-linear function. This allows the model to learn more complex data patterns and model intricate relationships between features"("ReLU Activation Function").

Furthermore, GeeksForGeeks states "this helps the neural network maintain the necessary complexity to learn patterns while avoiding some of the pitfalls associated with other activation functions, like the vanishing gradient problem" ("ReLU Activation Function").


3. If the model detects that the weights are bad at predicting, we assign new weights. $w_1^*,w_2^*,...w_p^*$.

This usually happens with a component called the optimizer. It tries to reduce the training set error of a neural network.

There is a optimizer called Adam that adaptively updates weights. Lets use that.

4. If there are additional hidden layers, repeat step 1 to 3. If there are no additional hidden layers, then go to the output layer.

**Part 3- The Output Layer:**

This is the last layer. Here we use a final activation function on $z$, which is $f(z)$. $f(z)$  depends on if we do regression or classification. The final numbers from the output layer will be our predicted numbers or classes.


* In my case, I am doing binary classification, so my activation function is the sigmoid function:

$$f(z) = \frac{e^z}{1+e^{z}}$$

* This activation function gives a probability. (It seems like we have been using a lot of sigmoid functions so far!)


* The Output Layer in this model will end up giving a probabilities. There will only be on output layer neuron in my model.

If the probablity is less than or equal to 0.5 ($\widehat p \le 0.5$), classify the Pokemon
as class 0, the non legendary class.



If the probablity is greater than 0.5 ($\widehat p > 0.5$), classify the Pokemon as class 1, the legendary class.


* As a honorable mention, in the event that have more than 2 classes, we will use the softmax activation function:

$$f(z_i) = softmax(z_i) = \frac{e^{z_i}}{\sum^n_{i=1} e^{z_j}}$$

Where $1 \le j \le n$.

(When we have more than 2 classes, XGBoost and Random forests use this soft max function as well!)

When we have more than 2 classes, the number of neurons in the output layer will become equal the number of neurons. EX: If we have 7 classes, we need 7 neurons for the output layer.

The class with the highest probablity will be assigned for the observation for both activation functions for sigmoid and softmax.  


---
###  How Neural Networks use the Testing Set


* After the neural network gets trained, the weights will NO longer be updated when we use testing set observations!

* We still plug the testing set observations the same way we plugged the training set observations.

* Like any other model, we will compare the predicted values with the test set observations and see how well the neural network's predictions did!

---

### What will perform better, less hidden layers or more hidden layers?

* For my first neural network model, I will chose only 1 hidden layer with 3 neurons.

* For the second neural network model, I will choose 3 hidden layers each with 3 neurons.

* For the third neural network model, I will choose 5 hidden layers each with 3 neurons.

* Just to be clear, we can put any amount of  hidden layers or neurons for the hidden layer part. But too much hidden layers and neurons may cause overfitting. the neural network may just predict the training set well, but not the testing set. (It is even possible to make a neural network with no hidden layers! But where is the fun in that?)


Lets see if the amount of layers affects how well the neural network predicts.



```python
# standardize the data first!

scaler_function = StandardScaler()

scaler_function.fit(X_train)

X_train_scaled = scaler_function.transform(X_train)
X_test_scaled = scaler_function.transform(X_test)

X_train_scaled
# It appears the the target encoded values are standardized as well
# But the still keep consistent values -> encoded information is still there

#X_test_scaled


# NOTE: I have tried using the standardized predictor values in my first run.
# The rMSE values were worse compared to the regular predictor values!!!
# It is good pracitce to standardize predictor values before using
# networks. But this is one case where standardization did worse.
```

### Two issues, weights and randomness!

* Weights:
We need to do just ONE more thing before making neural networks. Our classes are imbalanced, and neural networks might not predict well or consistently due to the class imbalances.



* Random predictions:
Even if we set a TensorFlow seed for neural network, we can still get different answers.

We want to set as much seeds as possible! Using the Python seed, NumPy seed, and TensorFlow seed should help make consistent predictions!




```python
from sklearn.utils import class_weight

class_labels = np.unique(y_train) # 0,1 -> non legendary, legendary
print(class_labels)

# our neural network weights:
class_weights = class_weight.compute_class_weight(class_weight = "balanced",
                                            classes = class_labels,
                                            y = y_train)
print(class_weights)

# zip() -> pairs elements with the same position, and makes it a tuple
# EX: zip(["a","b"],[0,1]) -> [(a,0),(b,1)]

# dict() -> makes a dictionary
# print(dict([("yes",1),("no",0)]))

nn_weights = dict(zip(class_labels,class_weights))

print(nn_weights)
```

### Classification Neural Network model 1:




```python
# neural network model 1: only 1 hidden layer with 3 neurons

# NOTE: For classification Neural Networks, I will use nn1_model = Sequential(Dense())

# For regression Neural Networks, I will use nn1_model = Sequential(), then
# use nn1_model.add(Dense())

# For regression Neural Networks, I will make a function that creates Neural Networks

# 0. import tensor flow imports
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense


# 1. Set seed for reproduciblity: SET ALL SEEDS POSSIBLE!
import random

random.seed(1) # python seed from random module
np.random.default_rng(1) # the old numpy set seed is: np.random.seed(1)
tf.random.set_seed(1) # tensorflow seed for neural networks

# ALERT: tensorflow seed is NOT enough for neural networks, it can still
# make different predictions! we need at least 2 other seeds!!!

# 2. intialize neural network using Sequential
# 3. Inside Sequential, create input layers, hidden layers, and output layer with Dense
# we need to use brackets since we are using a list of Dense()


nn1_model = Sequential([

# this is BOTH the input layer with 4 predictors/neurons and
# the hidden layer with 3 neurons and relu!
Dense(3, input_dim = 4, activation = "relu"),

# output layer with 1 neuron. use sigmoid for binary classification
Dense(1, activation = "sigmoid")

])




# 4. compile the model, AKA decribe its error metrics

# loss = "binary_crossentropy" -> tries to lose error for binary classification

# optimizer = "adam" -> the adam optimizer tries to adjust weights

# metrics = ["f1_score"] -> lets use f1_score


nn1_model.compile(loss = "binary_crossentropy", optimizer = "adam",
                  metrics = ["f1_score"])


# 5. fit the neural network model

# epoch = 100 -> the training set will use the neural network 100 times
# 100 is a good number. The neural network needs to go a bunch of times!

# batch_size = 32 -> 32 observations go through the neural network, THEN the
# weights get updated. Usually do batch_size = 2^k,
# base 2 exponents work more effcitently

# verbose = 0 -> print no output of training error.
# verbose = 1 and 2 -> prints many outputs of training error
# I will do verbose = 0 so that the project will not be cluttered


nn1_model.fit(X_train, y_train, epochs = 100, batch_size = 32,
              class_weight = nn_weights, verbose = 0)

```

```python
# 6. predict the model:

y_preds = nn1_model.predict(X_test)

# y_preds


y_preds = (y_preds > 0.5).astype(int)
# if pred greater than 0.5, classify as 1 (True)

y_preds = y_preds.ravel() # turn a multi-dim array into 1D array

results_df = pd.DataFrame({"Actual values": y_test, "Predicted values": y_preds})
results_df

```

```python
nn1_model_score =  round(accuracy_score(y_test, y_preds),3)
print("Neural Network accuracy score:", nn1_model_score)

nn1_model_f1 = round(f1_score(y_test, y_preds),3)
print("Neural Network F1 score:", nn1_model_f1)

f1_check(nn1_model_f1)
```

```python
mat = confusion_matrix(y_test, y_preds)


fig = plt.figure(figsize = (12,7))


sns.heatmap(mat, square = True, annot = True, fmt = "d", cbar = False,
            xticklabels = ["not legendary", "legendary"],
            yticklabels = ["not legendary", "legendary"])

plt.title("Confusion Matrix")
plt.xlabel("true label")
plt.ylabel("predicted label");
```

The results could vary due to the randomness of Neural Networks.

In my previous run, Neural Network model 1 predicted 10 legendaries which is a huge feat. But it also misclassified some Pokemon as legendaries as well.

It's F1 score is 60.6%, which is a great F1 score!


### Classification Neural Network model 2:

```python
# neural network model 2: 3 hidden layers each with 3 neurons


# 1. Set seed for reproducibility:
random.seed(1)
np.random.default_rng(1)
tf.random.set_seed(1)



# 2. intialize neural network using Sequential
# 3. Inside Sequential, create input layers, hidden layers, and output layer with Dense
# we need to use brackets since we are using a list of Dense()

nn2_model = Sequential([

# this is BOTH the input layer with 4 predictors/neurons and
# the 1st hidden layer with 3 neurons and relu!
Dense(3, input_dim = 4, activation = "relu"),

# 2nd hidden layer with 3 neurons
Dense(3, activation = "relu"),

# 3rd hidden layer with 3 neurons
Dense(3, activation = "relu"),

# output layer with 1 neuron. use sigmoid for binary classification
Dense(1, activation = "sigmoid")

])



# 4. compile the model, AKA decribe its error metrics

# loss = "binary_crossentropy" -> tries to lose error for binary classification
# optimizer = "adam" -> the adam optimizer tries to adjust weights
# metrics = ["f1_score"] -> lets use f1_score

nn2_model.compile(loss = "binary_crossentropy", optimizer = "adam",
                  metrics = ["f1_score"])


# 5. fit the neural network model

# epoch = 100 -> the training set will use the neural network 100 times
# 100 is a good number. The neural network needs to go a bunch of times!

# batch_size = 32 -> 32 observations go through the neural network, THEN the
# weights get updated. Usually do batch_size = 2^k,
# base 2 exponents work more effcitently

# verbose = 0 -> print no output of training error.
# verbose = 1 and 2 -> prints many outputs of training error
# I will do verbose = 0 so that the project will not be cluttered


nn2_model.fit(X_train, y_train, epochs = 100, batch_size = 32,
              class_weight = nn_weights, verbose = 0)

```

```python
# 6. predict the model:

y_preds = nn2_model.predict(X_test)

# y_preds


y_preds = (y_preds > 0.5).astype(int)
# if pred greater than 0.5, classify as 1 (True)

y_preds = y_preds.ravel() # turn a multi-dim array into 1D array

results_df = pd.DataFrame({"Actual values": y_test, "Predicted values": y_preds})
results_df

```

```python
nn2_model_score =  round(accuracy_score(y_test, y_preds),3)
print("Neural Network accuracy score:", nn2_model_score)

nn2_model_f1 = round(f1_score(y_test, y_preds),3)
print("Neural Network F1 score:", nn2_model_f1)

f1_check(nn2_model_f1)
```

```python
mat = confusion_matrix(y_test, y_preds)


fig = plt.figure(figsize = (12,7))


sns.heatmap(mat, square = True, annot = True, fmt = "d", cbar = False,
            xticklabels = ["not legendary", "legendary"],
            yticklabels = ["not legendary", "legendary"])

plt.title("Confusion Matrix")
plt.xlabel("true label")
plt.ylabel("predicted label");

```

```python
best_nn_f1_score = max([nn1_model_f1, nn2_model_f1, nn3_model_f1])
print(best_nn_f1_score)

# best_nn_score = None

def best_nn_score_finder(nn_f1_score):
  if nn_f1_score == nn1_model_f1:
    return nn1_model_score
  elif nn_f1_score == nn2_model_f1:
    return nn2_model_score
  elif nn_f1_score == nn3_model_f1:
    return nn3_model_score
  else:
    print("Recheck if the number is a neural_network F1 score")

best_nn_score = best_nn_score_finder(best_nn_f1_score)
print(best_nn_score)
```

### Conclusion for Classification Neural Network:

If all three Neural Networks' predictions are consistent enough, the first Neural Network is the best model with a F1 score of 60.6%! Compared to other models, it is tied with XGBoost's F1 score of 60.6%.


To address the problem of randomness here are some topics:

* If we really want to make the Neural Network to give consistent results, we can make the CPU fixed. But this method should only be done if we are working in serious data sets.

* One other issue of randomness is that the data set is not big. It only has over 1000 observations in total, and about 900 observations in the training set. Neural Networks do better when we have a lot more data like 100,000 observations.


* NOTE: I tried standardizing the predictors in the Neural Network models. But this gave worse rMSE values, so I ended up using the original predictor values and it gave better rMSE values. Standardization is still a good practice to do, but it will not guarantee better predictions.

### Do we use XGBoost or Neural Network?

Ultimately, we should use XGBoost.

For the next tie breaker, I could simply state that we use the model whose interpretation is more simple. However, there is a better reason to use XGBoost. The better reason is that XGBoost is less affected by randomness.


At least for this project, Neural Network's randomness is one of it's flaw. It could come up with random results and even with setting multiple seeds, it is possible that predictions could slightly vary.

Furthermore, trying to do cross validation to find the best amount of layers, neurons, and maybe even seeds can be too computationally expensive.

For XGBoost, we can just tune the parameters, and we would have similar results regardless of the seed. XGBoost is also robust enough that cross validation is not a necessary process.

Lets start by making our first regression model, Linear Regression!

## 🟢 Supervised Regression Model 1: Linear Regression (Multiple Linear Regression)

Linear Regression is a straight forward model. If we have one predictor variable and one response variable, we simply make a line of best fit and that is our linear regression model!


If we have multiple predictor variables and one response variable, we make a plane or hyperplane  of best fit and that is our linear regression model!


Here is the theoretical formula for linear regression:

$$y_i = \beta_0 + \beta_1 x_{i1} + \beta_2 x_{i2} + ... + \beta_p x_{ip} + e_i$$

And:

 $$e_i \sim N(0, \sigma^2)$$

Here is what each notation means:

* $y_i$ is the response variable value for the i-th observation.

* $\beta_0$ is the intercept coefficient. In a slope line of, $y = mx+b$, $\beta_0$ is $b$.

* $\beta_1,... \beta_p$ are the "slope" coefficients. In a slope line of $y = mx + b$, $\beta_1$ is $m$. In a plane of $z = ax +by + c$, $\beta_1$ is $a$ and $\beta_2$ is $b$.

* $x_{i1},x_{i2},...x_{ip}$ are the predictors. In our model, $x_{i1}$ is height_meters of a Pokemon in the i-th observation. Use the same logic for the next three predictors.  

* $e_i$ is the residual for the i-th observation. The word "residual" is a fancy name for "error". The theoretical residuals can NOT be calculated, but we can calculate the estimated residuals: $\widehat e_i = y_i - \widehat y_i$ (more on estimators in a few lines).

* $e_i \sim N(0, \sigma^2)$ means that the theoretical residuals should be normally distributed with a mean of 0 and have a constant variance of $\sigma^2$.

---

### Why the residual assumption is IMPORTANT!

* Linear Regression makes the best line or plane of fit with residuals! But if the residuals do NOT behave well, then this is a big sign Linear Regression is a bad model for the given data!!

* And be careful! Linear Regression can still fit with data that is either normally distributed or not normally distributed! The data is NOT what determines if Linear Regression is a good fit.

* The RESIDUALS determine if linear regression is a good fit! If the residuals are normally distributed, then Linear Regression is appropriate. If the residuals are not normally distributed, Linear Regression may not be the best model!


---
### We need to put hats when fitting a model!

Lets continue to be careful! when we use the training set to make our linear regression model, we need to put hats on the random variables. A further explanation of random variables will be given a few code chunks later, but for now think of random variables as variables that can get a random value.

$$\widehat y_i = \widehat \beta_0 + \widehat \beta_1 x_{i1} + \widehat \beta_2 x_{i2} + ... + \widehat \beta_p x_{ip} + \widehat e_i$$

Where:

 $$\widehat e_i \sim N(0, \widehat \sigma^2)$$

* The training set only has a sample from the population of data. Therefore a linear regression model will not get true values. This is why use hats like $\widehat \sigma^2$, any variables with hats are known as estimators (AKA statistics). These estimators are used to estimate the population value of $\sigma^2$, also known as the parameter.

* The main idea of linear regression is that the predicted value of $\widehat y_i$ tries to estimate the real value of $y_i$.





---

## Measuring errors of Regression Models, a Case Study

* Linear regression uses a formula called Residual Sums of Squares:

$$RSS = \sum^n_{i=1}(y_i - \widehat y_i)^2$$

RSS is essentially: $RSS = \sum^n_{i=1} (\widehat e_i)^2$

(Notice that we sum the reisduals and square them, that is how RSS got its name!)

This formula answers the question: "How much error did we get from our estimated model"?

* Another error formula that is widely used for both Linear Regression and everyone Mean Squared Error (MSE):

$$MSE = \frac{1}{n}\sum^n_{i=1}(y_i - \widehat y_i)^2$$

MSE is essentially: $MSE = \frac{RSS}{n}$

Notice how we take the mean of the error that is squared, that is where MSE got its name!

* IMPORTANT: The lower the RSS or MSE, the better the model fits!

* Through out this project, I will be using MSE to measure the error of a model! We can use any models such as random forests and regression Neural Networks uses MSE to measure how well the model performed! Lets see what regression model has the lowest MSE score!



```python
from sklearn.linear_model import LinearRegression # 1. from module, import class/function

linear_reg_model = LinearRegression(fit_intercept=True) # 2. intiate model

linear_reg_model.fit(X_train,y_train) # 3. fit model


y_preds = linear_reg_model.predict(X_test) # 4. based on fitted model, predict values

#print(y_preds)

results_df = pd.DataFrame({"Actual values": y_test, "Predicted values": y_preds})
results_df
```

```python
from sklearn.metrics import mean_squared_error

linear_reg_MSE = round(mean_squared_error(y_test, y_preds),3)
print("Linear Regression's MSE:", linear_reg_MSE)

linear_reg_rMSE = round(np.sqrt(linear_reg_MSE),3)
print("Linear Regression's rMSE", linear_reg_rMSE)
```

## Why rMSE?

rMSE simply means "root mean squared error". And there is a interpretation difference between MSE and rMSE.

MSE has squared units. rMSE has regular units.

So lets say someone kicks a soccer ball and we want to measure the distance of how far he can kick the ball, and this person does multiple kick attempts. Next, we try to make a model where we predict how far the ball travels.

Then we calculate the MSE and we have a slight issue! We get our error in squared units. It is a bit cumbersome to say that the error of our model is in $81 ~ meters^2$.

We could take the square root of the error and finally say our model has a error of about $9 ~ meters$.  The interpretation is much easier with regular units compared to squared units!


In this case, linear regression has a rMSE of about 67.58 in terms of pokemon total base stats.

### What is a good rMSE value?

It depends on the response variable on if our rMSE value is good. If our response variable ranges from 0 to 10,000, having a rMSE of 3 is extremely good. But if the response variable ranges from 0 to 7, a rMSE of 3 is terrible.

One effective way to gauge rMSE is by taking the standard deviaition of the response variable. If the rMSE is lower than the standard deviation, it is a good rMSE!

Recall that the sample variance formula is:

$$Var(y_i) = s^2 = \frac{1}{n-1} \sum^n_{i=1} (y_i - \bar y)^2$$

Where:
* $y_i$ is the ith-observation's response variable value.

* $\bar y = \frac{1}{n}\sum^n_{i=1} y_i$ is the mean of the response variable.

* $n$ is the number of observations in the data set.

The standard variation is just the square root of the variance:

$$SD(y_i) = s = \sqrt{\frac{1}{n-1}\sum^n_{i=1} (y_i - \bar y)^2}$$





```python
# standard deviation of response variable
SD_total_base_stats = testing_set["Base_Stats"].std()
print(SD_total_base_stats)

SD_total_base_stats_training = training_set["Base_Stats"].std()
print(SD_total_base_stats_training)


# function that checks the rMSE number of a model and compares it with the st dev
def rMSE_check(rMSE_number):
  if rMSE_number < 0.33 * SD_total_base_stats:
    print("Model's rMSE is GREAT!")
  elif rMSE_number < 0.66 * SD_total_base_stats:
    print("Model's rMSE is good.")
  elif rMSE_number < SD_total_base_stats:
    print("Model's rMSE is ok.")
  elif rMSE_number >= SD_total_base_stats:
    print("Model's rMSE is terrible...")
  else:
    print("This may not be a valid rMSE number, rMSE must be a positive number or 0.")

```

```python
rMSE_check(linear_reg_rMSE)
```

## Random Variables (RV) Case Study

Before we go on to residual assumptions lets review what a random variable is.

The definition of a random variable (RV): A variable that has a random value. The value is not a fixed constant.

* This is a example of a variable that is NOT a RV:

$$x + 2 = 5$$

In this case, $x = 3$ always. Notice $x$ is lower case here too. Lower case letters represent realized values.

* Example of a RV:

$$X \sim Unif(0,1)$$

Here, X can be any decimal value from 0 to 1 since it comes from a uniform distribution from 0 to 1. Notice that $X$ is capital here, we usually denote capital letters as a RV.

RVs apply in the real world as well! Lets say that someone tries to measure a random person's height in the United States. That person would not always be 65 inches. We could have a person with a height of 62.7 inches, 74.91 inches, and so on.

Lets pretend the people's heights are normally distributed with a population mean of 67 inches, and a population variance of 10 inches. A normal distribution is a reasonable assumption since people's heights are continous and can have decimal values, and a LOT of people live in the United States! The variable of a selected person's height is random and could be anything from this normal distribution!

The notation is:

$$X_i \sim N(\mu = 67, \sigma^2 = 10)$$

Where $X_i$ is the RV for the i-th person!

Therefore, in $e_i \sim N(0,\sigma^2)$, the residuals ($e_i$) are RVs since they can have a random value from the Normal distribution.

---

## Residual check:

Even though we do not need to check residuals for predicitng values, lets check them anyways to see if they match of $e_i \sim N(0,\sigma^2)$. We need good residuals to make the intrepretation of linear regression good.

Recall that residuals are the error term, calculated as  observed minus predicted.

$$\hat e_i = y_i - \hat y_i$$

So do the estimated residuals $\hat e_i$ follow the assumptions of the theoritical residuals $e_i$?

(In statistics, we put hats in our estimators, such as $\hat \mu$, because estimators attempt to estimate the value of the unknown true parameter $\mu$)

We need to check four assumptions of the residuals:

1. Have constant mean of 0, $E(e_i) = 0$

The expectation is the average value of a random variable. In theory, the residuals should have a average value of 0.

2. Have constant variance,  $Var(e_i) = \sigma^2$

The variance is the spread of a random variable. In theory, the residuals should be constant. The spread of the residuals should not be increasing or decreasing as we move along the x-axis.

3. Are Normally disributed,  $e_i \sim N(0,\sigma^2)$

The residuals should follow a Normal distribution with mean of 0, and variance of $\sigma^2$.

When using a histogram, the residuals should look roughly like a bell shape since thats the shape of a normal distribution.

We can also use a qqplot and see if the residuals are clustered in the qqline. The qqline is a 45 degree line.

4. Are independent.

A residual observation does NOT affect the next or previous residual observation.



```python
yhat_train = linear_reg_model.predict(X_train)


# recall that residuals = y_i - yhat_i
# we would use the training data for this!

residuals = y_train - yhat_train
print(residuals)
```

### Checking assumptions 1 and 2: Constant mean and constant variance:

For SLR (Simple Linear Regression) we only do one residual plot of $x$ and $y$! For MLR (Multiple Linear Regression), we have multiple residual plots of $x_1$, $x_2$, ... $x_p$ and $y$!

We are doing MLR!

```python
# we should do 4 subplots!

# plot residuals against training set!

# height_meters (weak), weight_kilograms (weak), capture_rate, Is_Legendary


# Figure (fig):  the entire canvas, a canvas contains multiple plots

# ax: a single subplot, so axes means multiple subplots

fig, axes = plt.subplots(2,2, figsize = (14,8)) # 2 by 2 -> 4 subplots
# we should increase the figsize so examining the residual plots are easier!!!

# This is the main title for the residual pltos:
fig.suptitle("Residuals plots for the four predictors!", )

axes[0,0].scatter(training_set["height_meters"], residuals,
                  alpha = 0.3)
axes[0,0].set_xlabel("height in meters")
axes[0,0].set_ylabel("residuals")
axes[0,0].set_title("Residual Plot 1")


# these lines are example of non constant variance
axes[0,0].plot([0, 30], [200, 0], color="red",
               linestyle='--', label="Decreasing variance")
axes[0,0].plot([0, 30], [-200, 0], color="red",
               linestyle='--', label="Decreasing variance")



axes[0,1].scatter(training_set["weight_kilograms"], residuals,
                  alpha = 0.3, color = "green")
axes[0,1].set_xlabel("weight in kilograms")
axes[0,1].set_ylabel("residuals")
axes[0,1].set_title("Residual Plot 2")



axes[1,0].scatter(training_set["capturing_rate"], residuals,
                  alpha = 0.3, color = "red")
axes[1,0].set_xlabel("capture rate")
axes[1,0].set_ylabel("residuals")
axes[1,0].set_title("Residual Plot 3")


# these lines are example of constant variance ->
axes[1,0].plot([0, 270], [200, 200], color="blue",
               linestyle='--', label= "Constant variance")
axes[1,0].plot([0, 270], [-200, -200], color="blue",
               linestyle='--', label= "Constant variance")





axes[1,1].scatter(training_set["Is_Legendary"], residuals,
                  alpha = 0.3, color = "purple")
axes[1,1].set_xlabel("legendary status")
axes[1,1].set_ylabel("residuals")
axes[1,1].set_title("Residual Plot 4")





plt.tight_layout() # this helps fit x label names
```

All points dont seem to increasing or decreasing or making a weird departure, as show in the points contained in the blue lines. Our assumption of constant variance is satisfied here.

## Bias and Variance Trade off for MSE:

Before we head towards more regression models, I want to mention a thoery that is important for getting the the best possible predictons.

If the model is too simple, it could have too much bias. Simple could mean too few predictors or the constraints are too heavy. A simple model is underfitted and will get a lot of error when using a testing set.  The underfitted model does bad in both the training and testing set.

If the model is complex, it could be have too much variance. It has too much predictors or there are minimal constraints.  The model is overfitted and will get a lot of error when using a testing set. The overfitted model does well with only the training set.

We are fixed to having 4 predictors, so we are not changing the number of predictors.

But next models have parameters that can be tuned, such as the number of tree branches, the lambda values, the learning rate, and more. The less complex we make the tuning parameters, the more bias it will have. The more complex they are, the more variance it will have.

Ultimately we want the bias and variance combination that gives the least amount of testing set MSE. Think of a U shape, and we want to reach the absolute bottom of that U.

---

## 🟢 Supervised Regression Model 2: Ridge

This explenation is similar to Logistic Ridge in part D, but there are some differences.

Ridge is a special type of linear regression where the beta coeffcients shrink towards zero, but never become zero.

The reason we use Ridge is to reduce mean squared error (MSE)!


We normally use Ridge if we believe the predictors are correlated. I don't think height and weight are strongly correlated with capture rates and legendary pokemon status, but we can still use Ridge anyways.


### Theory of Ridge Regression:




* In Ridge, a lambda($\lambda$) penalty term is used with the regular linear regression RSS (Residual Sums of Squares). The bigger the lambda penalty, the more the beta coefficients shrink towards 0. BUT the coefficients will NOT be exactly 0, even if $\lambda$ is big!

Here is how the Ridge works:

$$\vec{\widehat{\beta}^{R}} = arg ~ min_{\beta} \left\{ \sum^n_{i=1} (y_i - \widehat y_i )^2 + \lambda \sum^p_{j =1} \widehat \beta_j^2 \right\}$$

Where: $\lambda$ is the penalty term. The bigger the lambda, the more the beta coefficients shrink.

$\vec{\widehat \beta_R}$ is the vector of our modified beta coefficients.  The vector sign is a linear algebra notation for vector. (I am aware many people put a bold font lower case to represent a vector as well, $\mathbf{\widehat \beta_L}$ but the rendering of the bold font might not be working, so I will be sticking to the vector sign.)

This model's beta vector is: $\vec{\widehat \beta^R} =
\begin{bmatrix}
\widehat \beta_0^{R} \\
\widehat \beta_1^{R} \\
\vdots \\
\widehat \beta_4^{R}
\end{bmatrix}$

Lastly, the L2 norm: $||\beta||_2 = \sum^p_{j =1}  \widehat \beta_j^2$ is the reason why our beta coeffcients can become close to zero, but not exactly zero! Think of a 2D plot where the the range of values for x and y are [-1,1]. With L2 norm, we can only make circular lines, and due to the [-1,1] range, we end up making a circle shape. The circle has 4 reference points: (1,0),(0,1), (-1,0), (0,-1).

It is true that a x value ($\widehat \beta_1$) or a y value ($\widehat \beta_2$) can be 0, but the RSS countours usually touch a area near the reference points, not on it. That is why Ridge can only give values close to 0.


* Just like logistic ridge and logistic LASSO, we can not make the lambda penalty big and call it a day. Making the lambda penalty too big and making the beta coefficients being extremely close to zero is not helpful ($\widehat \beta_1^R \approx 0,...,\widehat \beta_p^R \approx 0$),  we just get slight sloped hyper plane.  We want to find the lambda value that gives the lowest possible MSE.

* Getting the lowest possible MSE relates to another theory called the bias-variance tradeoff.

We want to put a shrink the beta coeffcients so that the model has less variance, but this causes the model to be more biased. What lambda value can give the best compromise value for the least amount of test MSE?



* Cross validation (CV) is one way to find the lambda value that gives lowest possible MSE. So our models will try to implement CV.


### ALERT:
Python's scikit-learn uses the parameter of $\alpha$ instead of $\lambda$. So please keep this in mind!


```python
from sklearn.linear_model import RidgeCV
from sklearn.linear_model import ridge_regression # <- problem, has no CV for alpha

# we should do cross validation to find the best alpha value.
# cross validation defintely matters to get a significantly lower rMSE
# furthermore, the computation time is fast for ridgeCV

# one problem with RidgeCV is that it needs alpha guesses,
# RidgeCV

ridge_model = RidgeCV(alphas = [0.001,0.01,0.1,1,10,100, 1000], cv = 10)
ridge_model.fit(X_train,y_train)

y_preds = ridge_model.predict(X_test)

results_df = pd.DataFrame({"Actual values": y_test, "Predicted values": y_preds})
results_df
```

```python
linear_ridge_MSE = round(mean_squared_error(y_test,y_preds),3)
print("Ridge's MSE:",linear_ridge_MSE)

linear_ridge_rMSE = round(np.sqrt(linear_ridge_MSE),3)
print("Ridge's rMSE:", linear_ridge_rMSE)

rMSE_check(linear_ridge_rMSE)
```

Ridge regression had a rMSE of about 69.4 total base stats. A bit higher than regular linear regression's rMSE of 67.5 total base stats

## 🟢 Supervised Regression Model 3: LASSO (Least Absolute Shrinkage and Selection Operator)


This explenation is similar to Logistic LASSO in part D, but there are some differences.

LASSO is a special type of linear regression where the beta coeffcients shrink towards zero, and CAN become zero. This means LASSO acts as a variable selector. Lets say $\widehat \beta_2 = 0$, then the weight predictor of $x_2$ does nothing, since $0 * any~number = 0$.
The weight predictor is basically selected out of the model!

In the formula, the predicted value of $\widehat y_i$ is:

$$\widehat y_i = \widehat \beta_0 + \widehat \beta_1 x_{i1} + 0 * x_{i2} + ... + \widehat \beta_p x_{ip} + \widehat e_i$$

Which reduces to:

$$\widehat y_i = \widehat \beta_0 + \widehat \beta_1 x_{i1}  + ... + \widehat \beta_p x_{ip} + \widehat e_i$$

Just like Ridge, the reason we use LASSO is to reduce mean squared error (MSE)!


We normally use LASSO if it is possible some of the predictors are not important. Recall in part C, the Regression EDA showed that height and weight are weak predictors. Maybe LASSO will make the height and weight beta coefficients to be zero!

### Theory of LASSO Regression:

* In LASSO, a lambda($\lambda$) penalty term is used with the regular linear regression RSS (Residual Sums of Squares). The bigger the lambda penalty, the more the beta coefficients shrink towards 0. If the lambda penalty becomes big enough, some beta coefficients will be effectively 0.

Here is how the LASSO works:

$$\vec{\widehat{\beta}^{L}} = arg ~ min_{\beta} \left\{ \sum^n_{i=1} (y_i - \widehat y_i )^2 + \lambda \sum^p_{j =1}|  \widehat \beta_j| \right\}$$

Where: $\lambda$ is the penalty term. The bigger the lambda, the more the beta coefficients shrink.

$\vec{\widehat \beta^L}$ is the vector of our modified beta coefficients.   The vector sign is a linear algebra notation for vector.

This model's beta vector is: $\vec{\widehat \beta^L} =
\begin{bmatrix}
\widehat \beta_0^{L} \\
\widehat \beta_1^{L} \\
\vdots \\
\widehat \beta_4^{L}
\end{bmatrix}$

Lastly, the L1 norm: $||\beta||_1 = \sum^p_{j =1}|  \widehat \beta_j|$ is the reason why our beta coeffcients can become zero! Think of a 2D plot where the the range of values for x and y are [-1,1]. With L1 norm, we can only make straight lines, and due to the [-1,1] range, we end up making a diamond shape. There are 4 sharp axis points of the diamond: (1,0),(0,1), (-1,0), (0,-1). If the RSS countour touches one of those axis points is bound to be a x value ($\widehat \beta_1$) or a y value ($\widehat \beta_2$) that has a value equal to zero.


* Just like Ridge, we can not make the lambda penalty big and call it a day. Making the lambda penalty too big and making the beta coefficients being zero is not helpful ($\widehat \beta_1^L = 0,...,\widehat \beta_p^L = 0$),  we just get straight hyper plane.  We want to find the lambda value that gives the lowest possible MSE.

* Getting the lowest possible MSE relates to another theory called the bias-variance tradeoff.

We want to put a shrink the beta coeffcients so that the model has less variance, but this causes the model to be more biased. What lambda value can give the best compromise value for the least amount of test MSE?



* Cross validation (CV) is one way to find the lambda value that gives lowest possible MSE. So our models will try to implement CV.


### ALERT:
Python's scikit-learn uses the parameter of $\alpha$ instead of $\lambda$. So please keep this in mind!


```python
from sklearn.linear_model import LassoCV

# ALERT: max_iter must be equal to 10000 or more or else we may get keyboard error!
# set alphas = None to make LassoCV consider ALL alpha (lambda) values

lasso_model = LassoCV(alphas = None, cv = 10, max_iter = 10000)
lasso_model.fit(X_train, y_train)

y_preds = lasso_model.predict(X_test)

#print(y_preds)

results_df = pd.DataFrame({"Actual values": y_test, "Predicted values": y_preds})
results_df
```

I would like to say that LASSO and Ridge seem to have extremely similar predicted values, but some decimals numbers are different.  

```python
linear_lasso_MSE = round(mean_squared_error(y_test, y_preds),3)
print("LASSO's MSE:", linear_lasso_MSE)

linear_lasso_rMSE = round(np.sqrt(linear_lasso_MSE),3)
print("LASSO's rMSE:", linear_lasso_rMSE)

rMSE_check(linear_lasso_rMSE)
```

LASSO regression has a rMSE of about 69.63

```python
print("best alpha value for LASSO: ", lasso_model.alpha_)
print("beta coefficients for LASSO: ", lasso_model.coef_)
print("beta_0 coefficient for LASSO: ", lasso_model.intercept_ )
```

## 🟢 Supervised Regression Model 4: Principal Components Regression, PCR

The supervised model of Principal Components Regression (PCR) comes from the un-supervised model of Principal Components Analysis (PCA).

It is known as a dimension reduction technique since it tries to use less predictors than the original amount.

It uses the linear regression formula in the final step, but the predictors are a linear combination of ALL the predictors.


PCR is inspired by the bias and variance tradeoff. How can we reduce the number of predictor variables so that we can get a lower variance, with a slight increase in bias so that rMSE can be optimally lower?

---

### PCR Process:

1. PCR contains components called Principal Components (PCs). PCs come from a process called Principle Component Analysis (PCA).

For more information of PCA and how the process works, it is mentioned in "Un-Supervised Model 1 Principal Components Analysis (PCA)" under Section F.

2. Each PC contains a linear combination of ALL the predictor variables. Therefore, each PC contains information of all the predictors. Usually the first PC contains the most information, then the second PC contains the second most information, and so on.


$PC_{1,i} = A_1 x_{i,1} + A_2 x_{i,2} + ... + A_p x_{i,p}$

$PC_{2,i} = B_1 x_{i,1} + B_2 x_{i,2} + ... + B_p x_{i,p}$

$and~ so~ on~ for~ next~ PCs$

3. PCR ends up being a linear regression of:
$\widehat y_i = \widehat \beta_0 + \widehat \beta_1 PC_{1,i}  +  \widehat \beta_2 PC_{2,i} + ... + \widehat \beta_k PC_{k,i}$

Where $k < p$

Notation:

$A,B,...$ are the scalars, they are fixed numbers. For instance, $A_1 = 2.71$ and $A_2 = -7.21$ and this stays fixed for PC1 and for all observations.

$x_{i,1},x_{i,2},...,x_{i,p}$ are the predictors! "$i = 1,2,...n$" represents the observations.  Plug in the predictor values for observation 1, repeat for observation 2 to observation n.




First, we would want to standardize the training set and testing set before we apply PCA.

Lets also see what amount of PCs are needed to explain at least 90% of the variance. Those are the amount of PCs to use for PCR.

```python
# scaling is important for PCR, since PCA needs to be scaled

scaler_function = StandardScaler()

scaler_train = scaler_function.fit(X_train)
# Make the training data learn, NOT the testing data!


X_train_scaled = scaler_train.transform(X_train)

X_test_scaled = scaler_train.transform(X_test)

#print(X_train_scaled)
#print(X_test_scaled)




```

```python
from sklearn.decomposition import PCA # 1. from module, import class/function

pca_model = PCA(n_components= 4) # 2. initate model, get 4 PCs,

X_train_pca = pca_model.fit_transform(X_train_scaled)

X_train_pca_df = pd.DataFrame(X_train_pca, columns= ["PC1","PC2","PC3","PC4"])
X_train_pca_df

# print(X_train_pca) # contains PC1, PC2, PC3, PC4
# each value in each slot is the result of a linear combination.
```

```python
pca_model = PCA(n_components = 4)

X_test_pca = pca_model.fit_transform(X_test_scaled)

X_test_pca_df = pd.DataFrame(X_test_pca, columns= ["PC1","PC2","PC3","PC4"])
X_test_pca_df



# print(X_test_pca)
```

```python
# PCA under the hood walkthrough:
# This will be more clear in Un-Supervised Model 1: Principal Components Analysis

# our eigen vectors are here:
print(pca_model.components_)

# we have 4 predictors, and 4 components
# each PC loading for a PC is read by rows!
# so PC1 = 0.419x_1 +... + 0.496x_4

print() # make more space between outputs

# our eigen values are here:
print(pca_model.explained_variance_)

# the eigen values are out explained variance
# PC1 variance -> lambda_1 = 1.971

print()

print(pca_model.explained_variance_ratio_)
# the proportion of variance explained by each PC
# PC1 explains 0.422 of the data's variance,
# if we have more strong correlations, the PC1 may have a higher proportion...

# if we included all 9 PCs, the variance ratios of all 9 PCs add up to 1!
```

```python
pca_model_scree = PCA().fit(X_train_scaled)

plt.figure(figsize=(12,7))  # width= 12 inches, height= 7 inches


ax = plt.axes()

ax.plot(np.cumsum(pca_model_scree.explained_variance_ratio_),
        marker = "o",color = "green")
# we need to set a marker


ax.set_title("Training Set Scree plot of PCs. How much should we select?")
ax.set_xlabel("PCs / number of components")
ax.set_ylabel("Cumulative variance explained")

ax.set_xticks([0,1,2,3]) # use 4 ticks, 0 starts first
ax.set_xticklabels(["1","2","3","4"]) # relabel 0 as 1, then relable 1 as 2,...


```

We should probably use 3 PCs, since that is close to 90% of explaining the cumulative variance.

Using 4 PCs is pointless since that gives back all our information again, its basically using linear regression again!.

```python
PCR_model = LinearRegression(fit_intercept=True)

PCR_model.fit(X_train_pca_df[["PC1","PC2","PC3"]], y_train)

y_preds = PCR_model.predict(X_test_pca_df[["PC1","PC2","PC3"]])

#print(y_preds)


results_df = pd.DataFrame({"Actual values": y_test, "Predicted values": y_preds})
results_df
```

```python
pcr_MSE = round(mean_squared_error(y_test, y_preds),3)
print("PCR's MSE:", pcr_MSE)

pcr_rMSE = round(np.sqrt(pcr_MSE),3)
print("PCR's rMSE:", pcr_rMSE)

rMSE_check(pcr_rMSE)
```

### Conclusion for PCR:

PCR seems to have a big rMSE. Perhaps PCR is only more effective when we have many predictors.

Just by looking at the first 5 and last 5 Pokemon in the testing set, PCR did not do a good job at predicting Pokemons' total base stats.

Regular linear regression had a rMSE of 67.58. This is significantly less error than PCR's rMSE of 85.046.

## 🟢 Supervised Regression Model 5: Partial Least Squares, PLS

Partial Least Squares (PLS) is another dimension reduction technique. It uses the linear regression formula in the final step, but the predictors are a linear combination of ALL the predictors.  

Also PLS works differently than PCR.

PCR uses PCA, and it is concerned with the variance of just the predictor.

PLS is concerned with the covariance between the response variable and each predictor variable.


---
### What is Covariance?

* Covariance: How two different random variables change together.

* Covariance formula:

$$Cov(X,Y) = \frac{1}{n}\sum^n_{i=1} (x_i - \bar x)(y_i - \bar y)$$

* Interpretting Covariance:

Postive Covariance - X and Y tend to increase together. As a example, tall people tend to weigh more.

Negative Covariance - X and Y tend to move in opposite directions. As a example, during a winter day, the more it snows, the less schools tend to open.

Zero Covariance - X and Y are not linearly related. They may still have a non linear relationship. For example, people who are tall and the number of pets they own may have zero covariance.



---

### PLS Process

1. PLS tries to find the relationships between:

$$y \sim x_1, y \sim x_2, ..., y \sim x_p$$

2. PLS will try to calculate the covariance of $y$ and $x_j$, which is $cov(y,x_j)$. Where: $j = 1,...,p$

3. Each predictor in PLS is a linear combination of all the predictors, just like PCR. It can be expressed as:

$PLS_{1,i} = A_1 x_{i,1} + A_2 x_{i,2} + ... + A_p x_{i,p}$

$PLS_{2,i} = B_1 x_{i,1} + B_2 x_{i,2} + ... + B_p x_{i,p}$

$and~ so~ on~ for~ next~ PLS ~ components$

4. PLS ends up being a linear regression of:
$\widehat y_i = \widehat \beta_0 + \widehat \beta_1 PLS_{1,i}  +  \widehat \beta_2 PLS_{2,i} + ... + \widehat \beta_k PLS_{k,i}$

Where $k < p$

Notation:

$A,B,...$ are the scalars, they are fixed numbers. For instance, $A_1 = 2.71$ and $A_2 = -7.21$ and this stays fixed for PLS_1 and for all observations.

$x_{i,1},x_{i,2},...,x_{i,p}$ are the predictors! "$i = 1,2,...n$" represents the observations.  Plug in the predictor values for observation 1, repeat for observation 2 to observation n.


4. If the covariance between the response $y$ and the predictor $x_a$ is big, then PLS gives a big coefficient for $x_a$.

If the covariance between the response $y$ and the predictor $x_b$ is small, then PLS gives a small coefficient for $x_b$.

5. We would want to find the number of components that gives the lowest amount of rMSE. (This is similar to the first classification model. For KNN, we want to find the amount of neighbors that gives the lowest F1 score.)


```python
from sklearn.cross_decomposition import PLSRegression

rMSE_list = [] # create empty list

for k in range(1,5):
  pls_model = PLSRegression(n_components = k)
  pls_model.fit(X_train_scaled, y_train)

  y_preds = pls_model.predict(X_test_scaled)
  pls_MSE = mean_squared_error(y_test, y_preds)
  pls_r_MSE = np.sqrt(pls_MSE)

  rMSE_list.append(pls_r_MSE) # append means add to list

# print(rMSE_list)

best_k_components = np.argmin(rMSE_list) + 1
# need to add 1 since index starts at 0.

best_pls_rMSE = round(rMSE_list[best_k_components],3)

print("The best amount of components to use is:", best_k_components)
print("The PLS model with the lowest rMSE is:", best_pls_rMSE)

rMSE_check(best_pls_rMSE)
```

```python
plt.figure(figsize=(12,7))  # width= 12 inches, height= 7 inches


plt.plot([1,2,3,4], rMSE_list, color = "green", label = "rMSE" )
plt.scatter([1,2,3,4], rMSE_list, color = "green")

# Make one circle in point with lowest rMSE!

plt.scatter(best_k_components, best_pls_rMSE, s = 300, facecolors = "none",
            edgecolors = "red", linewidths = 2,
            label = "This amount of PLS components gives lowest rMSE")


plt.title("How many PLS components gives the lowest rMSE?")

plt.xlabel("The number of PLS components")
plt.ylabel("rMSE")


plt.legend();

```

### Conclusion for PLS:

PLS's rMSE of 67.58 is defintely better than PCR's rMSE of 85.046.

But PLS's rMSE is basically the same to regular Linear Regression's rMSE (67.58 when rounded).

In this case, Linear Regression would be the preferred model since we want the more simple model. If we have a complicated model and a simple model that both have similar rMSE, the simple model gets the edge since the interpretation is more understandable.

## 🟢 Supervised Regression Model 6: Support Vector Regression, SVR

Support Vector Regression (SVR) is basically the regression version of Support Vector Machine (SVM) from section D, classification model 6.

As you may have guessed, SVR uses higher dimensions just like SVM.

But SVR does NOT make a decision boundary. Instead, it makes a tolerance zone.

* SVM had a decision boundary with margins that contained support vectors.

* SVR has a tolerance zone (it is formally called a $\epsilon$ tube).

  * The tolerance zone is basically a cylinder with the radius value equal to the $\epsilon$ value.
  
  * Any points outside the tolerance zone are support vectors, since they adjust the SVR's regression line, plane, or hyperplane.

  * Any points inside the tolerance zone are non-support vectors. They do not  affect the SVR model.

Visualizing SVR can be easy. Think of a 2D line, some points cluttered around the line, and a 3D cylinder.

* The 2D line is the predicted relationship SVR created.

* The points are training set points

* And the 3D cylinder is the tolerance zone.

SVR still uses the same kernels as SVM. Lets continue to use the radial kernel.

---

### SVR Process:

1. Standardize the data.

2. Fit a SVR model with parameters of $C$ and $\gamma$.

3. For the training set, SVR model will use its kernel and create a regression line, plane, or hyperplane. The tolerance zone will adjust the regression shape.

4. Predict response values using the trained SVR model and compare it with the real testing set response values.

```python
from sklearn.svm import SVR

# scaling is important for SVR, or our predictions could be bad
# keep using X_train_scaled and X_test_scaled from PCR!

svr_model = SVR(kernel = "rbf", C = 1/4, gamma = 1)

svr_model.fit(X_train_scaled, y_train)

y_preds = svr_model.predict(X_test_scaled)

#print(y_preds)

results_df = pd.DataFrame({"Actual values": y_test, "Predicted values": y_preds})
results_df
```

```python
svr_MSE = round(mean_squared_error(y_test, y_preds),3)
print("SVR's MSE:", svr_MSE)

svr_rMSE = round(np.sqrt(svr_MSE),3)
print("SVR's rMSE:", svr_rMSE)

rMSE_check(svr_rMSE)
```

This SVR model's rMSE is really high compared to the other regression models.

Can we find the best $C$ and $\gamma$ value?

### Using a FOR loop to find the best C and gamma for SVR

Just like SVM for classification, for SVR, lets try to find the optimal tuning parameters of $\gamma$ and $C$.

```python
C_list = [0.001,0.01,0.1,1,10,100,1000]
gamma_list = [0.001,0.01,0.1,1,10,100,1000]

rMSE_list = []
y_preds_list = []
models_list = []

for C_value in C_list:
  for gamma_value in gamma_list:
    svr_model = SVR(kernel = "rbf", C = C_value, gamma = gamma_value)
    current_model = svr_model.fit(X_train_scaled, y_train)
    models_list.append(current_model) # we can actually put fitted models in a list!

    y_preds = svr_model.predict(X_test_scaled)
    y_preds_list.append(y_preds)

    svm_MSE = mean_squared_error(y_test,y_preds)
    svm_rMSE = np.sqrt(svm_MSE)
    rMSE_list.append(svm_rMSE)





lowest_index = np.argmin(rMSE_list)
# we want location of lowest rMSE!


best_svr_rMSE = round(rMSE_list[lowest_index],3) # we want lowest rMSE
print("The optimal SVR model has a rMSE of:", best_svr_rMSE)

optimal_svr_model = models_list[lowest_index]
print(optimal_svr_model)

print()

print("The C value of this SVR model is:", optimal_svr_model.C )
print("The gamma value of this SVR model is:", optimal_svr_model.gamma)
```

```python
# optimal SVR model predictions for first 5 pokemon, last 5 pokemon

results_df = pd.DataFrame({"Actual values": y_test, "Predicted values": y_preds})
results_df

```

### 3D rMSE Visualization


```python
# first we need to extract all model's C values and gamma values

C_coords = []
gamma_coords = []

for i in range(0,len(models_list)):
  current_svr_model = models_list[i]
  current_C = current_svr_model.C
  C_coords.append(current_C)

  current_svr_model = models_list[i]
  current_gamma = current_svr_model.gamma
  gamma_coords.append(current_gamma)


#print(C_coords)
#print(gamma_coords)
#print(rMSE_list)

print()
print(C_coords[lowest_index])
print(gamma_coords[lowest_index])
print(rMSE_list[lowest_index])

```

```python
# now we can plot!



fig = plt.figure(figsize = (20,12)) # width = 20, height = 12

ax = plt.axes(projection = "3d")

ax.scatter(C_coords, gamma_coords, rMSE_list,
           color = "green", label = "Model's rMSE", s = 100)


# Make one circle in point with lowest rMSE!
ax.scatter(optimal_svr_model.C, optimal_svr_model.gamma,
            best_svr_rMSE, facecolors="none", s = 300,
            edgecolors="blue", linewidths=2,
            label="This model gives lowest rMSE, the lowest point model")

ax.set_title("3D Plot of rMSE Values")
ax.set_xlabel("C values")
ax.set_ylabel("gamma values")
ax.set_zlabel("rMSE")

ax.view_init(10,75)

ax.legend();
```

Based on this plot, it appears to get low rMSE, we need a combination of:

(low $C$  of about 100, and a extremely low $\gamma$ value about 1) this combination gave the lowest rMSE in our case.

(high $C$ value of about 1000, and extremely low $\gamma$ value of about 1), this combination can still output a low rMSE.

Keep in mind that we can have similar combinations and still have a high rMSE.

### Support Vector analysis:

Support vectors in SVR work differently than support vectors in SVM.

1. What does "vector" mean?

* Some people may think word "vector" is a line, but this is not the case for SVR or even in Linear Algebra.

* The word "vector" for SVM are just the coordinates of the actual points.

* In Linear Algebra, we can draw a vector as a arrow from the origin point, but the vector is still just the coordinates of a point. More specifically, a vector just represents magnitude and direction.

2. Support vectors are simply the actual data points that are OUTSIDE the tolerance zone.

* In SVM, we have a decision boundary and two margins that are outside the boundary. For SVR, can think of those margins as the tolerance zone.

* The support vectors are usually the points that are outside the tolerance zone. (The tolerance zone is formally known as the $\epsilon$ tube.) The regression line, plane, or hyperplane is adjusted to be near the support vectors.

3. Effect of support vectors:

* If we remove these support vectors, the tolerance zone will change.

* But removing the non-support vectors will not affect the tolerance zone.



```python
### Checking out the support vectors:

optimal_svr_model = models_list[lowest_index]



number_sv = len(optimal_svr_model.support_vectors_)

print("Number of support vectors:", number_sv)
print("First support vector coordinates:", optimal_svr_model.support_vectors_[0])
print("Last support vector coordinates:", optimal_svr_model.support_vectors_[number_sv-1])
print()

print("Number of non-support vectors:", len(X_train_scaled) - number_sv)
print()

print("Epsilon value:", optimal_svr_model.epsilon)
print("Kernel:", optimal_svr_model.kernel)
```

There are only 2 non-support vectors for the optimal SVR model. It appears only 2 training set points are inside the tolerance zone. the other 903 training set points are outside the tolerance zone.

The epsilon value of 0.1 is the radius of the tolerance cylinder (zone).

### Conclusion for SVR:

The optimal SVR model has a rMSE of 56.14, which is so far the model with the lowest rMSE.

## 🟢 Supervised Regression Model 7: Decision Tree (regression version)

Recall that decision trees are a machine learning model that gives easy interpretation, but the prediction could be really off target or extremely biased, and this is especially true  for a regression tree.

In the section of "Supervised Classification Model 7: Decision Tree (classification version)" it talked about how NBA positions can be classified based on age, if we have a simple classification tree with just a age predictor.

Regression trees follow the same concepts as classification trees BUT we do NOT predict classes, we predict numbers!

For regression, usually use the mean of a kth partition, $\bar y_k$  as our prediction for a leaf. This is NOT always a good prediction for numbers and can be way off target! But in terms of partitioning, the mean of the partition is the best we can do.

As a recap, the mean formula is this:
$$\bar y = \frac{1}{n} \sum^n_{i=1} y_i $$

Where $n$ is the total number of observations, and $y_i$ is the i-th observation.

As a example, imagine if want to predict the yearly salary of a professional MLB base ball player. Lets say that we make a simple decision tree and we have a batting average predictor.

* If player 1 has a batting average of 0.300 or lower, he goes to the left branch and we predict that he earns:
$$\bar y_1 = 1.5 ~ million ~ dollars~ a ~ year$$

* If player 1 has a batting average of higher than 0.300, he goes to the right branch and we predict that he earns:
$$\bar y_2 = 4.7 ~ million ~ dollars ~ a ~ year$$

* The partition 1 would be any data equal or lower than a batting average of 0.300. We would make a line at 0.300 as our parition line. Next, partition 2 would be any data more than 0.300.

* Do this process for player 2 to player n so that we get all our predictions. Do this for both the training set and testing set observations.

IMPORTANT: As always, the training set makes the partitions. After the tree is built, we place the testing set observations into the partitions and see if the parititons are reliable.

---

### Regression Tree goals:

* Partition goal: Create partitions (territories) where the response values are similar.

Perhaps response values of $y_1 = 4.2$, $y_5 = 4.1$, $y_{17} =  4.6$ are grouped together.  A partition may contain these 3 values in a regression tree.


* Split goal: A split is chosen by minimizing MSE. We want to minimize the amount of variance if possible.

A regression tree tries multiple $X$ splits, and the best splits tend to be places where there are grouped $y$ values. Grouped $y$ values tend to have lower variance since there is less spread!

Recall that variance means the spread of data! The less spread, the more accurate our predicitions are!


---

### Variance Note:

* ML predictions: In ML predictions of $\hat y$, too much variance can cause high errors. We want to have lower variance.  

Lets keep in mind that models still must balance bias and variance.

Simple models with low variance tend to have high bias. This leads to underfitting.

We want the right amount of variance and the right amount of bias that gives the lowest MSE value.   


* Raw data: But in terms of raw data in a CSV file or a database, we may want as much variance as possible.

If everyone in a classroom all has brown eyes, that is not informative.

But if everyone in a classroom has blue, green, teal, and black eyes, then perhaps this information can help us discover  insights.

A future unsupervised ML model called PCA will benefit from the raw data's variance.

---

Now lets go make a regression tree!


```python
from sklearn.tree import DecisionTreeRegressor

tree_model = DecisionTreeRegressor(random_state = 1)

tree_model.fit(X_train, y_train)

y_preds = tree_model.predict(X_test)

#print(y_preds)

results_df = pd.DataFrame({"Actual values": y_test, "Predicted values": y_preds})
results_df
```

### Now lets make a more simple tree!

Recall that in the section of "Supervised Classification Model 7: Decision Tree (classification version)" we discussed why a simple tree reduces overfitting and as a result gives a better predictions and lower error.

Lets see if tree 2 (the pruned tree/ more simple tree) will have lower error than tree 1 (the un-pruned tree)!


```python
# random_state = 1 -> set.seed(1) in R, always make the same tree

# max_depth = 5 -> 5 is the max layers of a tree

# min_samples_split = 10 -> needs at least 10 observations in a node for a split
# a split makes a branch, no split means no branch is made

# min_samples_leaf = 5 -> a leaf needs at least 5 pokemon.
# The leaf is the end part of a tree, where we get our predictions

# all these parameters above can reduce overfitting and hopefully gives less error
# random_state does NOT contribute to reducing overfitting

tree_model = DecisionTreeRegressor(random_state= 1, max_depth= 5,
                                   min_samples_split = 10, min_samples_leaf = 5)

tree_model.fit(X_train, y_train)

y_preds = tree_model.predict(X_test)

#print(y_preds)

results_df = pd.DataFrame({"Actual values": y_test, "Predicted values": y_preds})
results_df
```

```python

regression_tree2_MSE = round(mean_squared_error(y_test, y_preds),3)
print("Tree 2's MSE:", regression_tree2_MSE)

regression_tree2_rMSE = round(np.sqrt(regression_tree2_MSE),3)
print("Tree 2's rMSE:", regression_tree2_rMSE)

rMSE_check(regression_tree2_rMSE)
```

## 🟢 Supervised Regression Model 8: Random Forest (regression version)


* In many situations, psychology tells us that a group decision is better than a single person's decision. The group is the Random Forest, and the single person is a Decision Tree. This is the intuition of Random Forest, multiple trees tend to give better results than single trees.

* The regression Random Forest model's prediction is actually simple. We use many decision trees, and each tree outputs a number. We take all the numbers and take the mean.

* For example, let's say we want to predict how much a house will cost. We observe the first house and its characteristics. Let's say that we make a Random Forest with 3 trees. Maybe tree 1 predicts the house to be 201k, then tree 2 predicts 210k, and tree 3 predicts 195k.
Next, the Random Forest takes the mean of these trees:

$$\widehat y_{house_1} = \frac{\widehat y_{tree_1} + \widehat y_{tree_2} + \widehat y_{tree_3}}{3} = \frac{201k + 210k + 195k}{3} = 202k$$

This Random Forest predicts the first house cost as 202k!

Repeat this process for observations/houses 2,...,n!

* We usually use 100 to 500 trees in Random Forests, but I used 3 trees as a example so calculation is easier to understand!

* Regression Random Forests easily benefit from multiple trees. This is because a single regression tree could have high variance AKA the prediction is far off target from the actual value. But if average big and small predictions with some moderate ones, we would get a more accurate prediction. Statistically, this is because the variance gets lower due to more sample sizes:

$$var(\bar y) = \frac{\sigma^2}{n}$$

The reduction in variance will most likely cause the rMSE to be lower as well.



Furthermore, we can apply this in psychology again. In a group setting, we want to get each person's opinion and then average the opinions out to get a group answer. And even if a person's opinion is too big or small, it still contains information.

(As a side note, in a school or work setting, we need someone to give out ideas, even if they are bad ideas so that a group of people can brainstorm other ideas. Sometimes even the bad ideas provides information.  Additionally, people rarely come up with perfect predictions or ideas. So one approach is to toss a bunch of ideas and average those ideas to find the best solution. This is similar to what Random Forests do!)


---

### Bias and Variance examples

Lets say the true value is 72.

* Variance example:

100, 70, 40. Taking the mean of these numbers give 70. $\frac{100 + 70 + 40}{3} = \frac{210}{3} = 70$

So averaging out the results gives takes us near the true value.

Random Forest does the above example since it takes the average value of trees!

* Bias example:

41,40,39. Taking the mean of these numbers give 40. $\frac{41 + 40 + 39}{3} = \frac{120}{3} = 40$

Yes, the variance is low (since the spread of these 3 numbers are low). But these numbers are so biased that it is not near the true value of 72.

Random Forest usually helps during variance problems. Bias problems are another story...

* Moving on, yes, we talked about bootstrapping earlier for classification Random Forests, but here is a refresher just in case!

---

### Bootstrapping:

* Bootstrapping: This is when we have an original sample of data, then randomly pick observations (re-sample) from that original sample with replacement. The new sample is called the bootstrap sample.

We do bootstrapping to mimic real-world data.

* Now this might seem weird, why would I want to re-pick the same data, shouldn't I just get more data? Doesn't bootstrapping feel "weird", why am I allowed to use the same observation multiple times as real observations?

* Well sometimes its expensive or just not possible to get more data, so this is why bootstrapping was made.

* So what about the "weird" part? Well, MANY objects, animals, or people in real life contain the same elements. For example, let's say we have a sample of 50 students.

Student 3 has blue eyes, is 6 feet tall, has yellow hair, and is American. Does that mean that student 3 is realistically the only person in the world who has blue eyes, is 6 feet, has yellow hair, and is American?

NO! Why? Because in the real world,  there are other people in the world who have blue eyes, are 6 feet, have yellow hair, and are American! The fact that things, animals, or people have common traits is why bootstrapping is so powerful! Bootstrapping may select student 3 multiple times, and this is okay to do!


* An example of bootstrapping: Say we have container 1 that contains 4 different candy bars that have different weights: 10g, 20g, 30g, 40g. A machine will replace any candy bar that is taken out of the container, and the candy bar's respective weight will be put in.

The person closes his eyes and picks the candy bars at random.  The goal is to put 10 candy bars in container 2.

Perhaps the person picks a 10g bar, he puts it at container 2, and the machine replaces the transferred 10g bar with another 10g bar. He does the process 9 more times. The process soon goes on and container 2 has 5 10g bars, 4 20g bars, 1 30g bar, and 0 40g bars.

Random re-sampling happened when the person closed his eyes and picked a candy bar from container 1 and moved it to container 2.  Replacement happened when the machine replaced the candy bar. The bootstrap sample is container 2.

Notice that some candy bar weights are picked multiple times due to replacement, and 40g bars are not picked at all since there is a rare chance a certain type of candy bar may not even be picked.

---

### The process of Random Forests:

1. A Random Forest model uses the training set, and uses bootstrapping on the training set for each tree. Now each tree will be different since bootstrapping is a random process. Each tree re-samples the same number of observations from the training set. Due to the random process of bootstrapping, roughly 60% of unique training set observations are used, some of the tree's bootstrap observations are bound to be repeats of some training set observations.

Let's say we have 50 observations in the training set. Tree 1 is trained on $obs_1$, $obs_1$, $obs_{37}$,..., $obs_{21}$. But observations: $obs_{14}$, $obs_{18}$, ..., $obs_{47}$ are NEVER selected to train tree 1.

Next, Tree 2 is trained on $obs_7$, $obs_7$, $obs_7$,... $obs_{40}$. But observations: $obs_{33}$, $obs_{35}$, ..., $obs_{50}$ are NEVER selected to train tree 2.

2. The observations that are not selected for training a tree are called Out Of Bag (OOB). These OOB observations are basically the internal testing set. Notice that $obs_1$  appeared multiple times for tree 1's training data, this is due to the randomness of bootstrapping.

For regression random forests, we use OOB observations to evaluate OOB rMSE.

A high OOB rMSE  signals that the Random Forest will probably predict the testing set observations poorly.

A low OOB rMSE signals that the Random Forest will predict observations in the actual testing set well. (We should note that there are a few cases where a Random Forest with low OOB rMSE predicts terribly.)


3. After the regression random forest is trained and the OOB rMSE is calculated, we test the random forest with an ACTUAL testing set. We get our testing set rMSE here.

Note that the trained bootstrapped trees remain the same when making predictions on the testing set. Even in an extreme case where a tree is trained on only 10 bootstrap observations while the testing set contains 1,000 observations, the same trained trees from steps 1 and 2 are used for predictions.

## 🟢 Supervised Regression Model 9: XGBoost (regression version)



* The explanation is similar to classification XGBoost, but some parts are different, such as the residual calculation and learning.

* XGBoost stands for Extreme Gradient Boosting. For regression, it sums the baseline number and predictions from multiple trees to  produce one number. Thats it! Regression XGBoost is simpler than classification XGBoost!

* Here is an intuition of XGBoost: Let us imagine an archer who is trying to shoot the bullseye. The archer already has some experience with shooting arrows, but he is not an expert. He shoots his first arrow, and it goes way left of the bullseye. He then tries to adjust his second arrow so that hits more to the right, but it still does not hit the bullseye.

He keeps shooting arrows, keeps learning from his mistakes, and keeps noting the distance from his arrow mark to the bullseye.

At his 500th arrow, he shoots the arrow and it finally hits the bullseye! Now, the arrow is not at the middle of the bullseye, but hitting any part of the bullseye is a big feat!

This is XGBoost. This model tries to give "bullseye" predictions.

Here each "arrow" is a tree. Each tree tries to learn the errors of the previous tree. The goal is to minimize the error distance between the prediction and the true value. Usually, the prediction won't exactly be the true value, but it can be really close.


* Random Forest is a parallel approach since it uses independent trees. When training  the trees, they are made randomly due to the bootstrap process.

* XGBoost is a sequential approach. The trees are trained by each tree predicting then learning:

1. We start with a baseline value.

2. Next, tree 1 learns from the baseline value and predicts a residual.

3. then tree 2 learns from the ensemble of tree 1 and the baseline value, and tree 2 gives its own residual. (Ensemble means collection of trees.)

4. Next, tree 3 learns  the ensemble of tree 2, tree 1, and the residual, and tree 3 gives its own predicition.

5. Repeat this for M trees. Combine all tree values plus the baseline value to get our final prediction for the observation.

6. Repeat these steps 1 to 5 for observations 2 to n.    


---

### Regression XGBoost Notation and Formula:

To understand the process of XGBoost, here is the mathematical formula of XGBoost:

$$\widehat y_M(\vec x) = \widehat f_0 + \eta \widehat f_1(\vec x) + \eta \widehat f_2(\vec x) + ... + \eta \widehat f_M(\vec x)$$

Where:

* $\vec x$ is the vector of predictors for a certain observation. We plug in those predictors for each tree.

For instance, for observation 1, the first Pokemon is Bulbasaur. We would put Bulbasaur's predictor values of height_meters = 0.7, weight_kilograms = 6.9, capturing_rate = 45, and Is_Legendary = 0.

The predictor vector for Bulbasaur is:
$\vec x_1 =
\begin{bmatrix}
 height~meters \\
weight~kilograms \\
capturing~rate \\
Is~Legendary
\end{bmatrix}  = \begin{bmatrix}
 0.7 \\
6.9\\
45 \\
0
\end{bmatrix}$

We would plug each these values for Bulbasaur in XGBoost and get our predictions. Do the same for the rest of the Pokemon.

* $\widehat f_0$ is the baseline prediction. For regression, it is usually the mean of the response variable $y$ which is $\bar y$.

* $\widehat f_m(\vec x)$ is the prediction from the m-th tree. The m-th tree has learned information of it's previous trees.

EX: $\widehat f_2(\vec x)$ is the prediction of tree 2.

Be careful, $\widehat y_2$ is the prediction of the baseline plus tree 1 with learning rate, and tree 2 with learning rate.

* $\eta$ is the learning rate. It is basically the multiplier of each tree's contribution.

We usually use small learning rates  such as 0.01 to 0.30. Each tree makes a small correction. This is a more stable approach but requires more trees such as 300 to 1000.

Big learning rates  such as 0.30 or higher carry some risk. Each tree makes a bigger corrections, but overfitting can happen and the predictions may over estimate.



* $M$ is the total number of trees.

* $\widehat y_M(x)$ is the final prediction after all trees.

Just to be clear, for regression XGBoost, the baseline value is usually the biggest number, then tree 1's predicted number is significantly small, and Tree M has the smallest number. We add up those numbers from baseline to tree M.

Lets use some pretend numbers as a example: $\widehat y_M = 190k + 1k + 0.9k + ... + 0.01k = 201.71k$

---

### Process of XGBoost for regression:

* Training Set process:

Let's say we are using XGBoost to predict house prices.

Perhaps factors such as number of bathrooms and the square feet of a house could be predictors for determining house price.


The first observation/house costs 200k.


1. XGBoost makes a baseline prediction which is usually just the mean of the response variable, $\widehat f_0 = \bar y$.

Say we find out that the mean of all houses in a certain neighborhood is 190k. So $\bar y = 190k$.

Then the error/residual is:
$$r_1 = y - \widehat f_0$$

$r_1 = 200k - 190k = 10k$

2. Tree 1 learns to about this residual.

Let $\eta = 0.10$ in this example.  


Be CAREFUL! Most times, the tree prediction might not equal the residual. Perhaps we get a result of tree 1 predicting 8k, even though the residual is 10k.

Perhaps tree 1 has leaf 1 ($square ~ feet \le 2000$) predicts a value of 5k. Meanwhile, leaf 2 ($square ~ feet > 2000$) predicts a value of 8k.

Lets say that observation 1 has more than 2000 square feet, so it falls in leaf 2.


$$\widehat y_1 = \widehat f_0 + \eta \widehat f_1(\vec x)$$

$$\widehat y_1 = 190k + (0.1)(8k) = 190.8k$$

Now the most recent residual is:
$$r_2 = y - \widehat y_1$$

$$r_2 = 200k - 190.8k = 9.2k$$


3. Tree 2 learns about this recent residual.


Perhaps tree 2 has leaf 1 ($bathrooms \le 2.5$) predicts a value of 3k. Meanwhile, leaf 2 ($bathrooms > 2.5$) predicts a value of 7k.

Lets say that observation 1 has 3 bathrooms, so it falls in leaf 2.



$$\widehat y_2 = \widehat f_0 + \eta \widehat f_1(\vec x) + \eta \widehat f_2(\vec x)$$

$$\widehat y_2 = 190 + (0.1)(8k) + (0.1)(7k) = 191.5k$$

Now the most recent residual is:

$$r_3 = y - \widehat y_2$$

$$r_3 = 200k - 191.5k = 8.5k$$


4. Repeat the process of step 3 for Tree 3 to Tree M, this will get our final prediction for observation 1.

Repeat steps 1 to 4 again for observations 2 to n.

---

If we were to make a XGBoost on just 2 trees, the final prediction will be:
$$\widehat y_2 = 191.5k$$

This is NOT close to 200k, but with many more trees, the prediction will be much closer!

Lets pretend that the final prediction for 500 trees is:

$$\widehat y_{500} = 201.11k$$, this is a solid prediction!

---

* Testing Set Process:

5. When using a testing set, we still use the same formula, we just plug in the predictor values of the testing set:

$$\widehat y_M(\vec x_{test}) = \widehat f_0 + \eta \widehat f_1(\vec x_{test}) + \eta \widehat f_2(\vec x_{test}) + ... + \eta \widehat f_M(\vec x_{test})$$

Be careful. We do NOT take the mean of the testing set response value! $\widehat f_0 = \bar y_{train}$ is always the mean of the training set response value!

We do not do ANY residual calculations when plugging in testing set values! We still use the same training set trees as well!

6. After we get the testing set predictions, compare them with the actual testing set response values. Compute the MSE and rMSE.

### Lets make a tuned XGBoost model!

```python

# XGBoost parameters:

# 1. objective = "reg:squarederror" -> use MSE
# this is the default, so no need to mention this!

# 2. n_estimators -> amount of trees to use, called boosting rounds for XGBoost
# n_estimators = 500 trees

# 3. learning_rate (eta) -> how much each tree learns from previous tree
# learning_rate = 0.05 -> eta = 0.05

# 4. max_depth -> how complex each tree is. depth = 1 -> 2 branches.
# depth = 2 -> at most 4 branches, depth = 3 -> at most 8 branches

# 5. subsample -> fraction of training rows used for each tree, prevents overfitting
# subsample = 0.8 -> only use 80% of training data for each tree

# 6. colsample_bytree -> fraction of total predictors to use for each tree
# similar to max_features in random forest
# colsample_bytree = 0.8 -> use 80% of predictors, in our case we have 4 so use 3.

# 7. gamma -> gamma = 0 makes big complex trees, gamma = 5 makes small simple trees

# 8. min_child_weight -> number of samples or weights needed to make a split.
# similar to min_sample_split in random forest, BUT if we use weights, can be different
# min_child_weight = 5 -> need at least 5 pokemon to make a split
# for instance, if we give a legendary a weight of 5, and regular pokemon a weight of 1,
# 2 legendary pokemon and 5 regular pokemon -> (2*5) + 5 = 15 as weight.

# 9. random_state = 1 -> set the seed as 1, this makes results reproducible.
# think of this as a minecraft seed, the same minecraft seed always gives a same world!



# xgboost model 2:

xgb2_model = XGBRegressor(n_estimators = 500,
                          learning_rate = 0.05, # eta
                          max_depth = 3,
                          subsample = 0.8,
                          colsample_bytree = 0.8,
                          gamma = 3,
                          min_child_weight = 10,
                          random_state = 1
                          )

xgb2_model.fit(X_train,y_train)

y_preds = xgb2_model.predict(X_test)

# print(y_preds)

results_df = pd.DataFrame({"Actual values": y_test, "Predicted values": y_preds})
results_df
```

```python
xgb2_MSE = round(mean_squared_error(y_test,y_preds),3)
print("XGBoost Model 2 MSE:",xgb2_MSE)

xgb2_rMSE = round(np.sqrt(xgb2_MSE),3)
print("XGBoost Model 2 rMSE:", xgb2_rMSE)

rMSE_check(xgb2_rMSE)
```

Tuning the XGBoost model defintely helped reduce the rMSE. The first XGBoost model has a rMSE of 68.032. The second XGBoost model has a rMSE of 59.159.

## 🟢 Supervised Regression Model 10: Neural Networks (regression version)

* Neural Networks for classification and regression are really similar. Their main differences are located in the output layer.

* So for the most part, we are doing the **nothing new**! It is just like the classic office meme: "Corportate needs you to find the difference between the first picture and the second picture." And Pam says "They are the same thing."

* There will be a few differences for regression Neural Networks though, so reading carefully is advised. For instance, the Adam optimizer will want to reduce MSE for regression tasks.


* Neural Networks are a machine learning model  inspired by our brain neurons. Brain neurons are connected with other neurons. Similarly, Neural Networks use artificial neurons (often symbolized as circles) that are connected with other neurons using lines.

* GeeksForGeeks states that "Neural networks have a non-linearity property. Neural networks are able to model and comprehend complicated relationships in data by virtue of the non-linear activation functions found in neurons which overcome the drawbacks of linear models" (Ghorakavi).

And a significant portion of our models used some form of linear regression, so let's see if Neural Networks can predict total base stats effectively!

* Furthermore, GeeksForGeeks states "Neural networks can be used for finances. Some examples are: Predicting stock prices, fraud detection, and risk management" (Ghorakavi).

Since Neural Networks can comprehend complicated relationships in data, it might be a great choice for financial applications as well.

* Here is a refresher of what Neural Networks do!

---
### Process for Neural Networks (Training Set)

**NOTE:** Before we even begin to make a neural network model, we need to standardize the data. We do this so the the one weight will not dominate another weight. We will talk about weights soon.      



Training a Neural Network consists of 3 parts:

**Part 1- The Input Layer:**

This is the first layer. There will be 4 neurons  in my neural network since I have 4  predictors variables of Base_Stats, capturing_rate, Type 1 encoded, and Type 2 encoded. (Some people call neurons as nodes.)

Also, the training set has around 900 observations, so I feed the neural network about 900 times.

Lets say for example my first Pokemon observation is Bulbasaur.  We would put Bulbasaur's predictor values of height_meters = 0.7, weight_kilograms = 6.9, capturing_rate = 45, and Is_Legendary = 0.

The model will input those values in the first layer. Repeat for observations 2 to n.

**Part 2- The Hidden Layer:**

These are the middle layers of the neural network. We can have 1 hidden layer or multiple hidden layers.

The hidden layer involves 4 steps:

1. The hidden layers have neurons that are a linear combination of weights and the variables in the input layer PLUS a bias number (b).

So each hidden layer node has the linear combination: $z = w_1 x_1 + w_2 x_2 + ... w_p x_p + b$.

Where:

$b$ is the bias number.

$x_k$ is a predictor variable.

$w_k$ is the weight number.

$z$ is the linear combination of $w_k$, $x_k$, and $b$.


2. We usually make the neural network use an activation function. A activation function simply converts our linear combination number of $z$ into a different number.

For the middle layer, we usually use the ReLU activation:

$$f(z) = ReLU(z) = max(0,z)$$

If $z$ is negative, it becomes a value 0. If $z$ is positive, the value remains as $z$.  

But why would we want to use a ReLU function in the first place? It is because Neural Networks want to capture non-linear information in predictors. ReLU is a piecewise function:

$$ReLU(z) =
 \begin{cases}
 z, & \text{if } z > 0 \\
 0, & \text{if } z \le 0 \\
 \end{cases}$$

As a refresher, a linear function is any straight line with a constant slope, $y = mx + b$.

It is true that each linear part separately is a linear function. The first part is a straight line that increases. The second part is a horizontal line with a slope of 0.

HOWEVER, a piecewise function in its TOTALITY is NOT linear! A line that has zero slope but suddenly has a slope of $z$ is NOT a linear function! This does not follow the linear function concept $y = mx + b$ for all $x$!

According to GeeksForGeeks, "ReLU has a Non-Linearity property. Although it seems like a piecewise linear function, ReLU is still a non-linear function. This allows the model to learn more complex data patterns and model intricate relationships between features"("ReLU Activation Function).

Furthermore, GeeksForGeeks states "this helps the neural network maintain the necessary complexity to learn patterns while avoiding some of the pitfalls associated with other activation functions, like the vanishing gradient problem" ("ReLU Activation Function").


* One more thing, what if we want to predict negative numbers for regression? Will our final output be 0?

NO! This is because the last hidden layer will make its $z$ value again with new weights. BUT the output layer will NOT use ReLU. Therefore, our $z$ value can be  negative and the final output is not forced to be 0.

3. If the model detects that the weights are bad at predicting, we assign new weights. $w_1^*,w_2^*,...w_p^*$.

This usually happens with a component called the optimizer. It tries to reduce the training set error of a neural network.

There is a optimizer called Adam that adaptively updates weights. Lets use that.

Adam tries to reduce the MSE  since it knows the training set's true $y$ response values. For regression neural networks, MSE is the loss function here and Adam wants to reduce the MSE by updating the weight values.

4. If there are additional hidden layers, repeat step 1 to 3. If there are no additional hidden layers, then go to the output layer.

**Part 3- The Output Layer:**

This is the last layer. Here we use a final activation function on $z$, which is $f(z)$. $f(z)$  depends on if we do regression or classification. The final numbers from the output layer will be our predicted numbers or classes.


* In my case, we are doing regression , so my activation function is the linear function:

$$f(z) = z$$

This activation function looks so simple. You might think its a typo, but it is NOT a typo!!!

The value of $z$ is already a continuous number. Recall that $z = w_1 x_1 + w_2 x_2 + ... w_p x_p + b$.  So in a way, $z$ acts as linear regression.

So since $z$ was already a continuous number from the start, we do NOT have to do anything else to it!  







---
###  How Neural Networks use the Testing Set


* After the neural network gets trained, the weights will NO longer be updated when we use testing set observations!

* We still plug the testing set observations the same way we plugged the training set observations.

* Like any other model, we will compare the predicted values with the test set observations and see how well the neural network's predictions did!

---

### Regression Neural Network 1:

```python
# 0. Set up the imports: (yes, we already did at classification neural network)
# but it would be a good to restate the imports as a habit

import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense



# 1. set seeds
import random # python's random number generator

random.seed(1)
np.random.default_rng(1)
tf.random.set_seed(1)

```

```python

# ALERT: Lets create a function that can make 3 neuron hidden layer neural networks!


def make_neural_network(hidden_layer_amount):
# 2. Intiate model with Sequential
  nn_model = Sequential()

# 3. Add input layers, hidden layers, output layers

# The input layer (4 neurons since 4 predictors) AND hidden layer (3 neurons with relu)
  nn_model.add(Dense(3, input_dim = 4, activation = "relu"))

  for x in range(0, hidden_layer_amount - 1):
    # ALERT! Since we already have one hidden layer already, if user writes 1,
    # we want this FOR loop to NOT run!!!

    # Add n-1 hidden layers, each with 3 neurons and relu :
    nn_model.add(Dense(3, activation = "relu"))

  # 4. Output layer only has 1 neuron. Use linear activation function
  # since this is regression
  nn_model.add(Dense(1, activation = "linear"))

  nn_model.compile(loss = "mse", optimizer = "adam",
                   metrics = ["mse"])

  # ALERT: We NEED to RETURN the model we created! OR function gives nothing!
  return nn_model


```

```python
# Lets set up the seeds one more time:


random.seed(1)
np.random.default_rng(1)
tf.random.set_seed(1)

nn1_model = make_neural_network(1)
# print(nn1_model)

# 5. fit the model:
# epochs = 100 -> the data set goes through the neural network model 100 times
# batch_size = 32, amount of observations that goes through neural network
# until the weights get updated

nn1_model.fit(X_train, y_train, epochs = 100, batch_size = 32,
              verbose = 0)
```

```python
y_preds = nn1_model.predict(X_test)

y_preds = y_preds.ravel() # make the multi-dim array into a 1D array


results_df = pd.DataFrame({"Actual values": y_test, "Predicted values": y_preds})
results_df
```

```python
nn1_MSE = round(mean_squared_error(y_test,y_preds),3)
print("Neural Network Model MSE:", nn1_MSE)

nn1_rMSE = round(np.sqrt(nn1_MSE),3)
print("Neural Network Model rMSE:", nn1_rMSE)

rMSE_check(nn1_rMSE)
```

## 🟢 Supervised Regression Model 11: K Nearest Neighbors Regressor (KNNR)

Since we started out supervised model journey with KNN for classification, we got to end it by using KNN for regression!

* Just like classification KNN, the regression KNN is simple. For each testing set point, this model uses the K nearest training set points. The predicted response variable is simply the average of the K response variables:

$$\hat y_i = \frac{1}{K} \sum^K_{i=1} y_i$$

* Just like classification KNN, KNN regressor usually select the nearest points by the shortest euclidian distance possible. Think of the distance formula. Here is a example with two variables: $d_E = \sqrt{(x_0 - x_1)^2 + (y_0 -y_1)^2}$




* A fun fact about KNNR is that it can technically be used as a imputer for NA or missing continous data. The KNNimputer class e in sklearn.imputer uses the same concept. It fills missing values by taking the average of the K neighbors! In a way, KNN imputer treats the row with a missing value as a testing set!   A Medium article says " KNN often provides more accurate results than simpler methods like mean  imputation, especially for complex datasets" (Kashyap).

* We do not have any categorical predictors that are strings so that is good. We do have a categorical predictor "Is_Legendary", but that uses numbers of 0 and 1.

### Standardization formula:

* We should try to standardize the predictors before using KNNR, we do not want any predictors dominating in terms of how big their numbers are. This can lead to inaccurate estimates of $\hat y_i$.

$x_{i,standardized} = \frac{x_i - \bar x}{\sigma}$

* Like KNN for classification, we can use Standard Scaler from sklearn.preprocessing.

```python
X_train_knn = training_set[["height_meters", "weight_kilograms",
                        "capturing_rate", "Is_Legendary"]]

X_test_knn = testing_set[["height_meters", "weight_kilograms",
                        "capturing_rate", "Is_Legendary"]]

scaler_function = StandardScaler()

X_train_knn_scaled = scaler_function.fit_transform(X_train_knn)
print(X_train_knn_scaled[0:5])

print()

X_test_knn_scaled = scaler_function.fit_transform(X_test_knn)
print(X_test_knn_scaled[0:5])

```

```python
from sklearn.neighbors import KNeighborsRegressor

knnr_model = KNeighborsRegressor(n_neighbors = 7)

knnr_model.fit(X_train_knn_scaled, y_train)

y_preds = knnr_model.predict(X_test_knn_scaled)

# print(y_preds)

results_df = pd.DataFrame({"Actual values": y_test, "Predicted values": y_preds})
results_df
```

```python
knnr_model_MSE = round(mean_squared_error(y_test, y_preds),3)
print("KNNR model's MSE: ", knnr_model_MSE)

knnr_model_rMSE = round(np.sqrt(knnr_model_MSE),3)
print("KNNR model's rMSE", knnr_model_rMSE)


rMSE_check(knnr_model_rMSE)
```

### Can we find the best K value

What K value gives the lowest rMSE?

```python
k_range = range(1,41,1) # start at 1, stop at 20 (since end is EXCLUSIVE), step by 1

k_scores = [] # empty list that stores our accuracy scores

# we want to use the scaled data!!!

for i in k_range:
  knn_model = KNeighborsRegressor(n_neighbors = i)
  knn_model.fit(X_train_knn_scaled, y_train)
  y_preds = knn_model.predict(X_test_knn_scaled)
  temp1_score = round(np.sqrt(mean_squared_error(y_test,y_preds)),3)
  k_scores.append(temp1_score)

# print(k_scores)

# print(k_scores)

# use the numpy aggregate function np.max() to get the max value of the array
# use vectorized scalar addition of + 1,
# this goes from index 0 to 19, we want 1 to 20 though

all_k_max = np.where(k_scores == np.min(k_scores))[0] + 1
# [0] returns a 1d array


print("Best possible k values for rMSE:", all_k_max)
print("Best rMSE from these k values: ", np.min(k_scores) )


# since K = 20 had the lowest rMSE, use that instead of K = 7!
best_knnr_rMSE = np.min(k_scores)

```

```python

fig = plt.figure(figsize = (15,8))

ax = plt.axes()

ax.plot(k_range, k_scores, label = "rMSE")
ax.scatter(k_range, k_scores) # lets plot each point to make visualization easier

lowest_index = np.argmin(k_scores) # find index with lowest rMSE
#print(lowest_index)
min_k_range = k_range[lowest_index]
min_k_score = k_scores[lowest_index]

# Make one circle in point with lowest rMSE!
plt.scatter(min_k_range, min_k_score, s=300, facecolors="none",
            edgecolors="red", linewidths=2,
            label="This amount of K neighbors gives lowest rMSE")

ax.set_xlabel("The K neighbors number")
ax.set_ylabel("rMSE")

ax.set_title("What amount of K neighbors gives us the lowest rMSE?");

ax.legend();
```

### Conclusion for KNN Regressor:

It appears that K = 20 gives a rMSE of about 66.981. KNN Regressor won't always give the best rMSE because it relies solely on distance. Distance is not a terrible metric to predict values, but it does not capture all relationships in the data.

## Do we have training and testing sets for Un-Supervised Learning???

Recall that for supervised learning (classification and regression) we have:
$$y \sim x_1,x_2,...,x_p$$

* We have a response variable that gets a value based on the predictor variables.

* We split data into training and testing sets to see if the model predicts the response variable well.

But for un-supervised learning (pattern) we have:

$$x_1,x_2,...,x_p$$

* Un-supervised learning is about learning patterns in the data, not about measuring a outcome. Therefore, there is no response variable.

* These patterns may be discovered by  dimension reduction, clusters, or outliers. Some people believe that un-supervised learning is solely about making clusters, but this is not entirely true.

The main goal for un-supervised learning is to learn patterns from the full data set! This is not about making predicting values for a response variable!

So we do NOT have training set and testing set for un-supervised learning! We use as much data as possible!

$$d_E = \sqrt{(x_0 - x_1)^2 + (y_0 - y_1)^2} $$

(This is just a example of distance of two observations with two variables. The distance formula may use more than two variables if we have more than two predictors!)

There are a significant amount of Un-Supervised models that use the Euclidean distance.

## 🔵 Un-Supervised Model 1: Principal Components Analysis (PCA)

**WARNING:** This whole explenation is long! If you just want a quick summary, feel free to skip the *PCA process* part!

I wanted to give a big overview of what PCA does since we will be using many PCA plots throughout the whole Un-Supervised section. Lets now start the explenation!


Recall that un-supervised models try to find patterns in data.

Principal Components Analysis (PCA), is a dimension reduction technique that reduces $p$ number of dimensions/predictors into $k$ amount of dimensions. Where: $k < p$

If $p \ge 4$, we can NOT visualize the $p$ amount of predictors! But we can do $2 \le k \le 3$. We can defintely visualize $k$ amount of predictors in this case! The $k$ predictors are known as Principle Components (PCs).



FURTHERMORE, PCA tries to capture as much information of the data as possible even with smaller dimensions. PCA can not capture 100% of the variance with smaller dimensions, but it can capture about 70% to 90%! (Now you may think variance is "bad". However, in the context of PCA, variance is good because variance is information! For instance, if everyone in a building has a brown eyes, there is no variance of eye color. That data is not informative.)


You may be wondering how PCA even captures information in the first place. This is because each principal component is a linear combinations of ALL the predictors! I will be using the same notation from Supervised Regression Model 4, Principal Components Regression Model, from Section E.

$PC_{1,i} = A_1 x_{i,1} + A_2 x_{i,2} + ... + A_p x_{i,p}$

$PC_{2,i} = B_1 x_{i,1} + B_2 x_{i,2} + ... + B_p x_{i,p}$

$and~ so~ on~ for~ next~ PCs$

**Notation:**

$A,B,...$ are the scalars, they are fixed numbers. For instance, $A_1 = 2.71$ and $A_2 = -7.21$ and this stays fixed for PC1 and for all observations.

$x_{i,1},x_{i,2},...,x_{i,p}$ are the predictors! "$i = 1,2,...n$" represents the obervations.  Plug in the predictor values for observation 1, repeat for observation 2 to observation n.

---

### PCA Process:

Fair warning, the PCA process can be challenging if this is your first time learning it!

PCA uses Linear Algebra concepts such as eigenvalues, eigenvectors, and spectral decomposition.

Do not worry if you do not understand all of the PCA process today!

As long as you understand steps 1,2,3,6, and 8, then you understand the statistical parts of PCA.

**1. We start with a predictor matrix / feature matrix.**

$$X_{[n \times p]}$$

Where:

* $n$ are the number of observations.

* $p$ are the number of predictors.

**2. Standardize the data.**

Standardization formula:

$x_{i,standardized} = \frac{x_i - \bar x}{\sigma}$

We want to center the data AKA subtract each observation by the mean AND divide each observation by the standard deviation. (Also known as centering and scaling respectively.) We do this so a predictor with big numbers does not dominate the the predictors with small numbers.

The StandardScalar function from sklearn.preprocessing can automatically calculate the above formula for each observation.

All entries of the predictor matrix $X_{[n \times p]}$ becomes scaled.



**3. We calculate the sample co-variance matrix from the scaled predictor Matrix $X$.**

$$S = \frac{X^T X}{n-1}$$

Note that the population co-variance matrix is denoted by $\Sigma$, and the sample co-variance matrix is denoted by $S$.


**4. Find the eigenvectors and eigenvalues.**


### Eigenvalues:

Find eigenvalues by taking this determinant:

$$det(S-\lambda I_n) = 0$$

Where:

* S is the sample covariance matrix.

* $\lambda$ represents the possible eigenvalues of $\lambda_1$, $\lambda_2$,..., $\lambda_n$. The amount of eigenvalues depends on the size of the square matrix. EX: If we have a $(5 \times 5)$ square matrix, we will have 5 eigenvalues. Eigenvalues are scalars.

* $I_n$ is our identity matrix. A identity matrix is a $(n \times n)$ square matrix that has a value of 1 for each left diagonal element. The other entries have a value of 0.  The identity matrix size is the same as the covariance matrix size.

* Eigenvalues are the variance for their respective PCs. The first eigenvalue corresponds to PC1 and has the largest variance, then the second eigenvalue corresponds to PC2 and has the second largest variance, and so on.

### Eigenvectors:

* Find eigenvectors by solving this equation:

$$(S-\lambda_i I_n) \vec x = \vec 0$$

* This equation is known as the Null Space in linear algebra.

* The eigen vectors are $\vec x$, for each of their respective eigen values $\lambda_i$. These eigen vectors come from the null space of $(S-\lambda_i I_n)$

*  The eigenvectors show the directions of where the maximum variance is. Projecting the eigenvectors to the data produces the principal components (PCs).


**5. We use spectral decomposition from Linear Algebra and obtain our eigenvectors and eigenvalues.**

Spectral decomposition is when we want to create a rank $k$ approximation of the original  matrix. Think of it is a "lower resolution photo", while
the original matrix has the "original resolution photo".

$$S = P D P^T$$

Where:

* $S$ is the sample covariance matrix. Recall that covariance matrices are symmetric, AKA $cov(x_i,x_j) = cov(x_j,x_i)$, so the transpose is equal to itself!

* $P$ is the matrix that contains the eigenvectors $\vec U_n$, $P = [\vec U_1, \vec U_2,..., \vec U_n]$ and $P^T$ is the transposed matrix.

* $D$ is the diagonal matrix that contains eigenvalues in the left diagonal entries. All other entries in this diagonal matrix are zero.

* NOTE: $S$, $P$, $P^T$, and $D$ are all $(nxn)$ square matrices!

* After doing spectral decompositon (just multiply the matrices above), $S$ can be expressed as:

$$S = \lambda_1 \vec U_1 \vec U^T_1 + \lambda_2 \vec U_2 \vec U^T_2 + ... + \lambda_n \vec U_n \vec U^T_n$$

* The beauty of spectral decomposition is that only using some parts  simply results in a "lower resolution photo", while spectral decomposition with all its parts is the "full resolution photo".

EX:

$$S' = \lambda_1 \vec U_1 \vec U^T_1 + \lambda_2 \vec U_2 \vec U^T_2$$

Is a lower resolution version of the original data. This modified sample covariance matrix $S$ is using the structure of two PCs.  
$\lambda_1 \vec U_1 \vec U^T_1$ represents the structure of the first PC.

**6. Pick $k$ amount of eigenvalues. $k < p$ and usually we pick $k$ so that it explains  90% or more of the variance in the data.**

Depending on the $k$ amount of eigenvalues we picked, we get a matrix of eigenvectors:

$$E_{[p \times k]}$$

Where:

* $p$ is the same number of predictors as $X_{[n \times p]}$.

* $k$ are the number of eigenvectors, corresponds with the number of eigenvalues we keep.



**7. Multiply the scaled predictor matrix $X_{[n \times p]}$ and the matrix of eigenvectors $E_{[p \times k]}$.**

This is called projection. A projection is the scenario where we take the original predictor space, and then reduce the dimensions.

By matrix multiplication recall that:
 $$A_{[x \times y]} * B_{[y \times z]} = C_{[x \times z]}$$

The inner dimensions must be the same. We can see that matrices $A$ and $B$ share the same inner dimension of $y$. The result has the outer dimensions. We can see that matrix $C$ has outer dimensions of $x$ and $y$.

Therefore:

$$X_{[n \times p]} * E_{[p \times k]} = Z_{[n \times k]}$$

$X_{[n \times p]}$ was the original predictor matrix, and now $Z_{[n \times k]}$ is the PC matrix, this matrix is smaller since $k < p$.

**8. $Z_{[n \times k]}$ is our final data with PCs as variables.**

We still have $k$ amount of PCs, and by matrix multiplication, our PCs are linear combinations of the original predictors.

To visualize PCA, we can only use 2 or 3 PCs.

---

One more thing. When you go read more of the project or if you already have experience with PCA, you may wonder why the first PC always has the largest variance/eigenvalue.

Well this is because the PCA algorithm sorts the eigenvalues from largest to smallest. As a result, the corresponding eigenvectors get sorted to match the eigenvalues.

In a Medium Article, Sen states that we first have our eigenvalues of:

$$ \lambda_1, \lambda_2,
\lambda_3 ...\lambda_n$$

Then we sort the eigenvalues and redefine the position numbers so that:

$$ \lambda_1', \lambda_2', \lambda_3', ...,\lambda_n'$$

Where:

$$ \lambda_1' \ge \lambda_2' \ge \lambda_3'  ... \ge \lambda_n'$$

This is why the first eigenvalue has the biggest variance and why the first eigenvector accounts for the most variance in the data (Sen).

That was a lot of steps and linear algebra, lets get started on making the actual PCA model!

```python
# we first have to standardize the data!
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler() # does standardization

X_scaled = scaler.fit_transform(X) # we need to fit the X data, and transform it!

#print(X_scaled)
```

```python
from sklearn.decomposition import PCA # 1. from module, import class/function

pca_model = PCA(n_components= 3) # 2. initate model, get 3 PCs

X_pca = pca_model.fit_transform(X_scaled)

print(X_pca) # contains PC1, PC2, PC3
# each value in each slot is the result of a linear combination.
```

```python
# PCA under the hood walkthrough:

# our eigen vectors are here:
print(pca_model.components_)

# we have 6 predictors, and 3 components
# each PC loading for a PC is read by rows!
# so PC1 = 0.424x_1 +... + 0.309x_6

print() # make more space between outputs

# our eigen values are here:
print(pca_model.explained_variance_)

# the eigen values are out explained variance
# PC1 variance -> lambda_1 = 03.805

print()

# the proportion of variance explained by each PC:
print(pca_model.explained_variance_ratio_)

# PC1 explains 0.4409 of the data's variance,
# if we have more strong correlations, the PC1 may have a higher proportion...

# if we included all 6 PCs, the variance ratios of all 6 PCs add up to 1!
```

### How much PC's do we select?

Usually, we want to select the amount of PC's that can explain at least 0.9 proportion of the variance. We usually do this by looking at a scree plot and check the amount of PCs that has at least 0.9 cumulative variance proportion.

Sometimes, people select the number of PCs by looking at the "elbow" part of the plot. Choosing the number of PCs can be subjective and depends on the person's goal.

The issue is that if we have more than 3 PC's, visualizing the PC data will be hard. So I will still be using 2 and 3 PCs in my plots.



```python
pca_model_scree = PCA().fit(X_scaled)


plt.figure(figsize=(12,7))  # width= 12 inches, height= 7 inches

ax = plt.axes()

ax.plot(np.cumsum(pca_model_scree.explained_variance_ratio_),
        marker = "o",color = "green")
# we need to set a marker


ax.set_title("Scree plot of PCs. How much should we select?")
ax.set_xlabel("PCs / number of components")
ax.set_ylabel("Cumulative variance explained")

ax.set_xticklabels(range(0,7));
# seems like range() is EXCLUSIVE for both start and end in this scenario
# for whatever reason...
# range(A,B) is usually inclusive at start A, and exclusive at end B



```

```python
print(pca_model_scree.explained_variance_ratio_)
# our 6 PCs, and how much proportional variance each PC explains
```

Based on the scree plot, I would select 4 components since it is the least amount of components that it is close to explaining 90% of the variance.

The "elbow" is not obvious here, 3 components looks like a possible elbow spot.

Again, for visualization, I will select 2 and 3 components.

## 🔵 Un-Supervised Model 2: K Means Clustering


K-Means is a simple clustering algorithm that has 4 steps:

* Step 1: We first CHOOSE the number of K clusters. A good starting point to choosing the number of K is looking at a scree plot of how much error each K clusters has. After choosing K clusters, the algorithim randomly put cluster centers in the middle region of the observations. Lastly, we should standardize our predictors, since K means will use the distance formula. We do NOT want any predictors dominating the distance! Make sure the predictors are continuous, because K means does not use categorical predictors.

* Step 2: This is formally known as the "Expectation Step". Assign each observation to the nearest cluster center. We will use the euclidean distance to determine this.

Recall that the eucldiean distance is just the distance formula, here is a example of a distance formula using variables x and y.

$$d_E = \sqrt{(x_1-x_0)^2 + (y_1-y_0)^2}$$

Now, here is  the distance formula of multiple predictors (We do NOT have a y variable since un-supervised learning only has predictor variables, which are x variables!):

$$d_{ik} = \sqrt{(x_{i1} - c_{k1})^2 + (x_{i2} -c_{k2})^2 + ... + (x_{ip} - c_{kp})^2}$$

Given $1 \le j \le p$:

* $x_{ij}$ is the value of the $j-th$ predictor for observation $i$.

* $c_{kj}$ is the value of the $j-th$ coordinate for the cluster center $k$.



Repeat for clusters 2 to K. Select the shortest distance and assign observation 1 to that cluster.

* Step 3: This is formally known as the "Maximization Step". Take the mean of each observations in  cluster 1. Then move the cluster 1 center to the mean of the points that belong to cluster 1. Do the same thing for cluster 2 to K.


* Step 4: Repeat Step 2 and Step 3 until the cluster centers do not move.

---
### How do we visualize K Means if we have too many predictors?

* HOWEVER, we have one problem. How do we visualize 4 or more predictors in a plot? We can not do this. So are we cooked???

* NO! We have PCA to save the day! We can use 2 PCs to visualize in 2d plots, or 3 PCs in 3d plots!

* But why can we use PCs for K means? Recall that Principal Components (PC's) contain information about ALL our predictors! Recall that PCs are just linear combinations of all the predictors! And remember, the first few PCs contain the most information!

We can use these 2 or 3 PCs, and make a K means for a 2d or 3d plot respectively!

And since our PCA model was already standardized, we do not have to standardize again for K means!

* NOW, I will also make a K means of the regular data set. Yes, K means can not be visualized with 4 or more predictors, but we can still get cluster labels for the data!

---

### How do we choose the amount of K clusters?

Since we have to set the number of K means in order to start the model, what is the optimal number?

We can make a Inertia based scree plot and find the "elbow". Inertia for K Means is just measuring the sum of squared distances of observations to the cluster center. A high inertia means that the points are not close to the center. A low inertia means that the points are close to the center.

BUT we have to be careful! Making too much clusters will result in extremely low inertia values. We may think that is great, but this leads to overfitting. Furthermore, if we have too much clusters, the clusters won't make sense logically.


### K Means on 2D PCA

```python
from sklearn.cluster import KMeans

# Our PCs came from standardized data, so no need to use StandardScalar!

# Lets make a K means on 2D PCA

X_kmeans_2d = pca_df[["PC1", "PC2"]]
```

```python
# lets make a scree plot
inertia_list = [] # initalize a empty list

K_values = range(1,16) # K values: 1 to 15, end value of 16 is EXCLUSIVE

for k in K_values:
  kmeans_model = KMeans(n_clusters = k, random_state = 1)
  kmeans_model.fit(X_kmeans_2d)
  inertia_list.append(kmeans_model.inertia_)


plt.figure(figsize=(15,8))

# "go-"= g -> green, o -> circle marker, - -> line!
plt.plot(K_values, inertia_list, "go-")

plt.xlabel("Number of Clusters")
plt.ylabel("Inertia (the sum of squared distance)")
plt.title("Scree plot of Inertia. Can we find the elbow/optimal number of clusters?");
```

From this plot, I would think that the most significant elbow is at 7 clusters.

It is true that there are elbows at 2,3, and 4. But the goal is to find the elbow that gives the lowest amount of Inertia while avoiding overfitting.

We should note that finding the elbow can be subjective at times. We also have  the YellowBrick library that can gives us a more objective approach on finding the optimal number of clusters.  

```python
from yellowbrick.cluster import KElbowVisualizer

kmeans_model = KMeans(random_state = 1)

# yellow brick was made from matplotlib, so we can make the plot bigger!
plt.figure(figsize = (15,8))

# metric = "distortion" gives inertia scores. timings = False removes time plot
scree_plot = KElbowVisualizer(kmeans_model, k = (1,16),
                              metric = "distortion", timings = False)

scree_plot.fit(X_kmeans_2d)

scree_plot.show();
```

Interesting, the Yellow Brick visualizer says that we should use 4 clusters.

Yellow Brick is a general rule, even if it suggests to use 4 clusters, if we think that the data uses 7 clusters, we can do that! We can NOT just blindly follow what a computer out put tells us!

Lets try 4 clusters and visualize the 2D PCA plot! Next, lets see if 7 clusters is any better or worse.

```python
kmeans_model = KMeans(n_clusters = 4, random_state = 1)

kmeans_model.fit(X_kmeans_2d)

# ALERT: For K means, predict() uses the same X matrix!
cluster_labels = kmeans_model.predict(X_kmeans_2d)
print("The first 25 observations and their clusters:",cluster_labels[0:25])

# print first 25 clusters, index 0 to 24.

print("Same cluster labels as above:", kmeans_model.labels_[0:25])

print(sum((cluster_labels == kmeans_model.labels_))/len(kmeans_model.labels_))
# The portion of clusters and kmeans_model.labels_ equaling each other is 1!

print()

centers = kmeans_model.cluster_centers_
print("Our cluster coordinates:",centers) # coordinates of the clusters
```

## 🔵 Un-Supervised Model 3: Spectral Clustering

Spectral Clustering is actually similar to K Means. The main difference is that K Means can not perform well in non-linear data. Spectral Clustering can save the day and help us cluster non-linear data.

To start off, lets define what linear data and non-linear data is:

* Linear data: Any data that can be approximated with a straight line or hyperplane.

EX: $y = mx+b$

* Non-linear data: Any data that can NOT be approximated with a straight line or hyperplane. Instead, it needs to be approximated with something like a polynomial.

EX: $y = ax^3 + b* cos(x) + c$

---

So lets imagine we have two curved shape data points.  These curves should belong in their own two clusters. But K Means might make clusters where a majority of of the points are from curve A, and some points are from curve B. This is one weakness of K Means, it is mainly effective in linear data. We need a new clustering method like Spectral Clustering.


A online python data science handbook states that Spectral Clustering is similar to SVM. "This situation is familiar to SVM, where we used a kernel transformation to project the data into a higher dimension. In higher dimensions, a linear separation is possible. Spectral Clustering uses a similar trick to allow K-Means to discover non-linear boundaries" (Jake VanderPlas, *In Depth: K-Means Clustering*).

To be clear, Spectral Clustering does not use the same method as SVM. Spectral Clustering actually uses a smaller amount of dimensions (variables) since it uses spectral decomposition (recall that PCA used spectral decomposition).

A GeeksForGeeks article states "It uses eigenvalues and eigenvectors of the data matrix to forecast the data into lower dimensions space to cluster the data points" (“Spectral Clustering in Machine Learning”).


---
### Process for Spectral Clustering


**1. Create lines between nearby points.**

Firstly, the data is examined. Any observations/ data points that are near each other are connected. If observations A,B, and C are near each other, they have 3 lines that connect each other. When two observations are closer, they get bigger similarity scores.
(In formal terms, the observations are called nodes, and the lines are called edges.)

If observation X and observation Y are far away, no line is made.

The first step is how Spectral Clustering performs well on non-linear data. The points will only connect with close points, they will not connect with far away points. On the other hand, for K Means, there are cases where far away points can be contained in a cluster...


How do we determine if observations are close or not? One way to determine if observations are close is by our old buddy, K Nearest Neighbors (KNN). In fact, in the SpectralCluster() model, there is parameter option called: affinity = "nearest_neighbors".

Once we create the lines, a Affinity Matrix ($A$) is made.



**2. Calculate the Laplacian Matrix $L$.**

It is simply:

$$L = D - A$$

Where:

* $A$ is the affinity matrix that we made in step 1.

* $D$ is called a degree matrix. It is defined as: $$D_{ij} =
 \begin{cases}
 d_i, & \text{if } i = j \\
 0, & \text{if } i \ne j \\
 \end{cases}$$

And: $d_i = \text{the sum of all column entries}$

Therefore, $D$ is basically a diagonal matrix. HOWEVER, it is NOT a identity matrix! A identity matrix is a matrix with left diagonal of ONLY 1s while the other entries are 0!

We denote identity matrix $I_n$ for a $n \times n$ matrix. Again, it is NOT the same thing as $D$! This is for a $3 \times 3$ identity matrix:

$$I_3 =
\begin{bmatrix}
1 & 0 & 0 \\
0 & 1 & 0 \\
0 & 0 & 1
\end{bmatrix}$$



* Here is a example of calculating $D$ and $L$ for 3 observations:

Suppose after step 1, the Affinity Matrix $A$ is:

$$A =
\begin{bmatrix}
0 & 5 & 3 \\
5 & 0 & 2 \\
3 & 2 & 0
\end{bmatrix}$$

Notice that the affinty matrix never has zeroes in the diagonal. This is because the diagonal element represents a observation COMPARING itself.

For example $d_{11}$ is observation 1 compared to observation 1. But that is pointless. We want to see the distances of other points. So we might as well give that element a value of 0.   

Meanwhile $d_{12}$ is row 1, column 2. This compares observation 1 with observation 2. A weight score of 5 is pretty high, so observations 1 and 2 must be close.

You may notice that $d_{12} = d_{21} = 5$. Since observations 1 and 2 always have the same euclidean distance (distance formula), these two entries will have the same value. We call this symmetry. In Linear Algebra, we call this scenario a symmetric matrix if for all $i$ and $j$, $d_{ij} = d_{ji}$. As a fun fact, if a matrix is symmetric, then $A = A^T$. The tranpose will be the same as regular matrix since the tranpose switches elements across each other, but the diagonal elements are not switched.

Sometimes in Spectral Clustering you may see a matrix have 1's and 0's. This is called a adjacency matrix (Chatterjee).  1's means that the two observations are similar, 0's means the two observations are not similar or is a diagonal element. This is another valid way to do Spectral Clustering.


But I will use weighted version of affinity matrix where higher numbers mean the two observations are really similar. This way, we will know the scale of how similar the observations are.


Then the Diagonal Matrix is:

$$D =
\begin{bmatrix}
8 & 0 & 0 \\
0 & 7 & 0 \\
0 & 0 & 5
\end{bmatrix}$$

Therefore the Laplacian Matrix L is:

$$L = D - A =
\begin{bmatrix}
8 & -5 & -3 \\
-5 & 7 & -2 \\
-3 & -2 & 2
\end{bmatrix}$$


**3. Do Spectral Decomposition on Laplcian Matrix $L$.**

For a refresher on spectral decomposition, refer the Un-Supervised Model 1: PCA section.

**4. Select the eigenvectors**

Here is the interesting part. Recall that in PCA that we select the largest $k$ amount of eigenvalues.

For Spectral Clustering, we select the SMALLEST $k$ amount of eigenvalues. This is NOT a typo!

A GeeksForGeeks article states "Notably, the eigenvectors associated with smaller eigenvalues often contain valuable insights regarding the underlying structure of clusters"(“Spectral Clustering: For Beginners”).

These smallest eigenvalues correspond with their following eigenvectors. The following eigenvectors are called Spectral Embedding.

**5. Do K-Means on Spectral Embedding**

Now we do K-Means on the Spectral Embedding.
Our clusters made by K-Means here will give more reliable clusters.


---
### One caution.

Now, there is a possibility that Spectral Clustering will not do well. After all, our PCA plots do not show any curvy parts or grouped data. The PCA plots just show a cloud points.

But maybe the data is non-linear in the original 6 predictors. The correlation plot in Section C, Pattern EDA had no correlation value that was greater than 0.5 or less than -0.5, so perhaps Spectral Clustering will do well here.

---
### Silhouette Plots

Lets first start by making silhouette plots. We will make multiple silhouette plots that each represents a eigenvalue number. The silhouette plot creates values of -1 to 1. Any point near 1 indicate that observations are matched well with its cluster. It also implies that the clusters are seperated well.  Meanwhile, scores near -1 signal that the observation may be in a wrong cluster (The scikit-yb developers).

The number of eigenvalues and number of clusters with the highest silhouette score will be selected.


```python
# Lets first visualize the silhoutte plot and see how much
# eigenvalues and clusters we want to use.

# To get our spectral embedding, we need to use the following import:
from sklearn.manifold import SpectralEmbedding
from yellowbrick.cluster import KElbowVisualizer



# plot 1: 2 eigenvalues

plt.figure(figsize=(12,7))

# n_components = number eigenvalues

spectral_embedding = SpectralEmbedding(n_components = 2, affinity = "nearest_neighbors",
                                       random_state = 1)

# embedding contains our eigenvectors
embedding = spectral_embedding.fit_transform(X)
#print(embedding)


kmeans_for_sc = KMeans()

scree_plot = KElbowVisualizer(kmeans_for_sc, k = (2,10), metric = "silhouette",
                              timings = False)

scree_plot.fit(embedding)

plt.title("Scree plot on 2 eigenvalues");

```

```python
# plot 2: 3 eigenvalues

plt.figure(figsize=(12,7))

# n_components = number eigenvalues

spectral_embedding = SpectralEmbedding(n_components = 3, affinity = "nearest_neighbors",
                                       random_state = 1)

embedding = spectral_embedding.fit_transform(X)
#print(embedding)


kmeans_for_sc = KMeans()

scree_plot = KElbowVisualizer(kmeans_for_sc, k = (2,10), metric = "silhouette",
                              timings = False)

scree_plot.fit(embedding)

plt.title("Scree plot on 3 eigenvalues");
```

```python
# plot 3: 4 eigenvalues

plt.figure(figsize=(12,7))


# n_components = number eigenvalues

spectral_embedding = SpectralEmbedding(n_components = 4, affinity = "nearest_neighbors",
                                       random_state = 1)

embedding = spectral_embedding.fit_transform(X)
#print(embedding)


kmeans_for_sc = KMeans()

scree_plot = KElbowVisualizer(kmeans_for_sc, k = (2,10), metric = "silhouette",
                              timings = False)

scree_plot.fit(embedding)

plt.title("Scree plot on 4 eigenvalues");

```

```python
# plot 4: 5 eigenvalues


plt.figure(figsize=(12,7))


# n_components = number eigenvalues

spectral_embedding = SpectralEmbedding(n_components = 5, affinity = "nearest_neighbors",
                                       random_state = 1)

embedding = spectral_embedding.fit_transform(X)
#print(embedding)


kmeans_for_sc = KMeans()

scree_plot = KElbowVisualizer(kmeans_for_sc, k = (2,10), metric = "silhouette",
                              timings = False)

scree_plot.fit(embedding)

plt.title("Scree plot on 5 eigenvalues");

```

* It appears that 2 eigenvalues, with 3 clusters gives the best silhouette score of about 0.590!

* It also appears that the more eigenvalues we use, the less the best possible silhouette score becomes.

* Notice that we did not start at 1 eigenvalue because we need at least 2 eigenvalues to compare how similar the observations are. We need at least 2 eigenvalues because we use their corresponding eigenvectors and compare them with each other.

It is similar to how we need points $x$ and $y$ to compare how similar $x$ and $y$ are. Having just point $x$ will not give us any information and how it is similar to other points.

```python
from sklearn.cluster import SpectralClustering

# SpectralClustering will automatically pick the best amount of eigenvalues
# We still need it to pick 3 clusters though.
# let n_clusters = 3

sc_model = SpectralClustering(n_clusters = 3, affinity = "nearest_neighbors",
                              assign_labels = "kmeans", random_state = 1)

# ALERT: the below code does NOT exist for SpectralCluster()!
# embedding = sc_model.fit_transform(X)

# Only SpectralEmbedding() has fit_transform()!


# both X and pca_df[["PC1","PC2"]] will give weird graphs!
#
cluster_labels = sc_model.fit_predict(X)
cluster_labels


# spectral clustering does not have a cluster centroid option
```

## 🔵 Un-Supervised Model 4: t-distributed Stochastic Neighbor Embedding (t-SNE)

Lets take a break from clustering models, and examine some more dimension reduction  models.


* PCA is not the only dimension (variable) reduction technique out there. We have a whole branch of models called Manifold Learning methods that can reduce dimensions as well.


* A GeeksForGeeks article states "Manifold learning is a dimensionality reduction techniques which turns complex, high-dimensional data into simpler form while keeping important patterns and features. It works well when the data has curved or non-linear shapes that simple methods like PCA can’t handle" ("Comparison of Manifold Learning Methods").

t-SNE (t-distributed Stochastic Neighbor Embedding) is one of those Manifold Learning models.

* One benefit of t-SNE is that it does double duty. It will turn our 6 dimensional Pokemon data into a 2 dimensional space AND place similar observations together.

If Magikarp has low individual base stats, then t-SNE would group it with other weak Pokemon such as Weedle.


---
### Process for t-SNE


**Step 1. Use Normal distribution:**

First, t-SNE tries to find pairwise similarities between observations. It actually does this with the Normal distribution when we have the original amount of dimensions.

If we use point 1 and place it at the center of the Normal distribution, and the points closest to point 1 would be inside the "big chunk part" of the Normal distribution. t-SNE would think that these points are related. Meanwhile, any points that are far away from point 1 lie in the tails of the Normal distribution.

Repeat for points 2 to n.

After this is completed, t-SNE makes a similarity matrix of all the points (Starmer). Each entry has a similarity score of 0 to 1, where any numbers near 1 means the two points are really similar. The diagonal entries will always be 1 since a point compared with itself will always be the same.   

**Step 2. Use t-distribution**

* As a fun fact from statistics, the t-distribution is related to the Normal distribution. In statistics, we use the t-distribution when we only know the sample variance ($s^2$). HOWEVER, this is NOT the reason why t-SNE uses the t-distribution!

* Furthermore, the t-distribution is known to be wider and have bigger tails than the Normal distribution. The main reason t-SNE uses the t-distribution is because of the bigger tails. I will touch on this after some sentences.  

Next, t-SNE tries to compress the multiple dimensions into 2 dimensions (3 dimensions if you want to see a 3D version). t-SNE tries to find pairwise distances in the 2D space this time. This time, we use a t-distribution, and repeat the same process that point 1 went through for the Normal distribution. Then repeat again for points 2 to n.  

**Step 3. Move the points**

Now we have our points in 2D. But t-SNE wants the points to be in the similar positions as the similarity matrix in Step 1 (Starmer).

**Why the t-distribution?**

If the Normal distribution was used in step 2, the clusters will just clump up and be harder to see (Starmer). If you have ever seen a Normal distribution before, you may have noticed that tails can be tiny.

This is why we use the t-distribution. It's probability distribution function (PDF) is designed so that the tails are bigger. This gives room for clusters to spread apart!

Now our points should be finally ready to see!  



---
### Some warnings!

First Warning:

t-SNE is usually used when we have extremely high amounts of dimensions, such as 100 predictors. t-SNE might give similar results to our other Un-Supervised models since we only have 6 predictors.

Second Warning:

t-SNE is only a dimension reduction technique like PCA. It will not find clusters for us. So we need to use a clustering model such as K-Means.

Third Warning:

We should just reuse the K-Means cluster labels from PCA or original data when visualizing t-SNE. We should NOT calculate K-Means on the t-SNE and make new cluster labels. Why?

Because the t-SNE space is non linear, so doing K-Means calculations on it won't bring much meaning. Meanwhile, the PCA space is linear, so doing K-Means calculations on it makes sense.

## 🔵 Un-Supervised Model 5: Multi-Dimensional scaling (MDS)

Multi-Dimensional Scaling is another dimension reduction like PCA and t-SNE. But it does not use variance as it's criteria for dimension reduction. It uses pairwise distances instead.

A GeeksForGeeks article states that, "MDS is a dimensionality reduction method that attempts to preserve pairwise distances between points while projecting them into a lower-dimensional space" (“Comparison of Manifold Learning Methods").

---
### MDS Process

There may be multiple ways that MDS creates its new dimensions. Since Scikit Learn uses a numerical optimization method, I will be describing the optimization method (“MDS Using Scikit Learn”).


* Step 1: Calculate Pairwise Distances


Annnnndddddd our math superhero comes to save the day again. We use the Euclidean Distance to find the pairwise distances. And his alias is the distance formula.

For the pairwise distnace of two points $x_i$ and $x_j$, we have:

$$d_{ij} = \sqrt{ \sum^m_{i=1} (x_{ik} - x_{jk})^2}$$

Where:

* $m$ is the amount of predictors we have. In this case, $m=6$ since we have 6 predictors of individual Pokemon stats: Attack, Defense, Sp. Attack, Sp. Defense, HP, Speed

* $d_{ij}$ is the pairwise distance of  $x_i$ and $x_j$.

Next, we create a Distance Matrix called $D$ (Jake VanderPlas, *In Depth: Manifold Learning*). The entries of $D$ are filled with the pair wise distances $d_{ij}$!



* Step 2: Random intialization

In the 2D or 3D space, the algorithim randomly assigns coordinates for each point.

Then compute the pairwise distances of these guessed point. Again, we want to call our friend, the Euclidean Distance:

$$\widehat d_{ij} = \sqrt{ \sum^m_{i=1} (\widehat x_{ik} - \widehat x_{jk})^2}$$


* Step 3: The Stress Function

To measure how well the guesses are, we use the stress function. The MDS GeeksForGeeks website states that the stress function is:

$$stress = \sqrt{ \frac{1}{2n^2} \sum^n_{i=1} \sum^n_{j=1} (d_{ij} - \widehat d_{ij})^2}$$



The lower the stress function, the more better MDS is estimating. Just like in real life, we want less stress to enjoy life more!

I should note that there are different stress functions, but the goal is the same. Less stress shows MDS is good at giving lower dimensions.  

* Step 4: Update the coordinates

After calculating the stress function in step 3, update the coordinates so that we can get a lower stress value. Keep repeating until a certain number of times.


* Step 5: The final coordinates.

Use the final 2D or 3D coordinates that give a lowest stress value.

---

## 🔵 Un-Supervised Model 6: Locally Linear Embedding (LLE)

Locally Linear Embedding is another dimension reduction technique like PCA, t-SNE, and MDS.

LLE has a big advantage that PCA and MDS does not have. LLE can do dimension reduction in NON-linear data more effectively than PCA and MDS.

This is because LLE only calculates pairwise distances of nearby points.

**EASY EXPLENATION:** LLE projects from higher dimensions into lower dimensions (2D or 3D) while preserving local relationships between points. PCA also projects from higher dimensions into lower dimensions. BUT PCA is different since it considers the variances of each PC to project down.


* For MDS, it literally calculates the pairwise distances between ALL points in the data.

* For LLE, it would just calculate pairwise distances of the 100 nearest points. (We can adjust the amount of points by using "n_neighbors = " if we think 100 is not good.) Overall, LLE only considers local structure of the data.

This makes LLE effective at capturing non-linear structure in data. This is similar to how in Spectral Clustering, each point tries to find its K nearest points/neighbors.  

---
### LLE Process


* Step 1: Select the number of neighbors.

We usually select 100 neighbors for each point, but this can change depending on the data set size and situation. For this process explanation, lets assume 100 neighbors is good enough.


* Step 2: Make Linear Regression for each point

For point 1, we use its 100 neighbors as weights or slope coefficients. We will be making a linear regression where these neighbors act as predictor variables. We want to make this linear regression so that is close to point 1's value (“LLE in Machine Learning”).

The idea behind making linear regression is that each neighbor contains information that can help predict point 1.

Repeat for points 2 to n.

* Step 3: Create the Weight matrix, $W$. Also create matrix $M$

The Weight matrix $W$ contains the weights of each point's neighbors. That's it.


Next, we need to calculate matrix $M$.

According to the LLE GeeksForGeeks website, we need to do the following matrix operations:

$$M = (I-W)^T (I-W)$$

Where:

* $I$ is the identity matrix.

* $W$ is the weight matrix.

We need to do this matrix calculation to prepare for the eigenvalue problem (“LLE in Machine Learning”).


* Step 4: Do Spectral Decomposition on $M$



We then do Spectral Decomposition on $M$. Refer to Un-Supervised Model 1, PCA Process, step 5 if you need a refresher on Spectral Decomposition.

If we want to make a 2D LLE plot, we only use 2 eigenvectors. If we want to make a 3D LLE plot, we simply use 3 eigenvectors. Recall that this is how we made 2D PCA and 3D PCA plots!

* Step 5: Get the final coordinates.

The final coordinates are simply the 2 eigenvectors for a 2D LLE plot, and the 3 eigenvectors for a 3D LLE plot. That's it.


---
### Lets use a legendary Pokemon marker for the plots!

For the 2D and 3D LLE plots, lets use the legendary Pokemon marker. I want to see how it compares to the 2D and 3D PCA plots' legendary Pokemon marker.





```python
from sklearn.manifold import LocallyLinearEmbedding

LLE_model = LocallyLinearEmbedding(n_neighbors = 100, n_components = 2,
                                   method = "modified", eigen_solver = "dense")

X_lle = LLE_model.fit_transform(X)


lle_df = pd.DataFrame(X_lle, columns = ["Component 1", "Component 2"])
print(lle_df)

```

## 🔵 Un-Supervised Model 7: DBSCAN

Ok lets jump back into clustering again!

* DBSCAN stands for "Density-Based Spatial Clustering of Applications and Noise".

* DBSCAN is a easy model to understand. It clusters points based on density. Its not about where the points are located, its about the distance between points. DBSCAN does well in clustering groups of points that has irregular shapes.  

* And guess what? We have our faithful friend, the Euclidean distance to measure the distance between points. We simply use the distance formula from middle school.

* As a bonus deal, DBSCAN can find outliers as well. If a point is too far away from other points, DBSCAN will just label that point as a outlier.  Outliers as labeled as "-1".

* Additionally, DBSCAN does not require a pre-determined number of clusters. Recall that K Means and Spectral Clustering needs the user to input the number of clusters to use!

* DBSCAN does require parameters of eps and min_samples.

eps stands for epsilon ($\epsilon$). eps is the max distance between two points to be considered neighbors.

min_samples is the lowest number of observations needed to make a cluster.

* One strength of DBSCAN is that we can keep changing the parameters based on how the 2D and 3D plots look like.

* As a warning, DBSCAN defintely needs to use standardized data! But our PCA data is already standardized, so we do not have to standardize a second time.



```python
from sklearn.cluster import DBSCAN

# by default, we set eps = 0.5
# however, DBSCAN uses experimenation.
# we adjust the eps value, then adjust its value to get clusters
# that are more understandable


# eps = 0.7 -> max distance between two points to be considered neighbors,
# in order to make a cluster

# min_samples = 5 -> A cluster needs at least 5 observations

dbscan_model = DBSCAN(eps = 0.7, min_samples = 5)

# ALERT: do not use X! recall that X is the regular data set of 6 predictors!
# USE X_pca!!! it contains our three PCs! Now, we want to make use 2 PCs
# for a 2D PC plot!

# set up 2D PCA data

dbscan_2d_df = pca_df[["PC1","PC2"]]


dbscan_model.fit(dbscan_2d_df)

dbscan_labels = dbscan_model.labels_
print(dbscan_labels)

print("Our outliers and clusters:", np.unique(dbscan_labels))
# -1 means outlier, cluster 0 and cluster 1 are our two clusters!

# lets store the dbscan labels in the pca_df
pca_df["dbscan_labels_2d"] = dbscan_labels
```

### 2D DBSCAN plot

```python

fig = plt.figure(figsize = (15,8))

sns.scatterplot(pca_df, x = "PC1", y = "PC2", hue = dbscan_labels,
                s = 100)

plt.title("DBSCAN 2D plot");

```

For this Pokemon data set, DBSCAN gave us the most understandable clusters.

* The middle cloud that contains the most points is the first cluster.  

* The "small island" of points is the second cluster.

* Other points that are far away from the clusters  are labeled as outliers.

### Why are we not taking summary statistics of each DBSCAN cluster?

You may have noticed that we took the mean values of all Pokemon stats for K Means cluster.

But for DBSCAN, taking the mean values does not really make sense. Firstly the clusters an have irregular shapes, so taking the mean might not represent the cluster well.

Furthermore, taking the mean of the outliers will just give a central value of the outliers. But outliers are extreme values, they are not meant to be summarized with a central value.

### 3D DBSCAN plot

## 🔵 Un-Supervised Model 8: Gaussian Mixture Model (GMM)

Recall our Un-Supervised Model 2, K Means. One problem with K Means is that it can not do well if the cluster shape is irregular.

Just like DBSCAN, the Gaussian Mixture Model (GMM) can help us make clusters when there is a irregular shape. The word Gaussian just means Normal distribution, that's it.

And our good old friend Pam from *The Office* tells us the clustering process of K Means and GMM are the same picture. (Okay, maybe not the same picture, but mostly similar.)

---
### GMM process:

* Step 1: We first CHOOSE the number of K normal distributions.


We will choose the number of K components by checking their AIC (Akaike Information Criterion) and BIC (Bayesian Information Criterion) score. The lower the AIC and BIC score, the better! The K components with the lowest AIC and BIC score will be selected.


* Step 2: This is formally known as the "Expectation Step".   For each point, find the probablities of belonging in the first normal distribution, the second normal distribution, and so on until the last normal distribution.

* Step 3: This is formally known as the "Maximization Step". Update each normal distributions' mean value, covariance value, and weights.

* Step 4: Repeat step 2 and step 3 until the mean value, covariance value, and weights stop changing ("Gaussian Mixture Model").

* Step 5: Point 1 has probabilities of belonging in a cluster. The cluster with the highest probability is the cluster label that is assigned to point 1. Repeat for points 2 to n.


---
### Covariance type

I mentioned that GMM can make clusters when the shapes of some clusters are irregular.

GMM does this by it's covariance matrix. If we use: covariance_type = "full", each cluster will have it's own covariance matrix, and try to adjust to based on that cluster's shape.



---
### GMM has many uses other than clustering!

Here are some uses of GMM:

1. Modeling density of data: If we use many K components, we can use it to model the overall density of the data (VanderPlas, *In Depth: Gaussian Mixture Models*).

2. Data generation: Once we model the overall density of the data, we can use those normal distributions to make synthetic data.


3. Outlier and Anomaly detection: GMMs can be used to find outliers. Any points with low probabilities are considered outliers. (I will talk about a more specific outlier detection model for Un-Supervised Model 10.)

There are many other uses for GMM, but lets not get too carried away. Lets see GMMs work in the Pokedex data.



```python
from sklearn.mixture import GaussianMixture

# make empty lists:
aic_list = []
bic_list = []


gmm_df = pca_df[["PC1","PC2"]]
# print(gmm_df)


for x in range(1,21):
  gmm_model = GaussianMixture(n_components = x, covariance_type = "full",
                              random_state = 1)
  gmm_model.fit(gmm_df)
  aic_score = gmm_model.aic(gmm_df)
  bic_score = gmm_model.bic(gmm_df)
  aic_list.append(aic_score)
  bic_list.append(bic_score)


fig = plt.figure(figsize = (15,8))

# ALERT: Instead of using range(1,21), lets use np.arange(1,11).
# np.arange() is faster since it is a numpy array

plt.plot(np.arange(1,21), aic_list, color = "blue", marker = "o",
         markersize = 10, linewidth = 2, label = "AIC score")

plt.plot(np.arange(1,21), bic_list, color = "red", marker = "o",
         markersize = 10, linewidth = 2, label = "BIC score")


# ALERT: we need to do + 1 to account for the fact the x-axis starts at 0!
lowest_aic_spot = np.argmin(aic_list) + 1
lowest_bic_spot = np.argmin(bic_list) + 1

# ALERT: We have to do - 1 in index to account for the + 1 above
plt.scatter(lowest_aic_spot, aic_list[lowest_aic_spot - 1], s = 300,
            marker = "o", facecolor = "none",
            edgecolor = "hotpink", linewidth = 3,
            label = "This amount of K gives lowest AIC score.")

# ALERT: We have to do - 1 in index to account for the + 1 above
plt.scatter(lowest_bic_spot, bic_list[lowest_bic_spot - 1], s = 300,
            marker = "o", facecolor = "none",
            edgecolor = "darkgreen", linewidth = 3,
            label = "This amount of K gives lowest BIC score.")



plt.title("What value of K do we select based on the AIC and BIC scores? (2D plot)")
plt.xlabel("The number of K Normal Distributions")
plt.ylabel("The AIC and BIC scores")



plt.legend();


# NOTE to self:
# maybe doing "ax = plt.axes()" can help with making better indices!


```

```python
print("Lowest AIC K location:", lowest_aic_spot )
print("Lowest AIC K value:", round(aic_list[lowest_aic_spot],3))

print()

print("Lowest BIC K location:", lowest_bic_spot )
print("Lowest BIC K value:", round(bic_list[lowest_bic_spot],3))



```

### What do we do if the K locations for lowest AIC and lowest BIC are different?

We can see that the location of K for the lowest AIC is K = 10.

We can also see that the location of K for the lowest BIC is K = 2.

So how much K do we select?

Some people suggest to do the compromise method. This is where we try to find the balance of the low AIC and BIC scores. We usually find this in "elbows".

The Medium article on AIC and BIC scores gives a good summary for choosing the lowest AIC or lowest BIC model:

"Choose the model with the lowest AIC if you’re focusing on fit with slight flexibility in complexity" (Diaries).

"Choose the model with the lowest BIC if you’re aiming for simplicity, especially with larger datasets" (Diaries).

---
### Lowest BIC is better for interpretation

I prefer the model with the lowest BIC score. More simplicity translates to more interpretation. 2 clusters would be easy to interpret!

The lowest AIC may fit the data better, but 10 clusters seems complicated. What exactly do the 10 clusters represent? I imagine that you the reader would not want to see 10 clusters and try to figure out what it represents.

## 🔵 Un-Supervised Model 9: Heirarchical Agglomerative Clustering (HAC)

HAC is a simple algorithm, here are the steps:

1. Each point starts as its own cluster.

2. Each point tries to find the closest point. These two points then make a cluster.

And how does a point try to find it's closest point? By our trusted measurement, the Euclidean Distance! (There are other measurements such as the Manhattan Distance. But for the sake of simplicity, lets use the Euclidean Distance.)

As a refresher, here is the Euclidean Distance Formula. Remember that he loves his code name, the distance formula from middle school:

$$d(A,B) = \sqrt{(x_{1A} - x_{1B})^2 + (x_{2A} - x_{2B})^2}$$

Where:

* $A$ is observation A, and $B$ is observation B

* $x_{1A}$ is observation A with predictor 1

* $x_{2A}$ is observation A with predictor 2


For $p$ amount of predictors, the Distance Formula is:

$$d(A,B) = \sqrt{ \sum^p_{i = 1}(x_{iA} - x_{iB})^2}$$  


3. The clusters then try to find another cluster to make a bigger cluster. But we have a problem. When two clusters have multiple points, we need a method to define distance between clusters.

This method is called linkage. One linkage type is called single linkage. This is when we measure the minimum distance between the points of cluster A and B.   


4. We repeat this process until we make a giant clusters or we make a final cluster that contains all the data.

Normally we only make big clusters since making a final cluster will not give us insights.

Selecting the number of big clusters is subjective, we can determine any amount. One way to visualize how much clusters we want to select is by making a dendrogram. I will give a example of a dendrogram later on.




---

### Linkages:

I mentioned in step 3 that Agglormerative clustering uses a method called Linkage. Single Linkage is not the only Linkage type.

Here are the 4 common linkage types:


1. **Single Linkage**

I mentioned this already, but lets dive deeper into what Single Linkage does.

This is when we try to use the closest points of cluster A and cluster B. If they give the lowest minimum distance out of the other clusters, then cluster A and cluster B merges and becomes a cluster AB.

It is true that Single Linkage is easy to understand, but it is not a robust linkage.

A GeeksForGeeks article states " This method creates long, chain-like clusters because it is sensitive to outliers"
(“Types of Linkages”).

Perhaps we should consider other linkages.


2. **Complete Linkage**

This is when we try to use the farthest points of cluster A and cluster B. If they give the lowest maximum distance out of the other clusters, then cluster A and cluster B merges and becomes a cluster AB.


Complete Linkage may seem counter intuitive. Why in the world would we want to find the farthest distances of two clusters?

Firstly, even if we find the farthest distances, we still want to find the lowest farthest distance of all cluster pairs. So lowest distance is still considered even if the criteria is about being the farthest.

Secondly, Complete Linkage  avoids the chained clusters problem from Single Linkage.



3. **Average Linkage**

As you guessed from this name, we use the average/mean.

For clusters A and B, we take pairwise distances of all points in A and B.

Even if cluster A has 10 points and cluster B has 3 points, this still works since we take pairwise distances. We find the distance of point 1 from A and point 1 from B, point 2 from A and point 1 from B, all the way to point 10 from A, and point 3 from B.

We then take the average of all these pairwise comparisons. In the above example, we are bound to have 30 pair wise distances, so add up these 30 pair wise distances, and divide it by 30.

Repeat this for other cluster combinations, such as cluster A to cluster C, or cluster Y to cluster Z.

The smallest average distance between all clusters will be selected, and those 2 clusters will merge. If cluster A and cluster B had the smallest average distance, then we have cluster AB.

Average Linkage is a solid Linkage choice. The GeeksForGeeks website says "It usually results in clusters that are moderately compact" (“Types of Linkages”).


4. **Ward Linkage**

Ward Linkage might actually be the most simple Linkage.

Firstly, cluster A finds the variance of its points. Recall that the variance is also known as the spread. The less spread, more compact the cluster is.

Here is the variance formula for cluster A:

$$Var(A_i) = \frac{1}{n-1} \sum^n_{i=1} (A_i - \bar A)^2$$

Where:

* $A_i$ is the ith observation in cluster A

* $n$ is the number of observations.

* $\bar A$ is the mean of observations in cluster A.

Repeat for cluster B,C, and so on.

Secondly, Ward Linkage tries to find which cluster merges gives the smallest increase in variance. If for some reason cluster G and cluster E results in the smallest increase in variance, then we merge them and make cluster GE.

Then we repeat until we get our desired number of clusters. That's it!

GeeksForGeeks reasonably states "This method creates compact, well-separated clusters by making sure that data within each cluster is as similar as possible" (“Types of Linkages”).

---
### What Linkage will I use?

I will be using the Ward Linkage.

Firstly, Ward Linkage creates compact clusters.

Additionally, we only do variance calculations for each cluster, and then do variance calculations for merged clusters.

We should note that Average Linkage also makes compact clusters, but there are a lot of pairwise calculations that go on. It appears computationally expensive, which matters when we are working in big data sets.  

---
### Silhouette scores

I did mention that choosing the number of clusters is subjective. And this is especially true when we make a dendrogram.


But a more objective way to select the number of clusters is by using silhouette scores.


Just like in Spectral Clustering, we want to pick the highest silhouette score.


```python
# lets first make silhouette scores
# lets determine which cluster has the best silhoutte score!

from sklearn.cluster import AgglomerativeClustering
from sklearn.metrics import silhouette_score

hac_df = pca_df[["PC1","PC2"]]



# create a empty list of silhoutte scores:

sil_score_list = []

for k in range(2,12):
  hac_model = AgglomerativeClustering(n_clusters = k, metric = "euclidean", linkage = "ward")
  cluster_labels = hac_model.fit_predict(hac_df)
  sil_score = silhouette_score(hac_df, cluster_labels)
  sil_score_list.append(sil_score)


# Make a plot of these silhoutte scores:

fig = plt.figure(figsize = (15,8))

plt.scatter(range(2,12), sil_score_list,
            s = 200, color = "orange")

plt.plot(range(2,12), sil_score_list,
            linewidth = 3, color = "red",
         marker = "o")

plt.title("What is the optimal number of clusters?")
plt.xlabel("The number of clusters")
plt.ylabel("Silhouette score");


```

Since 2 clusters has the highest silhouette scores, lets pick that!

```python


hac_model = AgglomerativeClustering(n_clusters = 2, # number of clusters
                                    metric = "euclidean", # distance formula
                                    linkage = "ward" # I will use ward linkage
                                    )

hac_labels_2d = hac_model.fit_predict(hac_df)
print(hac_labels_2d)

pca_df["hac_labels_2d"] = hac_labels_2d
```

### 2D HAC plot

```python


fig = plt.figure(figsize = (15,8))

sns.scatterplot(pca_df, x = "PC1", y = "PC2", hue = "hac_labels_2d",
                s = 100);

plt.title("2D HAC plot using PCA");

```

It appears HAC clusters weak and strong Pokemon. Its as if a Pokemon hits a threshold of having more than 500 total base stats, they are clustered as strong. If they are less than 500, they are clustered as weak.

We will be using the scipy module to make our dendrogram.

```python

fig = plt.figure(figsize = (15,8))


# the labels should equal the original mythical_df.index
# the scaled version: mythical_df_scaled, is a numpy array so it has no index

dendrogram(linkage_matrix, labels = mythical_df.index,
           leaf_rotation = 90, leaf_font_size = 8)


plt.tight_layout()

plt.title("Dendrogram for HAC, Ward Linkage")
plt.xlabel("The Mythical Pokemon")
plt.ylabel("The Euclidean Distance");
```

Interesting, it looks like  Manaphy, Jirachi, Mew, and Celebi has no leaves!

Victini also appears to have no leaves as well!

But in closer inspection. there is a actually a line in these 5 Mythical Pokemon.




One possible reason for this phenomenon that all 4 Pokemon have pairwise distances of zero. Look at their stats below:

```python
mythical_df.loc[["Mew","Manaphy","Jirachi","Celebi","Victini"]]
```

Since their euclidean distances are 0, HAC thinks they have perfect similarities and there, they do not need leaves. HAC may be thinking "their distances are 0, so we do not have any differences to show. Might as well do just draw a straight line to these 4 Pokemon!"

Lets compare these 4 Pokemon to Shaymin and Diance, the Pokemon next to them.



```python
mythical_df.loc[["Shaymin","Victini","Mew","Manaphy","Jirachi","Celebi","Diancie"]]
```

So it actually appears that Shaymin has the same stats as well. And HAC simply decided to give Shaymin a leaf.

This looks much better for inteprataion!

Now lets talk about how we read a dendrogram.

This may look like a decision tree, but it is NOT a decision tree! Be careful! A dendrogram does NOT give the same interpretations as a decision tree either!

Any Pokemon near each other are similar. If their shared branch is low, then the more similar they are. If their shared branch is higher, then they are less similar. This is because the y-axis is the euclidean distance. The more distance, the less simialar.

When someone looks a dendrogram, they can decide to make use a cut-off line. Here is a example below:


```python

fig = plt.figure(figsize = (15,8))


dendrogram(linkage_matrix, labels = mythical_df.index,
           leaf_rotation = 90, leaf_font_size = 8)

plt.axhline(y = 5, color = "hotpink", linewidth = 2, linestyle = "dashed")


plt.tight_layout()

plt.title("Dendrogram for HAC, Ward Linkage")
plt.xlabel("The Mythical Pokemon")
plt.ylabel("The Euclidean Distance");
```

## 🔵 Un-Supervised Model 10: Isolation Forests


Un-supervised learning models do not always have to be about obvious visual patterns or clusters. Un-supervised learning can also be about finding outliers/anomaly! Interestingly, outliers can be considered as extreme patterns.

This is what Isolation Forests do, this machine learning model tries to find outliers.

Isolation Forests simply use the original predictor variables when finding outliers, they do NOT use a response variable! That is why Isolation Forests are a Un-supervised learning model!


So whey even bother with Isolation Forests? Well their ability to detect outliers can help us see any unusual activity or abnormal results. Isolation Forests are commonly used in cybersecurity, finance, healthcare, retail, and even manufacturing ("What is a Isolation Forests?").



In healthcare, the GeeksForGeeks Isolation Forest article state that "Isolation Forests are used to detect anomalous patient data by identifying unusual patterns in health records such as abnormal lab results or unexpected combinations of diagnoses and treatments. It enables early detection of fraud, errors or emerging health issues which improves patient care" ("What is a Isolation Forest?").




---
### Isolation Forest Intuition:


The main idea of Isolation Forest is that outliers just need a few cuts/parititions to be seperated from the data (McLeod). If a point only needs 1 cut to be seperated from the data, it is most likely a outlier.

Meanwhile, a non-outlier point needs many cuts to be seperated from the data. This suggests that this point is next to many other points.   


---

### Isolation Forest Process:

**Part 1- Subsampling**

Create the first iso tree. Usually, iso trees do not use the whole data set, it only uses a random sample of the data set. According to the Isolation Forest Scikit Learn article, a iso tree is defaulted to be trained on 256 data points (Pedregosa).

Do be careful, if our data set has 256 or less data points, the iso tree will use the whole data set. So we may have to change the Isolation Forest parameters for smaller data sets.


**Part 2- Random Partitioning**

A iso tree make random partitions. This iso tree splits the data at a random value within that predictor range and divides the data into two parts ("What is a Isolation Forest?"). There are 4 steps involved for a iso tree's random partitioning.

1. For example, if the predictor variable of attack had a range of 45 to 300, the split can only be between 45 to 300.

Lets say the split is 250. Perhaps our dataset has: [120,50,65,70,210,145,265]

* The first node split is $attack < 250$.

* The left child will contain:  [120,50,65,70,210,145]. The left child has 5 points, these points are not isolated, so we need to split more.

* The right child will contain: [265]. The right child has 1 point. This becomes a leaf. This Pokemon with a attack of 265 has a path length = 1 for this iso tree. (The path length will be important if a observation is a outlier!)

2. Next the left child will continue splitting. Perhaps it will use a predictor variable of defense. For defense, repeat the split concept from step 1. (Keep in mind that the 5 remaining Pokemon will probably have different numbers for defense.)

3. Keep splitting until all leaves have only 1 point OR if the max depth of the tree is reached.

4. Repeat for iso tree 2 to n.

**Part 3- Anomaly Score**  

After using all the iso trees, each data point is given a anomaly score.

A anomaly score is calculated by averaging the path lengths from each iso trees of a data point.

If a point has a small average path length, it will assigned as a outlier. If a point has a big average path length, it will not be a outlier.



BE CAREFUL! Remember that in Random Forests, each tree tries to predict a value such as a NBA player being a point guard or center. Then all trees use majority vote to predict what position the NBA player is!

In Isolation Forests, each iso tree does **NOT** predict a value! Isolation Forests are trying to calculate with path lengths!
Each iso tree has a path length for each point. All iso trees then take the average of each point's path length.

Recall that a path length is equal to the number of tree splits it took to isolate the point.


**Part 4- Interpreting Results**

The predict() function will give two labels:

* -1 is the label for outlier.

* 1 is the label for non outlier.

The decision_function() function will give a rank each point on the range of hugely normal or hugely outlier.

* The range is -0.5 to 0.5.

* Points that are negative are most likely outliers. The more negative the point is, it implies it is a strong outlier.

* Positive points are most likely not outliers. The more positive the point is, it implies it is near other points.






```python
from sklearn.ensemble import IsolationForest # step 1. fetch the model

# parameters defined:
# 1. n_estimators = 100 -> use 100 iso trees, this is the default number too
# 2. contamination = 0.05 -> The percentage of points that we want to assign as outliers
# In this case, 0.05 is 5%
# 3. random_state = 1 -> the seed number, makes the results reproducible

# step 2. initate the model
IF_model = IsolationForest(n_estimators = 100, contamination = 0.05,
                           random_state = 1)

# step 3. fit the model
IF_model.fit(X)

# step 4. predict or transform points using the model,
# REMEMBER, un-supervised learning has no response vector!

# binary label of -1 (outlier) or 1 (non outlier)
preds = IF_model.predict(X)


# continous label of -0.5 ot 0.5
anomaly_scores = IF_model.decision_function(X)


print("These are the first 20 predictions:", preds[0:20])
print("These are the last 20 predictions:", preds[1005:1025])

print("These are the first 20 anomaly scores:", np.round(anomaly_scores[0:20],2))
print("These are the last 20 anomaly scores:", np.round(anomaly_scores[1005:1025],2))

```

Interesting, no outliers in the first 20 or last 20 observations. So where are they located?

```python
# lets do some organizing and renaming

IF_df = pd.DataFrame({"Predictions": preds, "Anomaly Scores": anomaly_scores})

# Lets assign -1 as outlier, 1 as non-outlier

IF_df["Predictions"]
```

```python
outlier_df = IF_df[IF_df["Predictions"] == -1]

#outlier_df

print(outlier_df.index) # these are the observations/pokemon with the outliers!

print()

print(len(outlier_df.index)) # we have 52 outliers
```

### Conclusion for Isolation Forests:

* It appears that Isolation Forests can only be powerful depending on the predictors we feed it. I only did individual stats, and it seemed to classify any small or big individual stat as a outlier.

* If we used any stat related predictor, extremely weak Pokemon such as Magikarp or extremely strong Pokemon such as Arceus would most likely consistently be outliers.

For instance, if a Pokemon's HP is less than 30 or greater than 150, it will most likely be a outlier. If not, Isolation Forests will check the other individual stats and see if any of them are extreme.   

* But if we used a non related predictor, maybe we may not have Magikarp and Arceus as outliers. Isolation Forest could give different Pokemon as outliers.

Lets say we make a Pokemon's ability as a predictor. Many legendary Pokemon have the ability called *Pressure* or many fire Pokemon have the ability called *Blaze*, so those Pokemon may not be outliers. But Ditto, a copycat Pokemon who transforms into a copy of the opponent Pokemon, has a ability called **Imposter**. Only Ditto has the **Imposter** ability, no other Pokemon has that. As a result, Ditto would most likely be a outlier.

(Isolation Forests usually does not use strings for their predictors, so we would probably do one hot encoding or label encoding for the Pokemon ability predictor.)
